### B. Un límite del diseño que conviene decidir antes de interpretar la red

El *backbone* SDSM necesita suficientes artefactos para que un par de actores pueda superar lo esperado.
Con **k = 8 tópicos** y actores ligados a 2–4 tópicos, el p-valor más pequeño posible ronda 0,12, así que
**la red temática no puede tener lazos con α = 0,05** aunque haya estructura (en simulación aparecen lazos
desde unos 12 artefactos). El Paso 10 imprime este diagnóstico. Opciones: usar la relación normativa
(subdimensiones) como principal; construir la temática con tópicos + subdimensiones juntos; o definir lazos
temáticos por similitud de perfiles con un nulo propio. Es una decisión de método, no de código.

### C. Qué aporta la literatura adjunta

* **Lazega y Snijders (2016), cap. 6 (Wang, Robins y Matous):** el ERGM multinivel modela la red meso con
  las redes dentro de nivel como dadas, igual que el método. Su efecto *C4AXB* (alineamiento entre niveles)
  es el análogo formal de la "cascada": vale la pena citarlo.
* **Cap. 7 (Zhu et al.)** respalda el análisis de correspondencias del espacio conjunto; **cap. 8 (Žiberna y
  Lazega)** recomienda analizar cada nivel por separado antes del análisis conjunto.
* **Rawlings et al. (2023, cap. 11):** las proyecciones bipartitas inflan el cierre (transitividad) y
  distorsionan la intermediación. Respalda la limitación 8; el cuaderno compara la transitividad con un nulo
  y no usa intermediación.
* **Estrada y Knight (2015):** la centralidad de vector propio se apoya en Perron-Frobenius (red conexa); en
  redes desconectadas se concentra en un componente. Se reporta el componente de cada actor.
* **Humphrey y LeBreton (2019, *Handbook of multilevel theory…*; el archivo se llama "Hox") y Finch, Bolin y
  Kelley:** la no independencia de observaciones anidadas infla el error de tipo I. Las universidades están
  anidadas en países y el bloc es un atributo del país, así que el Paso 11 añade ICC(1) por país y una
  permutación entre países. **Courgeau (2007):** falacias ecológica y atomista; por eso importa que la
  robustez R3 cambie la unidad (actor ↔ documento) y que no se infieran conclusiones de un nivel a otro.
* **Cherven (2015):** el Paso 10 exporta las redes en `.gexf` para Gephi.

```python
# ---- 1. Rutas -------------------------------------------------------------------
import os, re, json, pickle, warnings, math, itertools, unicodedata, urllib.request
from pathlib import Path
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib
if not EN_COLAB:
    matplotlib.use("Agg")
import matplotlib.pyplot as plt
plt.rcParams.update({"font.size": 9, "axes.spines.top": False, "axes.spines.right": False,
                     "axes.grid": True, "grid.color": "#e1e0d9", "grid.linewidth": 0.6, "axes.axisbelow": True,
                     "axes.edgecolor": "#c3c2b7", "axes.prop_cycle": matplotlib.cycler(color=["#2a78d6", "#eb6834", "#1baf7a", "#6250d6", "#eda100"])})
# Paleta validada para daltonismo (niveles: G azul, P naranja, U verde agua) y mapas de color secuencial y divergente
from matplotlib.colors import LinearSegmentedColormap
COLOR_NIVEL = {"G": "#2a78d6", "P": "#eb6834", "U": "#1baf7a"}
COLOR_GOB = {"Global (organizations)": "#2a78d6", "National (countries)": "#eb6834", "Institutional (universities)": "#1baf7a"}
for _cm in (LinearSegmentedColormap.from_list("seguile_azul", ["#f3f8fe", "#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#2a78d6", "#1c5cab", "#104281", "#0d366b"]),
            LinearSegmentedColormap.from_list("seguile_div", ["#104281", "#3987e5", "#9ec5f4", "#f0efec", "#f3b1b0", "#e34948", "#a32727"])):
    try:
        matplotlib.colormaps.register(_cm, force=True)
    except Exception:
        pass

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", message="This pattern is interpreted as a regular expression")
warnings.filterwarnings("ignore", message=".*keyword-only.*")
pd.set_option("display.max_columns", 50); pd.set_option("display.width", 200)

DIR_PROYECTO = Path(os.environ.get("SEGUILE_DIR", f"/content/{REPO_GH}" if EN_COLAB else os.getcwd()))
DIR_EXCEL = DIR_PROYECTO            # base_g.xlsx, base_p.xlsx, base_u.xlsx
DIR_TXT = DIR_PROYECTO              # all_g/, all_p/, all_u/ (.txt traducidos)
if EN_COLAB and GUARDAR_EN_DRIVE:
    DIR_SALIDA = Path("/content/drive/MyDrive") / f"{REPO_GH}_resultados"
else:
    DIR_SALIDA = Path(os.environ.get("SEGUILE_SALIDA", DIR_PROYECTO / "resultados"))
DIR_TAB, DIR_FIG, DIR_OBJ = DIR_SALIDA / "tablas", DIR_SALIDA / "figuras", DIR_SALIDA / "objetos"
DIR_RECURSOS = DIR_SALIDA / "recursos"
for d in (DIR_TAB, DIR_FIG, DIR_OBJ, DIR_RECURSOS):
    d.mkdir(parents=True, exist_ok=True)

BASES = pd.DataFrame({"gob_letra": ["G", "P", "U"],
                      "archivo_excel": ["base_g.xlsx", "base_p.xlsx", "base_u.xlsx"],
                      "carpeta_txt": ["all_g", "all_p", "all_u"]})
ETIQ_GOB = {"G": "Global (organizations)", "P": "National (countries)", "U": "Institutional (universities)"}

# ---- 2. Parámetros (las [D#] son decisiones de la investigadora) --------------------
SEMILLA = 1234
# Corpus (Paso 2; método 3.5)
N_REGISTROS_ESPERADOS = 527
SOLO_DOCUMENTOS_POLITICA = False      # True: aplica las exclusiones de cribado_manual.csv
REGEX_SUBNACIONAL = r"sub-?national"  # ADAPTAR a los valores de la columna SCOPE
# Nota de traducción al inicio de cada .txt (método 3.6). Revisa con la celda de inspección del Paso 2.
NOTA_TRADUCCION_REGEX = r"(?is)\A\s*\[?[^\n]{0,400}?(original language|translat)[^\n]*\n"
MIN_PALABRAS_CRIBADO = 300            # método 3.5: longitud mínima
PERCENTIL_NORMATIVO_CRIBADO = 0.05    # método 3.5: 5 % inferior de vocabulario normativo
DETENER_SI_FALTAN_TXT = False         # R se detiene; aquí se avisa y se sigue
# Preprocesamiento (Paso 3; método 3.8)
USAR_LEMATIZACION = True
MIN_DOCS_TERMINO = 3
MAX_PROP_TERMINO = 0.90
UMBRAL_CASI_DUP = 0.95
MIN_CARACTERES = 500
STOP_DOMINIO = ["ai", "artificial", "intelligence", "page", "pp", "http", "https", "www", "com",
                "org", "edu", "gov", "pdf", "etc", "figure", "table", "annex", "section", "chapter"]
# LDA (Paso 4; método 3.9)
EJECUTAR_CV_K = True                  # pasada 1: True; pasada 2: False
K_GRID = list(range(4, 21, 2))
N_FOLDS = 5
K_FINAL = 8                           # [D1]
ETIQUETAS_TOPICOS = None              # [D3] lista de K_FINAL nombres, p. ej. ["Ethics", "Education", ...]
LDA_MAX_ITER = 200
# RQ1a (Paso 5; método 3.10-3.11)
UMBRAL_MULTI = 0.20
K_PERFILES = 5                        # [D4]
# RQ1b (Paso 6; método 3.13)
MIN_DEONTICAS = 5
POR_MIL = 1000                        # densidades deónticas por 1.000 palabras (método 3.13)
ESCALA_OVERRIDES = "metodo"           # "metodo" (0-3) o "R_v1" (0-4, escala de los scripts en R)
# RQ1c (Paso 7; método 3.14)
ANIO_CORTE_GENAI = 2023
# Taxonomía (Paso 8; método 3.15)
POR_DIEZ_MIL = 10000                  # densidad de la taxonomía por 10.000 palabras
RHO_MIN_TRIANG = 0.20
# Referencias (Paso 9)
EXCLUIR_ANACRONICAS = True
N_MUESTRA_CITAS = 100
UMBRAL_PRECISION = 0.80
# Red (Paso 10; método 3.17)
ALFA_BACKBONE = 0.05
MTC_BACKBONE = "none"                 # "none" o "holm"
VAR_LAZO_NORMATIVO = "DENSIDAD"       # método 3.17: densidad > mediana. Los scripts R usaban "CUOTA"
N_NULO_BLOQUES = 1000                 # método 3.17: redes aleatorias que conservan lazos por bloque
N_NULO_BIPARTITO = 500                # scripts R: nulo bipartito curveball
K_POSICIONES = {"G": 3, "P": 4, "U": 4}   # [D5]
ESTIMAR_ERGM = False                  # True: estima los ERGM en R dentro de Colab (Paso 10b)
# Inferencia (método 3.18)
N_PERMUT = 1000
B_MONTECARLO = 5000

rng = np.random.default_rng(SEMILLA)

# ---- 3. Clasificación Norte / Sur (método 3.7; UNCTADstat) ----------------------------
ISO_NORTH = ["AD","AL","AT","BA","BE","BG","BY","CH","CY","CZ","DE","DK","EE","ES","FI","FR",
             "GB","GR","HR","HU","IE","IS","IT","LI","LT","LU","LV","MC","MD","ME","MK","MT",
             "NL","NO","PL","PT","RO","RS","RU","SE","SI","SK","SM","UA",
             "US","CA","AU","NZ","IL","JP","KR"]
ISO_SUR_A_NORTE = ["SG", "HK", "TW", "QA", "AE", "SA", "OM", "PR"]   # solo en BLOC_SENS
ISO_NORTE_A_SUR = ["RS"]                                             # solo en BLOC_SENS
ISO_ALIAS = {"UK": "GB", "EL": "GR"}

# patrón, BLOC principal, BLOC_SENS (alternativa: OECD -> North, OEI -> Mixed)
ORG_BLOC = [
    (r"^(European Union|European Commission|European Parliament|EU\b)", "North", "North"),
    (r"^Council of Europe", "North", "North"),
    (r"^G7", "North", "North"),
    (r"^NIST", "North", "North"),
    (r"^(African Union|AU\b)", "South", "South"),
    (r"^(Arab League|League of Arab States)", "South", "South"),
    (r"^(ECLAC|CEPAL)", "South", "South"),
    (r"^CLAD", "South", "South"),
    (r"^(IDB|Inter-American Development Bank)", "South", "South"),
    (r"^OEI", "South", "Mixed"),
    (r"^ASEAN", "South", "South"),
    (r"^(OAS|Organization of American States)", "Mixed", "Mixed"),
    (r"^APEC", "Mixed", "Mixed"),
    (r"^(OECD|Organisation for Economic)", "Global", "North"),
]
# Categoría institucional de las organizaciones (método 3.4, Tabla 2). Covariable del ERGM.
ORG_CATEGORIA = [
    (r"^(UNESCO|United Nations|UN\b|UNDP|ILO|ITU|WHO|UNICEF)", "UN"),
    (r"^(OECD|Organisation for Economic|Council of Europe)", "IGO"),
    (r"^(G7|GPAI|AI Safety Summit)", "MULT"),
    (r"^(ISO|IEC|IEEE)", "GLOB"),
    (r"^(World Bank|IDB|Inter-American Development Bank)", "FIN"),
    (r"^(European Union|European Commission|African Union|ASEAN|APEC|Arab League|League of Arab States|OAS|Organization of American States|CLAD|OEI|ECLAC|CEPAL)", "REG"),
]
REGIONES = {"0": "Global", "1": "Africa", "2": "Arab World", "3": "Asia", "4": "Europe",
            "5": "Latin America", "6": "North America", "7": "Oceania"}


def canonizar_org(x):
    """Nombres canónicos de organizaciones (Step1b) + tabla opcional org_canonica.csv (ORIGINAL, CANONICO)."""
    s = pd.Series(x, dtype="object").map(lambda v: re.sub(r"\s+", " ", str(v)).strip() if pd.notna(v) else np.nan)
    reglas = [(r"^(CEPAL|ECLAC)", "ECLAC"), (r"^OAS\b", "OAS"), (r"^AI Safety Summits?", "AI Safety Summits"),
              (r"^(UICEF|UNICEF)", "UNICEF")]
    for pat, can in reglas:
        s = s.where(~s.fillna("").str.contains(pat, regex=True), can)
    f = DIR_PROYECTO / "org_canonica.csv"
    if f.exists():
        eq = pd.read_csv(f, dtype=str)
        s = s.map(lambda v: dict(zip(eq.ORIGINAL, eq.CANONICO)).get(v, v))
    return s


def clasificar_org(org, columna):
    """BLOC de una organización según ORG_BLOC (columna 1 = principal, 2 = sensibilidad)."""
    org = pd.Series(org, dtype="object").fillna("")
    salida = pd.Series("Global", index=org.index)
    for fila in ORG_BLOC:
        salida[org.str.contains(fila[0], case=False, regex=True)] = fila[columna]
    return salida


def categoria_org(org):
    org = pd.Series(org, dtype="object").fillna("")
    salida = pd.Series("Other", index=org.index)
    for pat, cat in ORG_CATEGORIA:
        salida[org.str.contains(pat, case=False, regex=True) & (salida == "Other")] = cat
    return salida


# ---- 4. Registro de resultados ------------------------------------------------------
TAB, FIG = {}, {}
CALIDAD = pd.DataFrame(columns=["paso", "indicador", "valor"])


def guardar_tabla(df, nombre):
    df = df.copy()
    df.to_csv(DIR_TAB / f"{nombre}.csv", index=False)
    TAB[nombre] = df
    return df


def guardar_fig(fig, nombre, ancho=9, alto=6):
    fig.set_size_inches(ancho, alto)
    ruta = DIR_FIG / f"{nombre}.png"
    fig.savefig(ruta, dpi=300, bbox_inches="tight")
    FIG[nombre] = str(ruta)
    plt.close(fig)
    return ruta


def guardar_obj(x, nombre):
    with open(DIR_OBJ / f"{nombre}.pkl", "wb") as f:
        pickle.dump(x, f)


def existe_obj(nombre):
    return (DIR_OBJ / f"{nombre}.pkl").exists()


def leer_obj(nombre):
    f = DIR_OBJ / f"{nombre}.pkl"
    if not f.exists():
        raise FileNotFoundError(f"Falta {f}. Ejecute antes el paso que lo genera.")
    with open(f, "rb") as fh:
        return pickle.load(fh)


def registrar(paso, indicador, valor):
    global CALIDAD
    CALIDAD = pd.concat([CALIDAD, pd.DataFrame([{"paso": paso, "indicador": indicador, "valor": str(valor)}])],
                        ignore_index=True)
    CALIDAD.to_csv(DIR_TAB / "t00_reporte_calidad.csv", index=False)


if (DIR_TAB / "t00_reporte_calidad.csv").exists():   # conserva el registro entre sesiones
    CALIDAD = pd.read_csv(DIR_TAB / "t00_reporte_calidad.csv", dtype=str)

# ---- 5. Funciones estadísticas (método 3.18) -------------------------------------------
def hellinger(p, q):
    p, q = np.asarray(p, float), np.asarray(q, float)
    return float(np.sqrt(np.sum((np.sqrt(p) - np.sqrt(q)) ** 2)) / np.sqrt(2))


def entropia_norm(g):
    g = np.asarray(g, float)
    return float(-np.sum(g * np.log(np.maximum(g, 1e-12))) / np.log(len(g)))


def p_bh(p):
    """Benjamini-Hochberg con NA (como p.adjust(..., 'BH') en R)."""
    p = pd.Series(p, dtype=float)
    ok = p.notna()
    out = pd.Series(np.nan, index=p.index)
    if ok.sum():
        v = p[ok].values; n = len(v); orden = np.argsort(v)[::-1]
        aj = np.minimum.accumulate(v[orden] * n / np.arange(n, 0, -1))
        res = np.empty(n); res[orden] = np.minimum(aj, 1)
        out[ok] = res
    return out.values


def holm(p):
    p = np.asarray(p, float); n = len(p); orden = np.argsort(p)
    aj = np.maximum.accumulate((n - np.arange(n)) * p[orden])
    res = np.empty(n); res[orden] = np.minimum(aj, 1)
    return res


def comparar_2g(x, grupo, g1="North", g2="South", n_min=3):
    """Wilcoxon de rangos + correlación biserial de rangos (Kerby, 2014). _1 = g1, _2 = g2; r_rb > 0: mayor en g1."""
    x = pd.Series(np.asarray(x, float)); grupo = pd.Series(np.asarray(grupo, dtype=object))
    a = x[(grupo == g1).values & x.notna().values].values
    b = x[(grupo == g2).values & x.notna().values].values
    r = {"n_1": len(a), "n_2": len(b),
         "mediana_1": np.median(a) if len(a) else np.nan, "mediana_2": np.median(b) if len(b) else np.nan,
         "media_1": a.mean() if len(a) else np.nan, "media_2": b.mean() if len(b) else np.nan,
         "r_rb": np.nan, "p_valor": np.nan}
    if len(a) >= n_min and len(b) >= n_min:
        if np.ptp(np.r_[a, b]) == 0:
            r["r_rb"], r["p_valor"] = 0.0, 1.0
        else:
            u = stats.mannwhitneyu(a, b, alternative="two-sided", method="asymptotic", use_continuity=True)
            r["r_rb"] = 2 * u.statistic / (len(a) * len(b)) - 1
            r["p_valor"] = u.pvalue
    return pd.Series(r)


def kw_niveles(x, nivel):
    """Kruskal-Wallis con épsilon² = H / (n - 1) (Tomczak y Tomczak, 2014)."""
    df = pd.DataFrame({"x": np.asarray(x, float), "g": np.asarray(nivel, dtype=object)}).dropna()
    grupos = [v.values for _, v in df.groupby("g")["x"]]
    if len(grupos) < 2 or np.ptp(df.x.values) == 0:
        return pd.Series({"n": len(df), "H": np.nan, "p_valor": np.nan, "epsilon2": np.nan})
    h = stats.kruskal(*grupos)
    return pd.Series({"n": len(df), "H": h.statistic, "p_valor": h.pvalue, "epsilon2": h.statistic / (len(df) - 1)})


def perm_test(x, grupo, g1="North", g2="South", n=None, bloques=None):
    """Permutación de etiquetas (diferencia de medias, bilateral). Con `bloques`, permuta las etiquetas
    entre conglomerados (p. ej., países) en vez de entre individuos. [REVISIÓN]"""
    n = n or N_PERMUT
    df = pd.DataFrame({"x": np.asarray(x, float), "g": np.asarray(grupo, dtype=object)})
    if bloques is not None:
        df["b"] = np.asarray(bloques, dtype=object)
    df = df[df.x.notna() & df.g.isin([g1, g2])]
    if (df.g == g1).sum() < 3 or (df.g == g2).sum() < 3:
        return np.nan
    xv = df.x.values; gv = (df.g.values == g1)
    est = lambda es1: xv[es1].mean() - xv[~es1].mean()
    obs = est(gv)
    nulo = np.empty(n)
    if bloques is None:
        for i in range(n):
            nulo[i] = est(rng.permutation(gv))
    else:
        cod, inv = np.unique(df.b.astype(str).values, return_inverse=True)
        etiq = pd.Series(gv).groupby(inv).first().values          # bloc de cada conglomerado
        if etiq.all() or (~etiq).all():
            return np.nan
        for i in range(n):
            es1 = rng.permutation(etiq)[inv]
            nulo[i] = est(es1) if es1.any() and (~es1).any() else np.nan
    nulo = nulo[~np.isnan(nulo)]
    return (1 + np.sum(np.abs(nulo) >= abs(obs) - 1e-12)) / (len(nulo) + 1)


def _tabla(a, b):
    t = pd.crosstab(pd.Series(np.asarray(a, dtype=object), name="a"), pd.Series(np.asarray(b, dtype=object), name="b"))
    return t.loc[t.sum(axis=1) > 0, t.sum(axis=0) > 0]


def _tablas_aleatorias(a, b, B):
    """Tablas con márgenes fijos (permutando una variable), para p-valores Monte Carlo."""
    a = np.asarray(a, dtype=object); b = np.asarray(b, dtype=object)
    ok = pd.notna(a) & pd.notna(b); a, b = a[ok], b[ok]
    ua, ia = np.unique(a.astype(str), return_inverse=True); ub, ib = np.unique(b.astype(str), return_inverse=True)
    for _ in range(B):
        t = np.zeros((len(ua), len(ub)))
        np.add.at(t, (ia, rng.permutation(ib)), 1)
        yield t


def chi_sim(a, b, B=None):
    """Chi-cuadrado con p-valor Monte Carlo + V de Cramér (como chisq.test(simulate.p.value = TRUE))."""
    B = B or B_MONTECARLO
    t = _tabla(a, b)
    if min(t.shape) < 2:
        return pd.Series({"n": int(t.values.sum()), "chi2": np.nan, "p_valor_sim": np.nan, "cramer_v": np.nan})
    chi = lambda m: stats.contingency.chi2_contingency(m, correction=False)[0]
    obs = chi(t.values)
    nulo = np.array([chi(m) for m in _tablas_aleatorias(a, b, B)])
    n = t.values.sum()
    return pd.Series({"n": int(n), "chi2": obs, "p_valor_sim": (1 + np.sum(nulo >= obs - 1e-9)) / (B + 1),
                      "cramer_v": np.sqrt(obs / (n * (min(t.shape) - 1)))})


def fisher_sim(a, b, B=None):
    """Prueba exacta de Fisher r x c con p-valor simulado (estadístico: probabilidad de la tabla) + V de Cramér."""
    B = B or B_MONTECARLO
    t = _tabla(a, b)
    if min(t.shape) < 2:
        return pd.Series({"n": int(t.values.sum()), "p_fisher_sim": np.nan, "cramer_v": np.nan})
    from scipy.special import gammaln
    logp = lambda m: -np.sum(gammaln(m + 1))      # el resto es constante con márgenes fijos
    obs = logp(t.values)
    nulo = np.array([logp(m) for m in _tablas_aleatorias(a, b, B)])
    chi = stats.contingency.chi2_contingency(t.values, correction=False)[0]
    n = t.values.sum()
    return pd.Series({"n": int(n), "p_fisher_sim": (1 + np.sum(nulo <= obs + 1e-7)) / (B + 1),
                      "cramer_v": np.sqrt(chi / (n * (min(t.shape) - 1)))})


def icc1(x, grupo):
    """ICC(1) por ANOVA de un factor (Bliese, 2000): no independencia por conglomerado. [REVISIÓN]"""
    df = pd.DataFrame({"x": np.asarray(x, float), "g": np.asarray(grupo, dtype=object)}).dropna()
    g = df.groupby("g")["x"]
    k = g.size(); k = k[k > 0]
    if len(k) < 2 or len(df) <= len(k):
        return np.nan
    N, J = len(df), len(k)
    msb = np.sum(k * (g.mean() - df.x.mean()) ** 2) / (J - 1)
    msw = np.sum((df.x - g.transform("mean")) ** 2) / (N - J)
    k0 = (N - np.sum(k ** 2) / N) / (J - 1)
    den = msb + (k0 - 1) * msw
    return float((msb - msw) / den) if den > 0 else np.nan


def distancia_referente(df, clave, valor, excluir_ref=False):
    """Hellinger de cada actor al perfil medio de las organizaciones de alcance mundial (BLOC == 'Global').
    Para esas organizaciones el referente se calcula sin ellas mismas (leave-one-out)."""
    clave = [clave] if isinstance(clave, str) else list(clave)
    es_ref = (df.GOV_CODE == "G") & (df.BLOC == "Global")
    ref = df[es_ref].groupby(clave)[valor].agg(s_ref="sum", n_ref="size").reset_index()
    d = df.merge(ref, on=clave, how="left")
    d["ES_REF"] = (d.GOV_CODE == "G") & (d.BLOC == "Global")
    d["v_ref"] = np.where(d.ES_REF, (d.s_ref - d[valor]) / (d.n_ref - 1), d.s_ref / d.n_ref)
    out = []
    for (gc, ac), g in d.groupby(["GOV_CODE", "ACTOR"]):
        p, q = g[valor].values, g.v_ref.values
        ok = np.isfinite(p) & np.isfinite(q)
        h = hellinger(p[ok] / p[ok].sum(), q[ok] / q[ok].sum()) if ok.any() and p[ok].sum() > 0 and q[ok].sum() > 0 else np.nan
        out.append({"GOV_CODE": gc, "ACTOR": ac, "HELLINGER": h, "ES_REF": bool(g.ES_REF.iloc[0])})
    out = pd.DataFrame(out)
    return out[~out.ES_REF].drop(columns="ES_REF") if excluir_ref else out.drop(columns="ES_REF")


def resumen_por(df, por, func):
    """Aplica func(grupo) -> Series a cada grupo y devuelve un DataFrame plano."""
    filas = []
    for k, g in df.groupby(por, dropna=False):
        k = k if isinstance(k, tuple) else (k,)
        filas.append({**dict(zip(por if isinstance(por, list) else [por], k)), **func(g).to_dict()})
    return pd.DataFrame(filas)


# ---- 6. Recursos léxicos de tidytext/textstem (réplica exacta de R) -------------------
URL_STOP = "https://raw.githubusercontent.com/juliasilge/tidytext/main/data/stop_words.rda"
URL_LEMAS = "https://raw.githubusercontent.com/trinker/lexicon/master/data/hash_lemmas.rda"


def _cargar_rda(url, nombre):
    destino = DIR_RECURSOS / f"{nombre}.csv"
    if destino.exists():
        return pd.read_csv(destino, dtype=str, keep_default_na=False)
    try:
        import pyreadr
        tmp = DIR_RECURSOS / f"{nombre}.rda"
        urllib.request.urlretrieve(url, tmp)
        df = list(pyreadr.read_r(str(tmp)).values())[0]
        df.to_csv(destino, index=False)
        return df.astype(str)
    except Exception as e:
        print(f"[AVISO] No se pudo descargar {nombre} ({e}). Se usa la alternativa de NLTK/sklearn.")
        return None


STOP_WORDS = _cargar_rda(URL_STOP, "stop_words")
if STOP_WORDS is None:   # alternativa aproximada
    from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
    STOP_WORDS = pd.DataFrame({"word": sorted(ENGLISH_STOP_WORDS), "lexicon": "sklearn"})
STOP_TODAS = set(STOP_WORDS.word)
STOP_SNOWBALL = set(STOP_WORDS.word[STOP_WORDS.lexicon == "snowball"]) or STOP_TODAS

LEMAS = _cargar_rda(URL_LEMAS, "hash_lemmas")
DIC_LEMAS = dict(zip(LEMAS.token, LEMAS.lemma)) if LEMAS is not None else None
if DIC_LEMAS is None:
    import nltk
    nltk.download("wordnet", quiet=True)
    from nltk.stem import WordNetLemmatizer
    _wnl = WordNetLemmatizer()


def lematizar(palabras):
    """Equivale a textstem::lemmatize_words (diccionario lexicon::hash_lemmas)."""
    if DIC_LEMAS is not None:
        return [DIC_LEMAS.get(w, w) for w in palabras]
    return [_wnl.lemmatize(w) for w in palabras]


# Léxico deóntico (método 3.13). Se usa también en el cribado del Paso 2.
LEX_DEONTICO = pd.DataFrame([
    ("must not", "fuerte"), ("shall not", "fuerte"), ("is not permitted", "fuerte"), ("are not permitted", "fuerte"),
    ("is prohibited", "fuerte"), ("are prohibited", "fuerte"), ("prohibited from", "fuerte"),
    ("is required to", "fuerte"), ("are required to", "fuerte"), ("is obliged to", "fuerte"),
    ("are obliged to", "fuerte"), ("mandatory", "fuerte"), ("must", "fuerte"), ("shall", "fuerte"),
    ("should not", "debil"), ("ought to", "debil"), ("is recommended", "debil"), ("are recommended", "debil"),
    ("is encouraged to", "debil"), ("are encouraged to", "debil"), ("is advisable", "debil"),
    ("best practices", "debil"), ("best practice", "debil"),            # [REVISIÓN] el método lo incluye
    ("recommends", "debil"), ("recommend", "debil"), ("encouraged", "debil"), ("should", "debil"),
], columns=["patron", "tipo"])
LEX_DEONTICO = LEX_DEONTICO.assign(n=LEX_DEONTICO.patron.str.len()).sort_values("n", ascending=False).drop(columns="n")
_RX_DEONT = [(re.compile(r"\b" + re.escape(p) + r"\b"), t) for p, t in LEX_DEONTICO.itertuples(index=False)]


def contar_deonticos(txt):
    """Cuenta obligaciones fuertes y débiles; las frases largas primero, sin contar dos veces."""
    s = re.sub(r"\s+", " ", str(txt).lower()).strip()
    out = {"fuerte": 0, "debil": 0}
    for rx, tipo in _RX_DEONT:
        s, n = rx.subn(" # ", s)
        out[tipo] += n
    return out



# ---- 7. Funciones de redes (Paso 10 y robustez R6) ---------------------------------------------
# Backbone SDSM (Neal, 2014): modelo de configuración bipartito (BiCM) + binomial de Poisson.
def bicm(B, iters=5000, tol=1e-10):
    """Probabilidades del modelo de configuración bipartito (BiCM) que reproducen grados de fila y columna."""
    r = B.sum(axis=1).astype(float); c = B.sum(axis=0).astype(float); n, m = B.shape
    llena = r >= m
    x = np.where(llena, 1.0, r / np.sqrt(B.sum() + 1e-9)); y = c / np.sqrt(B.sum() + 1e-9)
    for _ in range(iters):
        xy = np.outer(x, y)
        x_new = np.where(llena, x, r / np.maximum((y[None, :] / (1 + xy)).sum(axis=1), 1e-300))
        xy = np.outer(x_new, y)
        y_new = c / np.maximum((x_new[:, None] / (1 + xy)).sum(axis=0), 1e-300)
        if np.max(np.abs(x_new - x)) < tol and np.max(np.abs(y_new - y)) < tol:
            x, y = x_new, y_new; break
        x, y = x_new, y_new
    xy = np.outer(x, y); P = xy / (1 + xy)
    P[llena, :] = 1.0
    return np.clip(P, 1e-12, 1 - 1e-12)


def _cola_superior(Q):
    """Q: (pares, m) probabilidades de Bernoulli. Devuelve P(X >= c) para c = 0..m (binomial de Poisson)."""
    dist = np.zeros((Q.shape[0], Q.shape[1] + 1)); dist[:, 0] = 1
    for k in range(Q.shape[1]):
        q = Q[:, k:k + 1]
        dist[:, 1:] = dist[:, 1:] * (1 - q) + dist[:, :-1] * q
        dist[:, 0] *= (1 - q[:, 0])
    return np.flip(np.cumsum(np.flip(dist, 1), 1), 1)


def _preparar(B):
    B = np.asarray(B, float); filas = np.where(B.sum(axis=1) > 0)[0]
    Bx = B[filas]; cols = np.where((Bx.sum(axis=0) > 0) & (Bx.sum(axis=0) < Bx.shape[0]))[0]
    return B, filas, Bx[:, cols]


def sdsm_pvalores(B):
    """p-valores SDSM (cola superior) para cada par de actores con al menos un artefacto."""
    B, filas, Bx = _preparar(B)
    if len(filas) < 3 or Bx.shape[1] < 2:
        return None
    P = bicm(Bx); iu = np.triu_indices(len(filas), 1)
    cola = _cola_superior(P[iu[0]] * P[iu[1]])
    O = (Bx @ Bx.T)[iu].astype(int)
    return {"filas": filas, "iu": iu, "p": cola[np.arange(len(O)), O], "cola": cola, "cols_ok": Bx.shape[1]}


def extraer_backbone(B, alpha=None, mtc=None):
    alpha = ALFA_BACKBONE if alpha is None else alpha; mtc = MTC_BACKBONE if mtc is None else mtc
    B = np.asarray(B, float); A = np.zeros((B.shape[0], B.shape[0]), dtype=int)
    s = sdsm_pvalores(B)
    if s is None:
        return A
    p = holm(s["p"]) if mtc == "holm" else s["p"]
    sel = p < alpha
    i, j = s["filas"][s["iu"][0][sel]], s["filas"][s["iu"][1][sel]]
    A[i, j] = A[j, i] = 1
    return A


def umbral_sdsm(B, alpha=None):
    """Para el nulo curveball: con grados fijos, el umbral de co-ocurrencia de cada par no cambia."""
    alpha = ALFA_BACKBONE if alpha is None else alpha
    s = sdsm_pvalores(B)
    if s is None:
        return None
    c_min = np.where(s["cola"] <= alpha, np.arange(s["cola"].shape[1])[None, :], 10 ** 9).min(1)
    B, filas, Bx = _preparar(B)
    cols = np.where((B[filas].sum(axis=0) > 0) & (B[filas].sum(axis=0) < len(filas)))[0]
    return {"filas": filas, "iu": s["iu"], "c_min": c_min, "cols": cols}


def backbone_con_umbral(B, u):
    A = np.zeros((B.shape[0], B.shape[0]), dtype=int)
    Bx = np.asarray(B, float)[u["filas"]][:, u["cols"]]
    O = (Bx @ Bx.T)[u["iu"]]
    sel = O >= u["c_min"]
    i, j = u["filas"][u["iu"][0][sel]], u["filas"][u["iu"][1][sel]]
    A[i, j] = A[j, i] = 1
    return A


def curveball(B, n_pasos=None):
    """Aleatorización curveball: conserva totales de fila y columna (Strona et al., 2014)."""
    B = np.asarray(B, int); n_pasos = n_pasos or 5 * B.shape[0]
    filas = [set(np.where(B[i] == 1)[0]) for i in range(B.shape[0])]
    for _ in range(n_pasos):
        a, b = rng.choice(B.shape[0], 2, replace=False)
        comunes = filas[a] & filas[b]; excl = list((filas[a] | filas[b]) - comunes)
        if len(excl) < 2:
            continue
        na = len(filas[a]) - len(comunes); excl = list(rng.permutation(excl))
        filas[a] = comunes | set(excl[:na]); filas[b] = comunes | set(excl[na:])
    R = np.zeros_like(B)
    for i, f in enumerate(filas):
        R[i, list(f)] = 1
    return R


def contexto_red(nodos):
    """Índices de universidades con país en la red, su país y las organizaciones (para el censo)."""
    ids = nodos.ID.tolist(); pos = {i: k for k, i in enumerate(ids)}; nivel = nodos.NIVEL.to_numpy(dtype=object)
    pais = nodos.PAIS_ID.to_numpy(dtype=object)
    U_idx = np.where((nivel == "U") & pd.notna(pais))[0]
    return {"ids": ids, "U_idx": U_idx, "P_de_U": np.array([pos[p] for p in pais[U_idx]], dtype=int),
            "iG": np.where(nivel == "G")[0]}


CELDAS = {7: "cascada (u-g, p-g, u-p)", 3: "relevo_nacional (p-g, u-p, sin u-g)", 6: "paralelo (u-g, p-g, sin u-p)",
          4: "bypass_puro (solo u-g)", 5: "u-g y u-p, sin p-g", 2: "solo p-g", 1: "solo u-p", 0: "desacoplado"}


def censo_configuraciones(A, ctx):
    """Censo de las 8 combinaciones (u-g, p-g, u-p) para cada par universidad-organización (método 3.17)."""
    U, P, G = ctx["U_idx"], ctx["P_de_U"], ctx["iG"]
    if not len(U) or not len(G):
        return {}
    ug = A[np.ix_(U, G)]; pg = A[np.ix_(P, G)]; up = A[U, P][:, None] * np.ones((1, len(G)), int)
    cod = ug * 4 + pg * 2 + up
    out = {nom: float(np.mean(cod == c)) for c, nom in CELDAS.items()}
    out["BYPASS_metodo (u-g sin u-p)"] = float(np.mean((ug == 1) & (up == 0)))
    out["alin_U_P"] = float(np.mean(A[U, P]))
    out["lift"] = float(ug[pg == 1].mean() - ug[pg == 0].mean()) if (pg == 1).any() and (pg == 0).any() else np.nan
    return out


def clasificar_universidades(A, ctx):
    """Clasificación de cada universidad (método 3.17); la categoría '101' no la nombra el método."""
    U, P, G = ctx["U_idx"], ctx["P_de_U"], ctx["iG"]
    ug = A[np.ix_(U, G)]; pg = A[np.ix_(P, G)]; up = A[U, P]
    clase = []
    for k in range(len(U)):
        if np.any(ug[k] & pg[k]) and up[k]: clase.append("Cascade")
        elif np.any(ug[k] & pg[k]): clase.append("Parallel alignment")
        elif np.any(ug[k]) and not up[k]: clase.append("Bypass")
        elif np.any(ug[k]): clase.append("Direct and national, country not aligned")
        elif up[k]: clase.append("National alignment only")
        else: clase.append("Decoupled")
    return pd.Series(clase, index=[ctx["ids"][i] for i in U])


# ---- 8. Limpieza de artefactos de la extracción de PDF y diccionario de inglés ------------------
def limpiar_texto(t):
    """Ligaduras (ﬁ, ﬂ), marcas de salto de página y palabras cortadas con guion al final de línea."""
    t = unicodedata.normalize("NFKC", str(t)).replace("\u00ad", "")
    t = re.sub(r"-{2,}\s*PAGE BREAK\s*-{2,}", " ", t, flags=re.I)
    t = re.sub(r"(\w)-[ \t]*\r?\n[ \t]*(\w)", r"\1\2", t)
    return t


def diccionario_ingles():
    """Formas reales del inglés: diccionario de lemas de textstem + lista de palabras de NLTK (si se puede descargar)."""
    pal = set()
    if LEMAS is not None:
        pal |= set(LEMAS.token.str.lower()) | set(LEMAS.lemma.str.lower())
    try:
        import nltk
        nltk.download("words", quiet=True)
        from nltk.corpus import words
        pal |= {w.lower() for w in words.words()}
    except Exception as e:
        print(f"[AVISO] No se pudo cargar la lista de palabras de NLTK ({e}); se usa solo el diccionario de lemas.")
    return pal


print("Paso 1 cargado. Resultados en:", DIR_SALIDA)
print(f"Stopwords: {len(STOP_TODAS)} | lemas: {'hash_lemmas (textstem)' if DIC_LEMAS else 'WordNet (alternativa)'}")
```

```python
PASO = "Step2"
LECTURA_OK = True

# ---- 1. Lectura de las bases (solo se descartan filas totalmente vacías) ----------------
def limpiar_nombres(cols):
    out = []
    for c in cols:
        c = re.sub(r"[\r\n]", " ", str(c)); c = re.sub(r"\s+", " ", c).strip().upper()
        c = c.replace("#", "NUM"); c = re.sub(r"[^A-Z0-9]+", "_", c).strip("_")
        out.append(c)
    return out


def leer_base(archivo, letra):
    df = pd.read_excel(DIR_EXCEL / archivo, dtype=str)
    vacias = [c for c in df.columns if str(c).startswith("Unnamed")]
    df = df.rename(columns={c: f"NOTES{i+1}" for i, c in enumerate(vacias)})
    df.columns = limpiar_nombres(df.columns)
    df = df.rename(columns={"DCO_TYPE": "DOC_TYPE"})
    df = df.apply(lambda s: s.map(lambda v: re.sub(r"\s+", " ", v).strip() if isinstance(v, str) else v)).replace({"": np.nan})
    df["FILA_EXCEL"] = df.index + 2
    df = df[df.drop(columns="FILA_EXCEL").notna().any(axis=1)].copy()
    df["BASE_ORIGEN"] = letra
    return df


for c in BASES.carpeta_txt:
    if not (DIR_TXT / c).is_dir():
        raise FileNotFoundError(f"No encuentro la carpeta: {DIR_TXT / c}")
meta_raw = pd.concat([leer_base(r.archivo_excel, r.gob_letra) for r in BASES.itertuples()], ignore_index=True)
for col in ["ORGANIZATION", "UNIVERSITY", "COUNTRY", "REGION", "SCOPE", "LANGUAGE", "DOC_NATURE", "DOC_TITLE", "YEAR", "DOC_TYPE"]:
    if col not in meta_raw:
        meta_raw[col] = np.nan
print(meta_raw.groupby("BASE_ORIGEN").size().rename("registros_leidos").to_string())
if meta_raw.CODE.isna().any():
    meta_raw[meta_raw.CODE.isna()].to_csv(DIR_TAB / "t02_ERROR_registros_sin_CODE.csv", index=False)
    print("ATENCIÓN: hay filas sin CODE (ver t02_ERROR_registros_sin_CODE.csv)."); LECTURA_OK = False
if len(meta_raw) != N_REGISTROS_ESPERADOS:
    print(f"ATENCIÓN: se leyeron {len(meta_raw)} registros y se esperaban {N_REGISTROS_ESPERADOS}."); LECTURA_OK = False

# ---- 2. Homologación, actor y bloc (método 3.7) ---------------------------------------------
HOMOLOGACION = {
    "guidelines and guides": "Guidelines and Guides", "internal policies": "Internal Policies",
    "ethical framework / principles": "Ethical Framework / Principles",
    "national strategy": "Strategy", "strategy": "Strategy", "ai strategy": "Strategy",
    "recommendation": "Recommendations", "recommendations": "Recommendations",
    "report": "Report", "institutional report": "Report", "standards": "Standards",
    "discussion paper": "Discussion Paper", "regulation": "Regulation",
    "treaty / convention": "Treaty / Convention", "no found": None}


def extraer_iso(pais):
    """Como en R: el ISO2 entre paréntesis en COUNTRY, p. ej. 'Colombia (CO)'.
    [REVISIÓN] Si no lo hay, se intenta por el nombre del país con pycountry (queda en t02_iso2_por_nombre)."""
    if not isinstance(pais, str):
        return np.nan, False
    m = re.search(r"\(([A-Za-z]{2})\)", pais)
    if m:
        iso = m.group(1).upper(); return ISO_ALIAS.get(iso, iso), False
    try:
        import pycountry
        nombre = re.sub(r"\s*\(.*?\)", "", pais).strip()
        try:
            return pycountry.countries.lookup(nombre).alpha_2, True
        except LookupError:
            return pycountry.countries.search_fuzzy(nombre)[0].alpha_2, True
    except Exception:
        return np.nan, False


meta = meta_raw.copy()
meta["REG_ID"] = [f"R{i:03d}" for i in range(1, len(meta) + 1)]
meta["GOV_CODE"] = meta.CODE.str[0].str.upper()
meta["GOVERNANCE"] = meta.GOV_CODE.map(ETIQ_GOB)
canon = canonizar_org(meta.ORGANIZATION).values
meta["ACTOR"] = np.select([meta.GOV_CODE == "G", meta.GOV_CODE == "P", meta.GOV_CODE == "U"],
                          [canon, meta.COUNTRY, meta.UNIVERSITY], default=None)
meta["YEAR"] = pd.to_numeric(meta.YEAR, errors="coerce").astype("Int64")
meta["DOC_TYPE_ORIG"] = meta.DOC_TYPE
meta["DOC_TYPE_KEY"] = meta.DOC_TYPE.str.replace(r"\s*/\s*", " / ", regex=True).str.lower()
meta["DOC_TYPE_H"] = meta.DOC_TYPE_KEY.map(HOMOLOGACION)
meta["SIN_DOCUMENTO"] = meta.DOC_TYPE_KEY.isin(["no found", "not found"]) | \
    (meta.DOC_TITLE.fillna("").str.lower() == "no document")
iso = [extraer_iso(p) for p in meta.COUNTRY]
meta["ISO2"] = [i for i, _ in iso]; meta["ISO2_POR_NOMBRE"] = [b for _, b in iso]
meta.loc[meta.GOV_CODE == "G", "ISO2"] = np.nan
es_g = (meta.GOV_CODE == "G").values
bloc = np.where(meta.ISO2.isin(ISO_NORTH), "North", np.where(meta.ISO2.notna(), "South", None))
meta["BLOC"] = np.where(es_g, clasificar_org(meta.ACTOR, 1), bloc)
meta["BLOC_SENS"] = np.where(es_g, clasificar_org(meta.ACTOR, 2),
                             np.where(meta.ISO2.isin(ISO_SUR_A_NORTE), "North",
                                      np.where(meta.ISO2.isin(ISO_NORTE_A_SUR), "South", meta.BLOC)))
meta["CASO_LIMITE"] = (meta.BLOC != meta.BLOC_SENS) & meta.BLOC.notna()
meta["ORG_CATEGORIA"] = np.where(es_g, categoria_org(meta.ACTOR), None)
# Covariables del método (3.5) que el R no construía
meta["DOC_NATURE"] = meta.DOC_NATURE.fillna("Policy document")
meta["COMUNICACION"] = meta.DOC_NATURE.str.contains(r"communicat|news|announce|press", case=False, regex=True)
meta["SUBNACIONAL"] = (meta.GOV_CODE == "P") & meta.SCOPE.fillna("").str.contains(REGEX_SUBNACIONAL, case=False, regex=True)

no_mapeadas = meta[~meta.SIN_DOCUMENTO & meta.DOC_TYPE_H.isna()].DOC_TYPE_ORIG.unique()
if len(no_mapeadas):
    print("ATENCIÓN: tipologías sin homologar:", "; ".join(map(str, no_mapeadas))); LECTURA_OK = False
if meta.ACTOR.isna().any():
    print("Aviso: registros sin actor:", "; ".join(meta.REG_ID[meta.ACTOR.isna()]))
if pd.isna(meta.BLOC).any():
    print("Aviso: registros sin BLOC (revise ISO2 en COUNTRY).")
# [REVISIÓN] Un mismo país escrito de dos formas en COUNTRY crearía dos actores distintos
dobles = meta[meta.GOV_CODE == "P"].dropna(subset=["ISO2"]).groupby("ISO2").ACTOR.unique()
dobles = dobles[dobles.map(len) > 1]
if len(dobles):
    print("ATENCIÓN: países con más de un nombre en COUNTRY (se contarían como actores distintos):",
          "; ".join(f"{k}: {list(v)}" for k, v in dobles.items()))
guardar_tabla(meta.groupby(["BASE_ORIGEN", "DOC_TYPE_ORIG", "DOC_TYPE_H"], dropna=False).size().reset_index(name="n"),
              "t02_homologacion_tipologias")
guardar_tabla(meta[["GOVERNANCE", "ACTOR", "ORGANIZATION", "COUNTRY", "ISO2", "BLOC", "BLOC_SENS", "CASO_LIMITE"]]
              .drop_duplicates().sort_values(["GOVERNANCE", "BLOC", "ACTOR"], na_position="last"), "t02_clasificacion_bloc")
guardar_tabla(meta[meta.ISO2_POR_NOMBRE][["REG_ID", "CODE", "COUNTRY", "ISO2"]], "t02_iso2_por_nombre")

# ---- 3. Índice de archivos .txt ---------------------------------------------------------------
norm_cod = lambda x: re.sub(r"[\s_.\-]", "", str(x)).upper()
filas = []
for r in BASES.itertuples():
    for f in sorted((DIR_TXT / r.carpeta_txt).rglob("*")):
        if f.is_file() and f.suffix.lower() == ".txt":
            filas.append({"CARPETA": r.gob_letra, "RUTA": f, "ARCHIVO": re.sub(r"\s+", " ", f.stem).strip()})
indice_txt = pd.DataFrame(filas, columns=["CARPETA", "RUTA", "ARCHIVO"])
if indice_txt.ARCHIVO.duplicated().any():
    print("Aviso: hay .txt con el mismo nombre en varias carpetas; se usa el primero.")
indice_txt = indice_txt.drop_duplicates("ARCHIVO").assign(CLAVE=lambda d: d.ARCHIVO.map(norm_cod))
print(indice_txt.groupby("CARPETA").size().rename("archivos_txt").to_string())

# ---- 4. Vínculo registro -> .txt (manual > exacto > normalizado) -------------------------------
f_mapa = DIR_PROYECTO / "mapa_txt.csv"            # columnas: CODE, ARCHIVO (varios CODE pueden compartir ARCHIVO)


def leer_mapa(f):
    m = pd.read_csv(f, dtype=str, encoding="utf-8-sig")
    if not {"CODE", "ARCHIVO"} <= set(m.columns):        # CSV guardado con punto y coma
        m = pd.read_csv(f, dtype=str, sep=";", encoding="utf-8-sig")
    if not {"CODE", "ARCHIVO"} <= set(m.columns):
        print("ATENCIÓN: mapa_txt.csv no tiene columnas CODE y ARCHIVO; se ignora.")
        return pd.DataFrame(columns=["CODE", "ARCHIVO_MANUAL"])
    m["CODE"] = m.CODE.str.strip()
    m["ARCHIVO"] = m.ARCHIVO.fillna("").str.strip().str.replace(r"(?i)\.txt$", "", regex=True)
    return m[m.ARCHIVO != ""].drop_duplicates("CODE")[["CODE", "ARCHIVO"]].rename(columns={"ARCHIVO": "ARCHIVO_MANUAL"})


mapa = leer_mapa(f_mapa) if f_mapa.exists() else pd.DataFrame(columns=["CODE", "ARCHIVO_MANUAL"])
malos = sorted(set(mapa.ARCHIVO_MANUAL) - set(indice_txt.ARCHIVO))
if malos:
    print("ATENCIÓN: mapa_txt.csv apunta a archivos que no existen (se ignoran):", "; ".join(malos))
    mapa = mapa[~mapa.ARCHIVO_MANUAL.isin(malos)]
print("Vínculos manuales leídos de mapa_txt.csv:", len(mapa))

registros = meta.merge(mapa, on="CODE", how="left")
existentes = set(indice_txt.ARCHIVO); clave_a_arch = dict(zip(indice_txt.CLAVE, indice_txt.ARCHIVO))
registros["ARCH_EXACTO"] = registros.CODE.where(registros.CODE.isin(existentes))
registros["ARCH_NORM"] = registros.CODE.map(lambda c: clave_a_arch.get(norm_cod(c)))
registros["TEXTO"] = registros.ARCHIVO_MANUAL.fillna(registros.ARCH_EXACTO).fillna(registros.ARCH_NORM)
registros.loc[registros.SIN_DOCUMENTO, "TEXTO"] = np.nan
registros["VINCULO"] = np.select(
    [registros.SIN_DOCUMENTO, registros.ARCHIVO_MANUAL.notna(), registros.ARCH_EXACTO.notna(), registros.ARCH_NORM.notna()],
    ["Sin documento (ausencia)", "Manual (mapa_txt.csv)", "CODE = nombre de archivo", "Nombre normalizado"], default="SIN ARCHIVO")
registros = registros.drop(columns=["ARCHIVO_MANUAL", "ARCH_EXACTO", "ARCH_NORM"]).rename(columns={"REGION": "REGION"})
print(registros.groupby(["GOVERNANCE", "VINCULO"]).size().to_string())
guardar_obj(registros, "registros")

# ---- 5. Si faltan vínculos: plantilla prellenada para revisar (antes Step2b) ------------------
STEP2_COMPLETO = False
sin_archivo = registros[registros.VINCULO == "SIN ARCHIVO"]
if len(sin_archivo):
    def lev(a, b):
        prev = list(range(len(b) + 1))
        for i, ca in enumerate(a, 1):
            cur = [i]
            for j, cb in enumerate(b, 1):
                cur.append(min(prev[j] + 1, cur[j - 1] + 1, prev[j - 1] + (ca != cb)))
            prev = cur
        return prev[-1]
    norm_tit = lambda x: re.sub(r"\s+", " ", re.sub(r"[^\w ]", " ", str(x) if pd.notna(x) else "").lower()).strip()
    tit_ref = (registros[registros.TEXTO.notna()].assign(TIT=lambda d: d.DOC_TITLE.map(norm_tit))
               .drop_duplicates("TIT").set_index("TIT").TEXTO)
    rev = []
    for r in sin_archivo.itertuples():
        d = [lev(norm_cod(r.CODE), k) for k in indice_txt.CLAVE]
        k = int(np.argmin(d)) if d else None
        sug_t = tit_ref.get(norm_tit(r.DOC_TITLE))
        sug_n = indice_txt.ARCHIVO.iloc[k] if k is not None else None
        dist = d[k] if k is not None else np.nan
        archivo = sug_t if sug_t is not None else (sug_n if dist <= 1 else "")
        criterio = "titulo (confirmar)" if sug_t is not None else ("nombre (confirmar)" if dist <= 1 else "REVISAR A MANO")
        rev.append({"CODE": r.CODE, "ARCHIVO": archivo, "CRITERIO": criterio, "REG_ID": r.REG_ID, "GOVERNANCE": r.GOVERNANCE,
                    "ACTOR": r.ACTOR, "DOC_TITLE": r.DOC_TITLE, "SUGERENCIA_POR_TITULO": sug_t,
                    "SUGERENCIA_POR_NOMBRE": sug_n, "DISTANCIA_NOMBRE": dist})
    rev = pd.DataFrame(rev)
    previo = pd.read_csv(f_mapa, dtype=str, encoding="utf-8-sig") if f_mapa.exists() else None
    pd.concat([previo, rev.astype(str).replace("nan", "")]).to_csv(DIR_PROYECTO / "mapa_txt_REVISION.csv", index=False)
    indice_txt[~indice_txt.ARCHIVO.isin(registros.TEXTO)][["CARPETA", "ARCHIVO"]].to_csv(DIR_PROYECTO / "txt_sin_registro.csv", index=False)
    print(rev.groupby(["GOVERNANCE", "CRITERIO"]).size().to_string())
    print(f"\n>>> FALTAN {len(sin_archivo)} VÍNCULOS. El Paso 2 se pausa (no se eliminó ningún registro).\n"
          ">>> 1) Abre mapa_txt_REVISION.csv en Excel; confirma o completa la columna ARCHIVO.\n"
          ">>> 2) Guárdalo como mapa_txt.csv (CSV UTF-8, delimitado por comas) y súbelo al repositorio.\n"
          ">>> 3) En Colab: !git -C /content/SEGUILE pull, y vuelve a correr el Paso 2.")
    if EN_COLAB:
        print(f"(Los archivos quedaron en {DIR_PROYECTO}; descárgalos desde el panel de archivos de Colab.)")
elif LECTURA_OK:
    STEP2_COMPLETO = True
else:
    print(">>> Corrige los avisos marcados como ATENCIÓN y vuelve a correr el Paso 2.")

# ---- 6-9. Solo se ejecuta cuando todos los registros están vinculados ---------------------------
if STEP2_COMPLETO:
    # 6. Estado de actores (ausencia = ningún registro con documento)
    primero = lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan
    actores_estado = (registros.groupby(["GOV_CODE", "GOVERNANCE", "ACTOR"], dropna=False)
                      .agg(ISO2=("ISO2", primero), REGION=("REGION", primero), BLOC=("BLOC", "first"),
                           BLOC_SENS=("BLOC_SENS", "first"), ORG_CATEGORIA=("ORG_CATEGORIA", "first"),
                           n_registros=("REG_ID", "size"), n_con_doc=("SIN_DOCUMENTO", lambda s: int((~s).sum())),
                           ANIO_MEDIANO=("YEAR", lambda s: s.dropna().median() if s.notna().any() else np.nan))
                      .reset_index())
    actores_estado["AUSENCIA"] = actores_estado.n_con_doc == 0

    # 7. Atributos por texto y vínculo registro-texto (CODE = texto; REG_ID = registro)
    con_doc = registros[~registros.SIN_DOCUMENTO]
    docs_meta = con_doc.groupby("TEXTO").agg(
        GOV_CODE=("GOV_CODE", "first"), GOVERNANCE=("GOVERNANCE", "first"), DOC_TITLE=("DOC_TITLE", "first"),
        YEAR=("YEAR", "max"),                                     # año más reciente de los registros vinculados
        DOC_TYPE_H=("DOC_TYPE_H", lambda s: " | ".join(sorted(set(s.dropna())))),
        IDIOMA=("LANGUAGE", primero), N_REGISTROS=("REG_ID", "size"), N_ACTORES=("ACTOR", "nunique"),
        NIVELES=("GOV_CODE", lambda s: "+".join(sorted(set(s)))), BLOC_MIXTO=("BLOC", lambda s: s.nunique() > 1),
        SUBNACIONAL=("SUBNACIONAL", "any"), COMUNICACION=("COMUNICACION", "any")).reset_index().rename(columns={"TEXTO": "CODE"})
    doc_actor_reg = (con_doc[["REG_ID", "TEXTO", "CODE", "GOV_CODE", "GOVERNANCE", "ACTOR", "ISO2", "REGION", "BLOC",
                              "BLOC_SENS", "ORG_CATEGORIA"]]
                     .rename(columns={"CODE": "CODE_REGISTRO"}).rename(columns={"TEXTO": "CODE"})
                     .merge(docs_meta[["CODE", "N_REGISTROS"]].rename(columns={"N_REGISTROS": "N_REGISTROS_TEXTO"}), on="CODE"))
    doc_actor_reg["TEXTO_COMPARTIDO"] = doc_actor_reg.N_REGISTROS_TEXTO > 1      # adopción compartida por pertenencia
    guardar_tabla(docs_meta[docs_meta.N_REGISTROS > 1].sort_values("N_REGISTROS", ascending=False), "t02_textos_compartidos")
    guardar_tabla(doc_actor_reg.groupby(["GOVERNANCE", "TEXTO_COMPARTIDO"]).size().reset_index(name="n"),
                  "t02_adopcion_compartida_por_nivel")
    # Para los análisis por actor, cada texto cuenta una vez por actor (método 3.5)
    doc_actor = doc_actor_reg.drop_duplicates(["CODE", "GOV_CODE", "ACTOR"])

    # 8. Lectura de textos y marcas (no eliminan registros)
    rutas = dict(zip(indice_txt.ARCHIVO, indice_txt.RUTA))
    corpus = docs_meta.copy()
    corpus["text"] = corpus.CODE.map(lambda c: Path(rutas[c]).read_text(encoding="utf-8", errors="replace")).map(limpiar_texto)
    if NOTA_TRADUCCION_REGEX:
        rx = re.compile(NOTA_TRADUCCION_REGEX, flags=re.I)
        corpus["NOTA"] = corpus.text.map(lambda t: bool(rx.search(t)))
        corpus["text"] = corpus.text.map(lambda t: rx.sub("", t, count=1))
    else:
        corpus["NOTA"] = np.nan
    corpus["TRADUCIDO"] = [(not re.search(r"^en|english|ingl", str(i).lower())) if pd.notna(i) else n
                           for i, n in zip(corpus.IDIOMA, corpus.NOTA)]
    corpus["N_CARACTERES"] = corpus.text.str.len()
    corpus["N_PALABRAS"] = corpus.text.str.split().str.len()
    corpus["CASI_VACIO"] = corpus.N_CARACTERES < MIN_CARACTERES
    corpus["ANIO_FALTANTE"] = corpus.YEAR.isna()
    registrar(PASO, "Textos con nota de traducción detectada", int(pd.Series(corpus.NOTA).fillna(False).sum()))

    # [REVISIÓN] Cuatro banderas de cribado del método 3.5 (para la revisión manual; no excluyen nada)
    RX_TITULO_PERIODISTICO = r"\b(news|press release|announc\w*|interview|blog|op-ed|webinar|event|newsletter|statement by)\b"
    RX_APERTURA_NOTICIA = r"\b(press release|posted on|published on|news|by our reporter|\(reuters\)|\(ap\)|read more)\b"
    deont = corpus.text.map(contar_deonticos)
    corpus["DENS_NORMATIVA"] = [(d["fuerte"] + d["debil"]) / max(n, 1) * POR_MIL for d, n in zip(deont, corpus.N_PALABRAS)]
    corpus["F_TITULO"] = corpus.DOC_TITLE.fillna("").str.contains(RX_TITULO_PERIODISTICO, case=False, regex=True)
    corpus["F_APERTURA"] = corpus.text.str[:600].str.contains(RX_APERTURA_NOTICIA, case=False, regex=True)
    corpus["F_CORTO"] = corpus.N_PALABRAS < MIN_PALABRAS_CRIBADO
    corpus["F_NORMATIVO_BAJO"] = corpus.DENS_NORMATIVA <= corpus.DENS_NORMATIVA.quantile(PERCENTIL_NORMATIVO_CRIBADO)
    banderas = ["F_TITULO", "F_APERTURA", "F_CORTO", "F_NORMATIVO_BAJO"]
    corpus["N_BANDERAS"] = corpus[banderas].sum(axis=1)
    registrar(PASO, "Textos con al menos una bandera de cribado", int((corpus.N_BANDERAS > 0).sum()))

    f_crib = DIR_PROYECTO / "cribado_manual.csv"   # CODE, DECISION (retener / sensibilidad / excluir), MOTIVO
    if f_crib.exists():
        crib = pd.read_csv(f_crib, dtype=str).fillna("")
        dec = dict(zip(crib.CODE, crib.DECISION.str.lower().str.strip()))
        corpus["DECISION_CRIBADO"] = corpus.CODE.map(dec).fillna("retener")
    else:
        corpus["DECISION_CRIBADO"] = "retener"
        (corpus[["CODE", "GOVERNANCE", "DOC_TITLE", "DOC_TYPE_H", "N_PALABRAS", "DENS_NORMATIVA", *banderas, "N_BANDERAS"]]
         .sort_values("N_BANDERAS", ascending=False).assign(DECISION="", MOTIVO="")
         .to_csv(DIR_PROYECTO / "cribado_manual_PLANTILLA.csv", index=False))
        print("Aviso: falta cribado_manual.csv; se generó la plantilla (ordenada por número de banderas).")
    corpus["MARCA_CRIBADO"] = corpus.DECISION_CRIBADO.eq("excluir")
    corpus["SENSIBILIDAD"] = corpus.DECISION_CRIBADO.eq("sensibilidad") | corpus.MARCA_CRIBADO
    if SOLO_DOCUMENTOS_POLITICA and corpus.MARCA_CRIBADO.any():
        excl = set(corpus.CODE[corpus.MARCA_CRIBADO])
        corpus = corpus[~corpus.CODE.isin(excl)].copy(); doc_actor = doc_actor[~doc_actor.CODE.isin(excl)]
        registrar(PASO, "Textos excluidos por cribado (SOLO_DOCUMENTOS_POLITICA)", len(excl))
    guardar_tabla(corpus[corpus.CASI_VACIO][["CODE", "GOVERNANCE", "N_CARACTERES"]], "t02_textos_cortos_marcados")

    # [REVISIÓN] Años posiblemente inconsistentes con el contenido (robustez 5 del método 3.18)
    RX_GENAI = r"\b(generative ai|genai|chatgpt|large language models?|llms?|gpt-?\d)\b"
    corpus["ANIO_SOSPECHOSO"] = (corpus.YEAR < ANIO_CORTE_GENAI).fillna(False) & corpus.text.str.contains(RX_GENAI, case=False, regex=True)
    f_anios = DIR_PROYECTO / "anios_inconsistentes.csv"        # una columna CODE, confirmados a mano
    corpus["ANIO_INCONSISTENTE"] = (corpus.CODE.isin(set(pd.read_csv(f_anios, dtype=str).CODE)) if f_anios.exists()
                                    else corpus.ANIO_SOSPECHOSO)
    guardar_tabla(corpus[corpus.ANIO_SOSPECHOSO][["CODE", "GOVERNANCE", "YEAR", "DOC_TITLE"]], "t02_anios_sospechosos")

    docs_meta = corpus.drop(columns=["text"])

    # 9. Verificaciones finales y guardado
    assert len(registros) == N_REGISTROS_ESPERADOS
    assert len(doc_actor_reg) == int((~registros.SIN_DOCUMENTO).sum())
    assert doc_actor.CODE.isin(corpus.CODE).all()
    flujo = pd.DataFrame({"etapa": [
        "Registros totales", "Registros con documento", "Registros sin documento (ausencia)",
        "Textos únicos (.txt) analizados", "Textos adoptados por más de un registro",
        "Registros que usan un texto compartido", "Actores", "Actores en ausencia",
        "Textos cortos (marcados, no excluidos)", "Textos marcados en cribado (no excluidos)"],
        "valor": [len(registros), int((~registros.SIN_DOCUMENTO).sum()), int(registros.SIN_DOCUMENTO.sum()),
                  len(corpus), int((docs_meta.N_REGISTROS > 1).sum()), int(doc_actor_reg.TEXTO_COMPARTIDO.sum()),
                  len(actores_estado), int(actores_estado.AUSENCIA.sum()),
                  int(corpus.CASI_VACIO.sum()), int(corpus.MARCA_CRIBADO.sum())]})
    print(flujo.to_string(index=False))
    guardar_tabla(flujo, "t02_flujo_corpus")
    guardar_tabla(registros[["REG_ID", "CODE", "TEXTO", "VINCULO", "GOVERNANCE", "ACTOR", "BLOC", "YEAR", "DOC_TYPE_H",
                             "SIN_DOCUMENTO"]], "t02_registros_527")
    for e, v in zip(flujo.etapa, flujo.valor):
        registrar(PASO, e, v)
    registrar(PASO, "Documentos únicos (CODE)", len(corpus))
    for nombre, obj in {"meta": meta, "actores_estado": actores_estado, "doc_actor": doc_actor,
                        "doc_actor_registros": doc_actor_reg, "docs_meta": docs_meta, "corpus": corpus}.items():
        guardar_obj(obj, nombre)
    print(f"\nPaso 2 listo: {len(registros)} registros conservados; {len(corpus)} textos únicos.")
```

```python
if globals().get("STEP2_COMPLETO", True):
    print("El Paso 2 terminó completo: no necesitas esta celda.")
else:
    # ==== Continuar el Paso 2 aunque falten algunos .txt ===========================
    # Corre esta celda JUSTO DESPUÉS del Paso 2, cuando el Paso 2 se haya pausado.
    # Los registros sin .txt se conservan, pero quedan fuera del análisis de texto hasta que
    # subas su archivo; quedan listados en t02_PENDIENTES_sin_txt y en el reporte de calidad.
    print("Diagnóstico del Paso 2:")
    print("  Registros leídos:", len(registros), "(esperados:", N_REGISTROS_ESPERADOS, ")")
    print("  Tipologías sin homologar:", list(no_mapeadas) if len(no_mapeadas) else "ninguna")
    registros["EXCLUIDO_SIN_TXT"] = registros.VINCULO.eq("SIN ARCHIVO")
    print("  Registros con documento pero sin .txt:", int(registros.EXCLUIDO_SIN_TXT.sum()))
    if registros.EXCLUIDO_SIN_TXT.any():
        print(registros[registros.EXCLUIDO_SIN_TXT][["CODE", "FILA_EXCEL", "ACTOR", "DOC_TITLE"]].to_string(index=False))
    dup = registros[registros.CODE.duplicated(keep=False)]
    if len(dup):
        print("  Aviso: CODE repetidos (se analizan igual, pero conviene corregirlos):", sorted(dup.CODE.unique()))
    if len(registros) != N_REGISTROS_ESPERADOS or len(no_mapeadas) or registros.CODE.isna().any():
        raise RuntimeError(">>> Hay avisos de ATENCIÓN (número de registros, tipologías o CODE vacíos). Pégalos en el chat.")

    # 6. Estado de actores (ausencia = ningún registro con documento)
    primero = lambda s: s.dropna().iloc[0] if s.notna().any() else np.nan
    actores_estado = (registros.groupby(["GOV_CODE", "GOVERNANCE", "ACTOR"], dropna=False)
                      .agg(ISO2=("ISO2", primero), REGION=("REGION", primero), BLOC=("BLOC", "first"),
                           BLOC_SENS=("BLOC_SENS", "first"), ORG_CATEGORIA=("ORG_CATEGORIA", "first"),
                           n_registros=("REG_ID", "size"), n_con_doc=("SIN_DOCUMENTO", lambda s: int((~s).sum())),
                           ANIO_MEDIANO=("YEAR", lambda s: s.dropna().median() if s.notna().any() else np.nan))
                      .reset_index())
    actores_estado["AUSENCIA"] = actores_estado.n_con_doc == 0

    # 7. Atributos por texto y vínculo registro-texto (CODE = texto; REG_ID = registro)
    con_doc = registros[~registros.SIN_DOCUMENTO & ~registros.EXCLUIDO_SIN_TXT]
    docs_meta = con_doc.groupby("TEXTO").agg(
        GOV_CODE=("GOV_CODE", "first"), GOVERNANCE=("GOVERNANCE", "first"), DOC_TITLE=("DOC_TITLE", "first"),
        YEAR=("YEAR", "max"),                                     # año más reciente de los registros vinculados
        DOC_TYPE_H=("DOC_TYPE_H", lambda s: " | ".join(sorted(set(s.dropna())))),
        IDIOMA=("LANGUAGE", primero), N_REGISTROS=("REG_ID", "size"), N_ACTORES=("ACTOR", "nunique"),
        NIVELES=("GOV_CODE", lambda s: "+".join(sorted(set(s)))), BLOC_MIXTO=("BLOC", lambda s: s.nunique() > 1),
        SUBNACIONAL=("SUBNACIONAL", "any"), COMUNICACION=("COMUNICACION", "any")).reset_index().rename(columns={"TEXTO": "CODE"})
    doc_actor_reg = (con_doc[["REG_ID", "TEXTO", "CODE", "GOV_CODE", "GOVERNANCE", "ACTOR", "ISO2", "REGION", "BLOC",
                              "BLOC_SENS", "ORG_CATEGORIA"]]
                     .rename(columns={"CODE": "CODE_REGISTRO"}).rename(columns={"TEXTO": "CODE"})
                     .merge(docs_meta[["CODE", "N_REGISTROS"]].rename(columns={"N_REGISTROS": "N_REGISTROS_TEXTO"}), on="CODE"))
    doc_actor_reg["TEXTO_COMPARTIDO"] = doc_actor_reg.N_REGISTROS_TEXTO > 1      # adopción compartida por pertenencia
    guardar_tabla(docs_meta[docs_meta.N_REGISTROS > 1].sort_values("N_REGISTROS", ascending=False), "t02_textos_compartidos")
    guardar_tabla(doc_actor_reg.groupby(["GOVERNANCE", "TEXTO_COMPARTIDO"]).size().reset_index(name="n"),
                  "t02_adopcion_compartida_por_nivel")
    # Para los análisis por actor, cada texto cuenta una vez por actor (método 3.5)
    doc_actor = doc_actor_reg.drop_duplicates(["CODE", "GOV_CODE", "ACTOR"])

    # 8. Lectura de textos y marcas (no eliminan registros)
    rutas = dict(zip(indice_txt.ARCHIVO, indice_txt.RUTA))
    corpus = docs_meta.copy()
    corpus["text"] = corpus.CODE.map(lambda c: Path(rutas[c]).read_text(encoding="utf-8", errors="replace")).map(limpiar_texto)
    if NOTA_TRADUCCION_REGEX:
        rx = re.compile(NOTA_TRADUCCION_REGEX, flags=re.I)
        corpus["NOTA"] = corpus.text.map(lambda t: bool(rx.search(t)))
        corpus["text"] = corpus.text.map(lambda t: rx.sub("", t, count=1))
    else:
        corpus["NOTA"] = np.nan
    corpus["TRADUCIDO"] = [(not re.search(r"^en|english|ingl", str(i).lower())) if pd.notna(i) else n
                           for i, n in zip(corpus.IDIOMA, corpus.NOTA)]
    corpus["N_CARACTERES"] = corpus.text.str.len()
    corpus["N_PALABRAS"] = corpus.text.str.split().str.len()
    corpus["CASI_VACIO"] = corpus.N_CARACTERES < MIN_CARACTERES
    corpus["ANIO_FALTANTE"] = corpus.YEAR.isna()
    registrar(PASO, "Textos con nota de traducción detectada", int(pd.Series(corpus.NOTA).fillna(False).sum()))

    # [REVISIÓN] Cuatro banderas de cribado del método 3.5 (para la revisión manual; no excluyen nada)
    RX_TITULO_PERIODISTICO = r"\b(news|press release|announc\w*|interview|blog|op-ed|webinar|event|newsletter|statement by)\b"
    RX_APERTURA_NOTICIA = r"\b(press release|posted on|published on|news|by our reporter|\(reuters\)|\(ap\)|read more)\b"
    deont = corpus.text.map(contar_deonticos)
    corpus["DENS_NORMATIVA"] = [(d["fuerte"] + d["debil"]) / max(n, 1) * POR_MIL for d, n in zip(deont, corpus.N_PALABRAS)]
    corpus["F_TITULO"] = corpus.DOC_TITLE.fillna("").str.contains(RX_TITULO_PERIODISTICO, case=False, regex=True)
    corpus["F_APERTURA"] = corpus.text.str[:600].str.contains(RX_APERTURA_NOTICIA, case=False, regex=True)
    corpus["F_CORTO"] = corpus.N_PALABRAS < MIN_PALABRAS_CRIBADO
    corpus["F_NORMATIVO_BAJO"] = corpus.DENS_NORMATIVA <= corpus.DENS_NORMATIVA.quantile(PERCENTIL_NORMATIVO_CRIBADO)
    banderas = ["F_TITULO", "F_APERTURA", "F_CORTO", "F_NORMATIVO_BAJO"]
    corpus["N_BANDERAS"] = corpus[banderas].sum(axis=1)
    registrar(PASO, "Textos con al menos una bandera de cribado", int((corpus.N_BANDERAS > 0).sum()))

    f_crib = DIR_PROYECTO / "cribado_manual.csv"   # CODE, DECISION (retener / sensibilidad / excluir), MOTIVO
    if f_crib.exists():
        crib = pd.read_csv(f_crib, dtype=str).fillna("")
        dec = dict(zip(crib.CODE, crib.DECISION.str.lower().str.strip()))
        corpus["DECISION_CRIBADO"] = corpus.CODE.map(dec).fillna("retener")
    else:
        corpus["DECISION_CRIBADO"] = "retener"
        (corpus[["CODE", "GOVERNANCE", "DOC_TITLE", "DOC_TYPE_H", "N_PALABRAS", "DENS_NORMATIVA", *banderas, "N_BANDERAS"]]
         .sort_values("N_BANDERAS", ascending=False).assign(DECISION="", MOTIVO="")
         .to_csv(DIR_PROYECTO / "cribado_manual_PLANTILLA.csv", index=False))
        print("Aviso: falta cribado_manual.csv; se generó la plantilla (ordenada por número de banderas).")
    corpus["MARCA_CRIBADO"] = corpus.DECISION_CRIBADO.eq("excluir")
    corpus["SENSIBILIDAD"] = corpus.DECISION_CRIBADO.eq("sensibilidad") | corpus.MARCA_CRIBADO
    if SOLO_DOCUMENTOS_POLITICA and corpus.MARCA_CRIBADO.any():
        excl = set(corpus.CODE[corpus.MARCA_CRIBADO])
        corpus = corpus[~corpus.CODE.isin(excl)].copy(); doc_actor = doc_actor[~doc_actor.CODE.isin(excl)]
        registrar(PASO, "Textos excluidos por cribado (SOLO_DOCUMENTOS_POLITICA)", len(excl))
    guardar_tabla(corpus[corpus.CASI_VACIO][["CODE", "GOVERNANCE", "N_CARACTERES"]], "t02_textos_cortos_marcados")

    # [REVISIÓN] Años posiblemente inconsistentes con el contenido (robustez 5 del método 3.18)
    RX_GENAI = r"\b(generative ai|genai|chatgpt|large language models?|llms?|gpt-?\d)\b"
    corpus["ANIO_SOSPECHOSO"] = (corpus.YEAR < ANIO_CORTE_GENAI).fillna(False) & corpus.text.str.contains(RX_GENAI, case=False, regex=True)
    f_anios = DIR_PROYECTO / "anios_inconsistentes.csv"        # una columna CODE, confirmados a mano
    corpus["ANIO_INCONSISTENTE"] = (corpus.CODE.isin(set(pd.read_csv(f_anios, dtype=str).CODE)) if f_anios.exists()
                                    else corpus.ANIO_SOSPECHOSO)
    guardar_tabla(corpus[corpus.ANIO_SOSPECHOSO][["CODE", "GOVERNANCE", "YEAR", "DOC_TITLE"]], "t02_anios_sospechosos")

    docs_meta = corpus.drop(columns=["text"])

    # 9. Verificaciones finales y guardado
    assert len(registros) == N_REGISTROS_ESPERADOS
    assert len(doc_actor_reg) == int((~registros.SIN_DOCUMENTO & ~registros.EXCLUIDO_SIN_TXT).sum())
    assert doc_actor.CODE.isin(corpus.CODE).all()
    flujo = pd.DataFrame({"etapa": [
        "Registros totales", "Registros con documento", "Registros sin documento (ausencia)",
        "Textos únicos (.txt) analizados", "Textos adoptados por más de un registro",
        "Registros que usan un texto compartido", "Actores", "Actores en ausencia",
        "Textos cortos (marcados, no excluidos)", "Textos marcados en cribado (no excluidos)"],
        "valor": [len(registros), int((~registros.SIN_DOCUMENTO).sum()), int(registros.SIN_DOCUMENTO.sum()),
                  len(corpus), int((docs_meta.N_REGISTROS > 1).sum()), int(doc_actor_reg.TEXTO_COMPARTIDO.sum()),
                  len(actores_estado), int(actores_estado.AUSENCIA.sum()),
                  int(corpus.CASI_VACIO.sum()), int(corpus.MARCA_CRIBADO.sum())]})
    print(flujo.to_string(index=False))
    guardar_tabla(flujo, "t02_flujo_corpus")
    guardar_tabla(registros[["REG_ID", "CODE", "TEXTO", "VINCULO", "GOVERNANCE", "ACTOR", "BLOC", "YEAR", "DOC_TYPE_H",
                             "SIN_DOCUMENTO"]], "t02_registros_527")
    for e, v in zip(flujo.etapa, flujo.valor):
        registrar(PASO, e, v)
    registrar(PASO, "Documentos únicos (CODE)", len(corpus))
    for nombre, obj in {"meta": meta, "actores_estado": actores_estado, "doc_actor": doc_actor,
                        "doc_actor_registros": doc_actor_reg, "docs_meta": docs_meta, "corpus": corpus}.items():
        guardar_obj(obj, nombre)
    registrar(PASO, "Registros con documento pero sin .txt (excluidos temporalmente)", int(registros.EXCLUIDO_SIN_TXT.sum()))
    guardar_tabla(registros[registros.EXCLUIDO_SIN_TXT][["REG_ID", "CODE", "GOVERNANCE", "ACTOR", "DOC_TITLE"]], "t02_PENDIENTES_sin_txt")
    guardar_obj(registros, "registros")
    print(f"\nPaso 2 listo: {len(registros)} registros conservados; {len(corpus)} textos únicos.")
```

## Paso 3. Preprocesamiento del texto (método 3.8)
Seis pasos del método: limpieza, tokenización, *stop words* (onix + SMART + Snowball de tidytext),
lematización (mismo diccionario que `textstem`), *stop words* de dominio y poda relativa
(< 3 documentos o > 90 % de los documentos). Detecta además casi duplicados (coseno ≥ 0,95).
Produce: `dtm`, `casi_dup`.

```python
from scipy import sparse
PASO = "Step3"
corpus = leer_obj("corpus")

RX_URL = re.compile(r"https?://\S+|www\.\S+|\S+@\S+")
RX_TOKEN = re.compile(r"[^\W_]+(?:'[^\W_]+)*")


def es_palabra(w):
    return len(w) >= 3 and w[0].isalpha() and w[-1].isalpha() and all(c.isalpha() or c in "'-" for c in w)


def tokens_crudos(txt):
    txt = re.sub(r"[’‘]", "'", txt)
    txt = RX_URL.sub(" ", txt).lower()
    return [re.sub(r"'s$|^'+|'+$", "", w) for w in RX_TOKEN.findall(txt)]


# [REVISIÓN] Reparación de palabras partidas por la extracción del PDF («confo rmity» -> «conformity»).
# Se unen dos trozos contiguos x, y cuando (a) al menos uno no es palabra del inglés y (b) la unión x+y sí lo es,
# o bien la unión aparece en el corpus en al menos el triple de documentos que cada trozo por separado.
# Cada reparación queda registrada en t03_palabras_reparadas para revisarla.
crudos = {code: tokens_crudos(txt) for code, txt in zip(corpus.CODE, corpus.text)}
DICC_EN = diccionario_ingles()
FREC_CRUDA = pd.Series([w for v in crudos.values() for w in set(v)]).value_counts().to_dict()
from collections import Counter
reparadas = Counter()


def es_union(x, y):
    if not (x.isalpha() and y.isalpha()) or (x in DICC_EN and y in DICC_EN) or (x in STOP_TODAS and y in STOP_TODAS):
        return False
    xy = x + y
    if xy in DICC_EN:
        return True
    f = FREC_CRUDA.get(xy, 0)
    return f >= 3 and f >= 3 * max(FREC_CRUDA.get(x, 0), FREC_CRUDA.get(y, 0))


def unir_fragmentos(toks):
    out, i = [], 0
    while i < len(toks):
        if i + 1 < len(toks) and es_union(toks[i], toks[i + 1]):
            reparadas[f"{toks[i]} {toks[i + 1]} -> {toks[i] + toks[i + 1]}"] += 1
            out.append(toks[i] + toks[i + 1]); i += 2
            continue
        out.append(toks[i]); i += 1
    return out


STOP_DOM = set(STOP_DOMINIO)
tokens = {}
for code, toks in crudos.items():
    t = [w for w in unir_fragmentos(toks) if es_palabra(w) and w not in STOP_TODAS]
    if USAR_LEMATIZACION:
        t = lematizar(t)
    tokens[code] = [w for w in t if w not in STOP_DOM and w not in STOP_TODAS]
guardar_tabla(pd.DataFrame(reparadas.most_common(), columns=["REPARACION", "n"]), "t03_palabras_reparadas")
registrar(PASO, "Palabras partidas reparadas (ocurrencias)", sum(reparadas.values()))

# Frecuencia documental y poda relativa (Maier et al., 2018)
n_docs = sum(1 for v in tokens.values() if v)
df_freq = pd.Series([w for v in tokens.values() for w in set(v)]).value_counts()
frec_doc = pd.DataFrame({"word": df_freq.index, "DOC_FREQ": df_freq.values})
frec_doc["PROP_DOCS"] = frec_doc.DOC_FREQ / n_docs
guardar_tabla(frec_doc[frec_doc.PROP_DOCS > MAX_PROP_TERMINO].sort_values("PROP_DOCS", ascending=False),
              "t03_terminos_ubicuos")
vocab = sorted(frec_doc.word[(frec_doc.DOC_FREQ >= MIN_DOCS_TERMINO) & (frec_doc.PROP_DOCS <= MAX_PROP_TERMINO)])
idx_vocab = {w: i for i, w in enumerate(vocab)}

# [REVISIÓN] Auditoría del vocabulario. Clasifica los términos que no están en el diccionario de inglés:
#  · «posible fragmento»: es el inicio o el final de otra palabra del vocabulario (p. ej., «olicy» de «policy»);
#    casi seguro es un resto de la extracción del PDF -> agrégalo a STOP_DOMINIO (Paso 1) o corrige el texto.
#  · «fuera del diccionario»: sigla, nombre propio o término técnico (p. ej., «unesco», «chatgpt»); suele ser real.
voc_largo = [w for w in vocab if len(w) >= 6]
def tipo_termino(w):
    if w in DICC_EN:
        return None
    if len(w) <= 6 and any(v != w and (v.startswith(w) or v.endswith(w)) and len(v) >= len(w) + 2 for v in voc_largo):
        return "posible fragmento"
    return "fuera del diccionario"
frec_doc["TIPO"] = frec_doc.word.map(lambda w: tipo_termino(w) if w in idx_vocab else None)
dudosos = frec_doc[frec_doc.TIPO.notna()].sort_values(["TIPO", "DOC_FREQ"], ascending=[False, False]).copy()
ejemplo = {}
for code, toks in tokens.items():
    for w in set(toks) & set(dudosos.word):
        ejemplo.setdefault(w, code)
dudosos["EJEMPLO_CODE"] = dudosos.word.map(ejemplo)
guardar_tabla(dudosos, "t03_terminos_dudosos")
frag = dudosos[dudosos.TIPO == "posible fragmento"]
registrar(PASO, "Términos que parecen fragmentos de palabras", len(frag))
registrar(PASO, "Términos fuera del diccionario (siglas, nombres propios, técnicos)", int((dudosos.TIPO == "fuera del diccionario").sum()))
if len(frag):
    print(f"[AVISO] {len(frag)} términos parecen fragmentos de palabras (ver t03_terminos_dudosos):", ", ".join(frag.word.head(40)))

filas, cols, vals, docs_dtm = [], [], [], []
for code in corpus.CODE:
    c = pd.Series([w for w in tokens[code] if w in idx_vocab]).value_counts()
    if len(c) == 0:
        continue
    i = len(docs_dtm); docs_dtm.append(code)
    filas += [i] * len(c); cols += [idx_vocab[w] for w in c.index]; vals += list(c.values)
X = sparse.csr_matrix((vals, (filas, cols)), shape=(len(docs_dtm), len(vocab)), dtype=np.int64)
dtm = {"X": X, "docs": docs_dtm, "terms": vocab}

guardar_tabla(pd.DataFrame({"CODE": sorted(set(corpus.CODE) - set(docs_dtm))}), "t03_docs_sin_terminos")
guardar_tabla(corpus[corpus.N_CARACTERES < MIN_CARACTERES][["CODE", "GOVERNANCE", "DOC_TITLE", "N_CARACTERES"]],
              "t03_docs_cortos")

# ---- Casi duplicados (revisión manual; el Paso 12 los excluye como robustez) ----------
Xn = X.astype(float)
normas = np.sqrt(np.asarray(Xn.multiply(Xn).sum(axis=1)).ravel())
Xn = sparse.diags(1 / np.where(normas > 0, normas, 1)) @ Xn
sim = (Xn @ Xn.T).toarray()
iu = np.triu_indices_from(sim, k=1)
sel = sim[iu] >= UMBRAL_CASI_DUP
meta_c = corpus.set_index("CODE")
casi_dup = pd.DataFrame({"CODE_1": np.array(docs_dtm)[iu[0][sel]], "CODE_2": np.array(docs_dtm)[iu[1][sel]],
                         "COSENO": sim[iu][sel]})
for k in ("1", "2"):
    casi_dup[f"ANIO_{k}"] = casi_dup[f"CODE_{k}"].map(meta_c.YEAR)
    casi_dup[f"TIT_{k}"] = casi_dup[f"CODE_{k}"].map(meta_c.DOC_TITLE)
casi_dup = casi_dup.sort_values("COSENO", ascending=False)
guardar_tabla(casi_dup, "t03_casi_duplicados")

registrar(PASO, "Documentos en la DTM", X.shape[0])
registrar(PASO, "Términos en la DTM", X.shape[1])
registrar(PASO, "Documentos sin términos tras la poda", len(TAB["t03_docs_sin_terminos"]))
registrar(PASO, f"Documentos con menos de {MIN_CARACTERES} caracteres", len(TAB["t03_docs_cortos"]))
registrar(PASO, "Pares casi duplicados", len(casi_dup))
guardar_obj(dtm, "dtm"); guardar_obj(casi_dup, "casi_dup")
print(f"Paso 3 listo: DTM {X.shape[0]} documentos x {X.shape[1]} términos; {len(casi_dup)} pares casi duplicados.")
```

## Paso 4. Modelo de tópicos LDA (método 3.9; RQ1a)
Selección de *k* por validación cruzada de 5 particiones con perplejidad en datos retenidos
(*k* = 4 a 20, de 2 en 2), modelo final, matrices β y γ y etiquetas de tópicos.

**[REVISIÓN] Diferencia con R:** `topicmodels` estima LDA por EM variacional y además estima α;
`scikit-learn` usa Bayes variacional con priors fijos (α = η = 1/k). Los tópicos son comparables,
pero **no idénticos** a los de R: si ya nombraste los tópicos con los resultados de R, revisa de nuevo
`t04_top_terminos` y `t04_docs_representativos` antes de fijar `ETIQUETAS_TOPICOS` **[D3]**.
Si cambias de software, actualiza la sección 3.9 y 3.19 del método.

```python
from sklearn.decomposition import LatentDirichletAllocation
import time
PASO = "Step4"
dtm = leer_obj("dtm"); X = dtm["X"]; terms = np.array(dtm["terms"])


def ajustar_lda(Xm, k, semilla=SEMILLA):
    return LatentDirichletAllocation(n_components=k, learning_method="batch", max_iter=LDA_MAX_ITER,
                                     random_state=semilla, n_jobs=-1).fit(Xm)


# ---- 1. Selección de k [D1] -----------------------------------------------------------
if EJECUTAR_CV_K:
    folds = np.random.default_rng(SEMILLA).permutation(np.resize(np.arange(1, N_FOLDS + 1), X.shape[0]))
    filas = []
    for k in K_GRID:
        t0 = time.time()
        for f in range(1, N_FOLDS + 1):
            mod = ajustar_lda(X[folds != f], k)
            filas.append({"k": k, "fold": f, "perplejidad": mod.perplexity(X[folds == f])})
        print(f"  k = {k}: {time.time() - t0:.0f} s")
    cv_resumen = (pd.DataFrame(filas).groupby("k").perplejidad.agg(perplejidad_media="mean", perplejidad_de="std")
                  .reset_index())
    guardar_obj(cv_resumen, "cv_k")
else:
    cv_resumen = leer_obj("cv_k") if existe_obj("cv_k") else None

if cv_resumen is not None:
    guardar_tabla(cv_resumen, "t04_seleccion_k")
    fig, ax = plt.subplots()
    ax.errorbar(cv_resumen.k, cv_resumen.perplejidad_media, yerr=cv_resumen.perplejidad_de, color="#2a78d6",
                marker="o", capsize=3)
    ax.set_xticks(K_GRID); ax.set_xlabel("Number of topics (k)"); ax.set_ylabel("Mean held-out perplexity")
    guardar_fig(fig, "f04_seleccion_k", 8, 5)

# ---- 2. Modelo final --------------------------------------------------------------------
lda_final = ajustar_lda(X, K_FINAL)
beta = lda_final.components_ / lda_final.components_.sum(1, keepdims=True)
gamma = lda_final.transform(X)
beta_td = pd.DataFrame([(t + 1, terms[j], beta[t, j]) for t in range(K_FINAL) for j in range(len(terms))],
                       columns=["topic", "term", "beta"])
gamma_td = pd.DataFrame([(dtm["docs"][i], t + 1, gamma[i, t]) for i in range(gamma.shape[0]) for t in range(K_FINAL)],
                        columns=["CODE", "topic", "gamma"])

top_terminos = (beta_td.sort_values(["topic", "beta"], ascending=[True, False]).groupby("topic").head(15)
                .reset_index(drop=True))
if ETIQUETAS_TOPICOS is not None and len(ETIQUETAS_TOPICOS) != K_FINAL:
    raise ValueError("ETIQUETAS_TOPICOS debe tener K_FINAL elementos.")
etiquetas = top_terminos.groupby("topic").term.apply(lambda s: ", ".join(s.head(4))).reset_index(name="TOP_TERMS")
etiquetas["TOPIC_LABEL"] = [f"T{t}: {ETIQUETAS_TOPICOS[t-1]}" if ETIQUETAS_TOPICOS else f"T{t}: {tt}"
                            for t, tt in zip(etiquetas.topic, etiquetas.TOP_TERMS)]
try:
    _t = pd.read_csv(DIR_TAB / "t03_terminos_dudosos.csv")
    _dudosos = set(_t.word[_t.TIPO == "posible fragmento"]) if "TIPO" in _t else set(_t.word)
except Exception:
    _dudosos = set()
top_terminos["DUDOSO"] = top_terminos.term.isin(_dudosos)
guardar_tabla(top_terminos.merge(etiquetas, on="topic"), "t04_top_terminos")
if top_terminos.DUDOSO.any():
    print("[AVISO] Posibles fragmentos de palabras entre los 15 términos más probables de algún tópico:",
          ", ".join(sorted(set(top_terminos.term[top_terminos.DUDOSO]))),
          "\n  Si son restos del PDF, agrégalos a STOP_DOMINIO en el Paso 1 y vuelve a correr los pasos 3 y 4.")

# Documentos con mayor γ por tópico, para leerlos y nombrar los tópicos [D3]
corpus = leer_obj("corpus")
guardar_tabla(gamma_td.sort_values("gamma", ascending=False).groupby("topic").head(5)
              .merge(corpus[["CODE", "GOVERNANCE", "DOC_TITLE"]], on="CODE", how="left").sort_values(["topic", "gamma"]),
              "t04_docs_representativos")

ncol = 4; nfil = math.ceil(K_FINAL / ncol)
fig, axs = plt.subplots(nfil, ncol, squeeze=False)
for t, ax in zip(range(1, K_FINAL + 1), axs.ravel()):
    d = top_terminos[top_terminos.topic == t].iloc[::-1]
    ax.barh(d.term, d.beta, color="#2a78d6"); ax.set_title(f"topic {t}", fontsize=8); ax.tick_params(labelsize=6)
for ax in axs.ravel()[K_FINAL:]:
    ax.axis("off")
guardar_fig(fig, "f04_beta_topicos", 12, 3 * nfil)

registrar(PASO, "k final", K_FINAL)
for nombre, obj in {"lda_final": lda_final, "beta_td": beta_td, "gamma_td": gamma_td, "etiquetas": etiquetas}.items():
    guardar_obj(obj, nombre)
print(f"Paso 4 listo: LDA con k = {K_FINAL}")
print(etiquetas.to_string(index=False))
```

## Paso 5. RQ1a: narrativas por nivel y validación de la tipología (método 3.10–3.11)
* Perfiles de documento: tópico dominante y secundario, número de tópicos con γ ≥ 0,20, entropía normalizada.
* Validación de la tipología por título (procedimiento de Silge y Robinson, cap. 6): tópico de consenso
  por tipo, documentos discrepantes, tópicos emergentes, χ² y V de Cramér (descriptivos).
* Perfiles empíricos: Ward (ward.D2) sobre distancias de Hellinger.
* Perfiles de actor (γ medio de sus documentos) y distancia de Hellinger al referente global.
Produce: `metricas_doc`, `actor_topico`, `actor_metricas`, `clusters_docs`.

```python
from scipy.cluster.hierarchy import linkage, fcluster
PASO = "Step5"
docs_meta = leer_obj("docs_meta"); doc_actor = leer_obj("doc_actor")
gamma_td = leer_obj("gamma_td"); etiquetas = leer_obj("etiquetas")
K = gamma_td.topic.nunique()
mat_g = gamma_td.pivot(index="CODE", columns="topic", values="gamma").sort_index(axis=1)
mat_g.columns = [f"T{t}" for t in mat_g.columns]

# ---- 1. Métricas por documento (método 3.11) -------------------------------------------
def perfil_doc(g):
    g = g.sort_values("gamma", ascending=False)
    rel = g[g.gamma >= UMBRAL_MULTI]
    return pd.Series({"TOPIC_DOM": g.topic.iloc[0], "GAMMA_DOM": g.gamma.iloc[0],
                      "TOPIC_SEC": g.topic.iloc[1], "GAMMA_SEC": g.gamma.iloc[1],
                      "N_TOPICS_RELEVANT": len(rel), "ENTROPY_NORM": entropia_norm(g.gamma.values),
                      "PROFILE": " + ".join(f"T{t} ({v:.2f})" for t, v in zip(rel.topic, rel.gamma)) or "None"})


metricas_doc = resumen_por(gamma_td, "CODE", perfil_doc)
metricas_doc["PROFILE_TYPE"] = pd.cut(metricas_doc.N_TOPICS_RELEVANT, [-1, 0, 1, 2, 99],
                                      labels=["Diffuse", "Monothematic", "Bithematic", "Multithematic"]).astype(str)
metricas_doc = metricas_doc.merge(docs_meta[["CODE", "GOV_CODE", "GOVERNANCE", "DOC_TYPE_H", "YEAR"]], on="CODE", how="left")
for c in ("TOPIC_DOM", "TOPIC_SEC", "N_TOPICS_RELEVANT"):
    metricas_doc[c] = metricas_doc[c].astype(int)
guardar_tabla(metricas_doc, "t05_metricas_documento")
tp = metricas_doc.groupby(["GOVERNANCE", "PROFILE_TYPE"]).size().reset_index(name="n")
tp["prop"] = tp.n / tp.groupby("GOVERNANCE").n.transform("sum")
guardar_tabla(tp, "t05_tipo_perfil_por_nivel")

# ---- 2. Validación de la tipología por título (método 3.10) ------------------------------
dom = metricas_doc[~metricas_doc.DOC_TYPE_H.fillna("").str.contains(r"\|") & (metricas_doc.DOC_TYPE_H.fillna("") != "")].copy()
consenso = dom.groupby("DOC_TYPE_H").TOPIC_DOM.agg(lambda s: s.value_counts().index[0])
dom["TOPICO_CONSENSO"] = dom.DOC_TYPE_H.map(consenso)
dom["DISCREPANTE"] = dom.TOPIC_DOM != dom.TOPICO_CONSENSO
coincidencia = lambda tipo, top: np.mean(top == pd.Series(top).groupby(tipo).transform(lambda s: s.value_counts().index[0]).values)
c_obs = coincidencia(dom.DOC_TYPE_H.values, dom.TOPIC_DOM.values)
c_nula = np.array([coincidencia(rng.permutation(dom.DOC_TYPE_H.values), dom.TOPIC_DOM.values) for _ in range(N_PERMUT)])
asoc = chi_sim(dom.DOC_TYPE_H, dom.TOPIC_DOM)
guardar_tabla(pd.DataFrame([{"n_docs": len(dom), "chi2": asoc.chi2, "cramer_v": asoc.cramer_v, "p_chi_sim": asoc.p_valor_sim,
                             "coincidencia_obs": c_obs, "coincidencia_nula_media": c_nula.mean(),
                             "p_perm": (1 + np.sum(c_nula >= c_obs)) / (N_PERMUT + 1),
                             "n_discrepantes": int(dom.DISCREPANTE.sum())}]), "t05_validacion_tipologias")
modal = dom.groupby("DOC_TYPE_H").agg(n=("CODE", "size"), topico_consenso=("TOPICO_CONSENSO", "first"),
                                      prop_consenso=("DISCREPANTE", lambda s: 1 - s.mean())).reset_index()
guardar_tabla(modal, "t05_tipologia_topico_modal")
guardar_tabla(dom[dom.DISCREPANTE].merge(docs_meta[["CODE", "DOC_TITLE"]], on="CODE")
              [["CODE", "GOVERNANCE", "DOC_TITLE", "DOC_TYPE_H", "TOPICO_CONSENSO", "TOPIC_DOM", "GAMMA_DOM", "PROFILE"]],
              "t05_docs_discrepantes_LEER")                                   # paso 3: leer para distinguir errores
emergentes = sorted(set(range(1, K + 1)) - set(consenso.values))
guardar_tabla(etiquetas[etiquetas.topic.isin(emergentes)], "t05_topicos_emergentes")   # paso 4 del método

tipo_topico = (gamma_td.merge(dom[["CODE", "DOC_TYPE_H"]], on="CODE").groupby(["DOC_TYPE_H", "topic"]).gamma.mean()
               .unstack())
fig, ax = plt.subplots()
im = ax.imshow(tipo_topico.values, cmap="seguile_azul", aspect="auto")
ax.set_xticks(range(K)); ax.set_xticklabels(etiquetas.set_index("topic").TOPIC_LABEL.reindex(tipo_topico.columns), rotation=45, ha="right")
ax.set_yticks(range(len(tipo_topico))); ax.set_yticklabels(tipo_topico.index)
for i in range(tipo_topico.shape[0]):
    for j in range(K):
        ax.text(j, i, f"{tipo_topico.values[i, j]:.2f}", ha="center", va="center", fontsize=7,
                color="white" if tipo_topico.values[i, j] > tipo_topico.values.max() * .6 else "black")
fig.colorbar(im, label="Mean γ"); guardar_fig(fig, "f05_tipologia_por_topico", 12, 6)

# ---- 3. Perfiles empíricos (Hellinger + Ward; método 3.11) ---------------------------------
Z = linkage(np.sqrt(mat_g.values) / np.sqrt(2), method="ward")
clusters = pd.DataFrame({"CODE": mat_g.index, "PROFILE_CLUSTER": [f"C{c}" for c in fcluster(Z, K_PERFILES, "maxclust")]})
clusters = clusters.merge(docs_meta[["CODE", "GOVERNANCE", "DOC_TYPE_H"]], on="CODE", how="left")
guardar_tabla(mat_g.reset_index().merge(clusters, on="CODE").groupby("PROFILE_CLUSTER")
              .agg(n=("CODE", "size"), **{c: (c, "mean") for c in mat_g.columns}).reset_index(), "t05_perfiles_empiricos")
guardar_tabla(pd.DataFrame([{**chi_sim(clusters.PROFILE_CLUSTER, clusters.GOVERNANCE).to_dict(), "cruce": "cluster x nivel"},
                            {**chi_sim(clusters.PROFILE_CLUSTER, clusters.DOC_TYPE_H).to_dict(), "cruce": "cluster x tipologia"}]),
              "t05_perfiles_asociacion")

# ---- 4. Perfil por actor y diferencias entre niveles ----------------------------------------
CLAVES_ACTOR = ["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]
actor_topico = (gamma_td.merge(doc_actor, on="CODE").groupby(CLAVES_ACTOR + ["topic"], dropna=False)
                .agg(gamma=("gamma", "mean"), n_docs=("CODE", "nunique")).reset_index())

kw_topicos = resumen_por(actor_topico, "topic", lambda g: kw_niveles(g.gamma, g.GOVERNANCE))
kw_topicos["p_BH"] = p_bh(kw_topicos.p_valor)
medias = actor_topico.groupby(["topic", "GOV_CODE"]).gamma.mean().unstack().add_prefix("gamma_media_").reset_index()
guardar_tabla(kw_topicos.merge(medias, on="topic").merge(etiquetas, on="topic"), "t05_RQ1a_topicos_por_nivel")

dist_global = distancia_referente(actor_topico, "topic", "gamma").rename(columns={"HELLINGER": "HELLINGER_GLOBAL"})
actor_metricas = (actor_topico.groupby(CLAVES_ACTOR, dropna=False)
                  .apply(lambda g: pd.Series({"n_docs": g.n_docs.max(), "TOPIC_DOM": int(g.topic.iloc[g.gamma.values.argmax()]),
                                              "GAMMA_DOM": g.gamma.max(), "ENTROPY_NORM": entropia_norm(g.gamma.values)}),
                         include_groups=False).reset_index()
                  .merge(dist_global, on=["GOV_CODE", "ACTOR"], how="left"))
actor_metricas["ES_REFERENTE"] = (actor_metricas.GOV_CODE == "G") & (actor_metricas.BLOC == "Global")
sin_ref = actor_metricas[~actor_metricas.ES_REFERENTE]     # [REVISIÓN] método 3.15: se excluye el referente
guardar_tabla(pd.DataFrame([{**kw_niveles(sin_ref.HELLINGER_GLOBAL, sin_ref.GOVERNANCE).to_dict(),
                             "indicador": "Hellinger al referente global (sin el referente)"},
                            {**kw_niveles(actor_metricas.ENTROPY_NORM, actor_metricas.GOVERNANCE).to_dict(),
                             "indicador": "Entropía normalizada"}]), "t05_RQ1a_indicadores_por_nivel")

nivel_top = actor_topico.groupby(["GOVERNANCE", "topic"]).gamma.mean().unstack()
fig, ax = plt.subplots()
ax.imshow(nivel_top.values, cmap="seguile_azul", aspect="auto")
ax.set_xticks(range(K)); ax.set_xticklabels(etiquetas.set_index("topic").TOPIC_LABEL.reindex(nivel_top.columns), rotation=45, ha="right")
ax.set_yticks(range(len(nivel_top))); ax.set_yticklabels(nivel_top.index)
for i in range(nivel_top.shape[0]):
    for j in range(K):
        ax.text(j, i, f"{nivel_top.values[i, j]:.2f}", ha="center", va="center", fontsize=7,
                color="white" if nivel_top.values[i, j] > nivel_top.values.max() * .6 else "black")
guardar_fig(fig, "f05_RQ1a_perfil_por_nivel", 12, 4.5)

for nombre, obj in {"metricas_doc": metricas_doc, "actor_topico": actor_topico, "actor_metricas": actor_metricas,
                    "clusters_docs": clusters}.items():
    guardar_obj(obj, nombre)

# ---- 5. Perfil de los 527 registros (el texto compartido se asigna a cada registro) ----------
registros = leer_obj("registros")
perfil_registros = (registros[["REG_ID", "TEXTO", "GOVERNANCE", "ACTOR", "BLOC", "YEAR", "DOC_TYPE_H", "SIN_DOCUMENTO"]]
                    .merge(metricas_doc[["CODE", "TOPIC_DOM", "GAMMA_DOM", "PROFILE", "PROFILE_TYPE", "ENTROPY_NORM"]]
                           .rename(columns={"CODE": "TEXTO"}), on="TEXTO", how="left")
                    .merge(mat_g.reset_index().rename(columns={"CODE": "TEXTO"}), on="TEXTO", how="left"))
faltan_gamma = perfil_registros[~perfil_registros.SIN_DOCUMENTO & perfil_registros.TOPIC_DOM.isna()]
if len(faltan_gamma):
    guardar_tabla(faltan_gamma, "t05_ERROR_registros_sin_gamma")
    print(f"[AVISO] {len(faltan_gamma)} registros sin γ (texto sin términos o sin archivo). Ver t05_ERROR_registros_sin_gamma.")
guardar_tabla(perfil_registros, "t05_perfil_527_registros")
print("Paso 5 (RQ1a) listo.")
```

## Paso 6. RQ1b: fuerza normativa (método 3.13)
**Capa 1, forma del instrumento**, en las cinco categorías del método (no las seis de los scripts R):

| Valor | Categoría | Tipologías armonizadas |
|---|---|---|
| 3 | Legally binding | Treaty / Convention, Regulation |
| 2 | Institutionally mandatory | Internal Policies |
| 1 | Voluntary or soft | Strategy, Recommendations, Ethical Framework / Principles, Standards, Guidelines and Guides |
| 0 | Descriptive | Report, Discussion Paper |
| −1 | Absence | registros sin documento |

Codificación en tres pasos: regla por tipología → cribado de títulos (*Act, Decree, Directive,
Instruction, Code of Conduct, voluntary, draft*) → revisión manual **[D2]** en `fuerza_overrides.csv`.

**Capa 2, intensidad deóntica:** obligaciones fuertes y débiles por 1.000 palabras e índice
fuerte / (fuerte + débil), solo con ≥ 5 expresiones. Validez convergente con Kruskal-Wallis del índice
entre categorías y fuerza de cada narrativa (índice medio ponderado por γ).
Produce: `fuerza_doc`, `fuerza_actor`.

```python
PASO = "Step6"
corpus = leer_obj("corpus"); doc_actor = leer_obj("doc_actor"); actores_estado = leer_obj("actores_estado")

FUERZA_TIPO = {"Treaty / Convention": 3, "Regulation": 3, "Internal Policies": 2, "Strategy": 1,
               "Recommendations": 1, "Ethical Framework / Principles": 1, "Standards": 1,
               "Guidelines and Guides": 1, "Report": 0, "Discussion Paper": 0}
ETIQ_FUERZA = {-1: "Absence", 0: "Descriptive", 1: "Voluntary or soft", 2: "Institutionally mandatory",
               3: "Legally binding"}
INDICADORES_TITULO = [(r"\b(act|decree|law)\b", 3), (r"\b(directive|instruction|code of conduct)\b", 2),
                      (r"\b(voluntary|draft)\b", 1)]

# ---- 1. Capa 1: forma del instrumento --------------------------------------------------
capa1 = corpus[["CODE", "GOV_CODE", "DOC_TITLE", "DOC_TYPE_H"]].copy()
capa1["TIPOS"] = capa1.DOC_TYPE_H.fillna("").str.split(r" \| ")
capa1["NIVEL_TIPO"] = capa1.TIPOS.map(lambda ts: max(FUERZA_TIPO.get(t, -9) for t in ts))
if (capa1.NIVEL_TIPO == -9).any():
    raise ValueError(f"Tipologías sin categoría de fuerza: {capa1[capa1.NIVEL_TIPO == -9].DOC_TYPE_H.unique()}")


def indicio_titulo(t):
    for pat, nivel in INDICADORES_TITULO:
        if re.search(pat, str(t), flags=re.I):
            return nivel
    return np.nan


capa1["INDICIO_TITULO"] = capa1.DOC_TITLE.map(indicio_titulo)
capa1["REVISAR"] = ((capa1.TIPOS.str.len() > 1) | capa1.TIPOS.map(lambda ts: "Standards" in ts)
                    | (capa1.INDICIO_TITULO.notna() & (capa1.INDICIO_TITULO != capa1.NIVEL_TIPO)))
plantilla = capa1[capa1.REVISAR].assign(NIVEL_FUERZA_FINAL="", JUSTIFICACION="")
plantilla.drop(columns="TIPOS").to_csv(DIR_PROYECTO / "fuerza_overrides_PLANTILLA.csv", index=False)
registrar(PASO, "Registros marcados para revisión de fuerza", int(capa1.REVISAR.sum()))

f_over = DIR_PROYECTO / "fuerza_overrides.csv"      # CODE, NIVEL_FUERZA_FINAL, JUSTIFICACION
capa1["NIVEL_FUERZA_FINAL"] = np.nan; capa1["JUSTIFICACION"] = None
if f_over.exists():
    over = pd.read_csv(f_over, dtype={"CODE": str})
    over["NIVEL_FUERZA_FINAL"] = pd.to_numeric(over.NIVEL_FUERZA_FINAL, errors="coerce")
    if ESCALA_OVERRIDES == "R_v1":                  # escala 0-4 de los scripts en R
        over["NIVEL_FUERZA_FINAL"] = over.NIVEL_FUERZA_FINAL.map({4: 3, 3: 2, 2: 1, 1: 1, 0: 0})
    capa1 = capa1.drop(columns=["NIVEL_FUERZA_FINAL", "JUSTIFICACION"]).merge(
        over[["CODE", "NIVEL_FUERZA_FINAL", "JUSTIFICACION"]], on="CODE", how="left")
else:
    print("[AVISO] [D2] Falta fuerza_overrides.csv: se usa solo la tipología.")
capa1["NIVEL_FUERZA"] = capa1.NIVEL_FUERZA_FINAL.fillna(capa1.NIVEL_TIPO).astype(int)
guardar_tabla(capa1[capa1.NIVEL_FUERZA_FINAL.notna()].drop(columns="TIPOS"), "t06_overrides_aplicados")
registrar(PASO, "Registros reclasificados a mano", int(capa1.NIVEL_FUERZA_FINAL.notna().sum()))

# ---- 2. Capa 2: intensidad deóntica (texto completo, antes de quitar stop words) -----------
cuentas = pd.DataFrame(list(corpus.text.map(contar_deonticos)), index=corpus.index)
fuerza_doc = capa1.drop(columns="TIPOS").merge(
    corpus[["CODE", "N_PALABRAS", "TRADUCIDO", "YEAR", "GOVERNANCE"]].assign(FUERTE=cuentas.fuerte.values, DEBIL=cuentas.debil.values),
    on="CODE")
fuerza_doc["TOTAL_DEONT"] = fuerza_doc.FUERTE + fuerza_doc.DEBIL
fuerza_doc["FUERTE_1000"] = fuerza_doc.FUERTE / fuerza_doc.N_PALABRAS * POR_MIL
fuerza_doc["DEBIL_1000"] = fuerza_doc.DEBIL / fuerza_doc.N_PALABRAS * POR_MIL
fuerza_doc["DENS_DEONTICA"] = fuerza_doc.TOTAL_DEONT / fuerza_doc.N_PALABRAS * POR_MIL
fuerza_doc["INDICE_DEONTICO"] = np.where(fuerza_doc.TOTAL_DEONT >= MIN_DEONTICAS,
                                         fuerza_doc.FUERTE / fuerza_doc.TOTAL_DEONT.where(fuerza_doc.TOTAL_DEONT > 0), np.nan)
fuerza_doc["FUERZA_ETIQ"] = fuerza_doc.NIVEL_FUERZA.map(ETIQ_FUERZA)
guardar_tabla(fuerza_doc.drop(columns="DOC_TITLE"), "t06_fuerza_documento")

# ---- 3. Validez convergente y fuerza de las narrativas (método 3.13) -------------------------
ct = stats.spearmanr(fuerza_doc.NIVEL_FUERZA, fuerza_doc.INDICE_DEONTICO, nan_policy="omit")
guardar_tabla(pd.DataFrame([{**kw_niveles(fuerza_doc.INDICE_DEONTICO, fuerza_doc.FUERZA_ETIQ).to_dict(),
                             "prueba": "Índice deóntico entre categorías de instrumento (KW)"},
                            {"n": int(fuerza_doc.INDICE_DEONTICO.notna().sum()), "rho": ct.statistic, "p_valor": ct.pvalue,
                             "prueba": "Spearman capa 1 x índice (complementaria)"}]), "t06_validez_convergente")
guardar_tabla(fuerza_doc.groupby("FUERZA_ETIQ").agg(n=("CODE", "size"), indice_mediana=("INDICE_DEONTICO", "median"),
                                                     fuerte_1000=("FUERTE_1000", "median"), debil_1000=("DEBIL_1000", "median"))
              .reset_index(), "t06_indice_por_categoria")
if existe_obj("gamma_td"):
    gt = leer_obj("gamma_td").merge(fuerza_doc[["CODE", "INDICE_DEONTICO", "FUERZA_ETIQ"]], on="CODE")
    ok = gt.INDICE_DEONTICO.notna()
    fz = (gt[ok].assign(p=gt.gamma * gt.INDICE_DEONTICO).groupby("topic")
          .apply(lambda g: pd.Series({"fuerza_narrativa": g.p.sum() / g.gamma.sum(), "n_docs": len(g)}), include_groups=False)
          .reset_index())
    cuota = gt.groupby(["topic", "FUERZA_ETIQ"]).gamma.sum().unstack(fill_value=0)
    cuota = cuota.div(cuota.sum(axis=1), axis=0).add_prefix("cuota_").reset_index()
    guardar_tabla(fz.merge(cuota, on="topic").merge(leer_obj("etiquetas"), on="topic"), "t06_fuerza_de_las_narrativas")

# ---- 4. Nivel de actor (incluye ausencia) -----------------------------------------------------
fa = fuerza_doc.drop(columns="GOVERNANCE").merge(doc_actor[["CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]], on="CODE")
fuerza_actor = (fa.groupby(["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"], dropna=False)
                .agg(n_docs=("CODE", "nunique"), FUERZA_MAX=("NIVEL_FUERZA", "max"), FUERZA_MEDIANA=("NIVEL_FUERZA", "median"),
                     PROP_OBLIGATORIO=("NIVEL_FUERZA", lambda s: np.mean(s >= 2)), INDICE_DEONTICO=("INDICE_DEONTICO", "mean"),
                     FUERTE_1000=("FUERTE_1000", "mean"), DEBIL_1000=("DEBIL_1000", "mean"),
                     DENS_DEONTICA=("DENS_DEONTICA", "mean"), PROP_TRADUCIDO=("TRADUCIDO", lambda s: pd.to_numeric(s, errors="coerce").mean()))
                .reset_index())
aus = actores_estado[actores_estado.AUSENCIA][["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]].assign(n_docs=0, FUERZA_MAX=-1)
fuerza_actor = pd.concat([fuerza_actor, aus], ignore_index=True)
fuerza_actor["FUERZA_ETIQ"] = pd.Categorical(fuerza_actor.FUERZA_MAX.map(ETIQ_FUERZA), categories=list(ETIQ_FUERZA.values()))
fuerza_actor = fuerza_actor.rename(columns={"PROP_OBLIGATORIO": "PROP_VINCULANTE"})   # nombre usado en el Paso 11

# ---- 5. Comparación entre niveles ------------------------------------------------------------
t = fuerza_actor.groupby(["GOVERNANCE", "FUERZA_ETIQ"], observed=True).size().reset_index(name="n")
t["prop"] = t.n / t.groupby("GOVERNANCE").n.transform("sum")
guardar_tabla(t, "t06_RQ1b_capa1_por_nivel")
con_doc = fuerza_actor[fuerza_actor.n_docs > 0]
guardar_tabla(pd.DataFrame([
    {**chi_sim(fuerza_actor.GOVERNANCE, fuerza_actor.FUERZA_ETIQ.astype(str)).to_dict(), "prueba": "Capa 1 x nivel (incluye ausencia)"},
    {**kw_niveles(con_doc.FUERZA_MAX, con_doc.GOVERNANCE).to_dict(), "prueba": "Capa 1 ordinal (KW)"},
    {**kw_niveles(con_doc.PROP_VINCULANTE, con_doc.GOVERNANCE).to_dict(), "prueba": "Proporción de instrumentos obligatorios (KW)"},
    {**kw_niveles(con_doc.INDICE_DEONTICO, con_doc.GOVERNANCE).to_dict(), "prueba": "Índice deóntico (KW)"},
    {**kw_niveles(con_doc.FUERTE_1000, con_doc.GOVERNANCE).to_dict(), "prueba": "Obligaciones fuertes por 1.000 palabras (KW)"},
    {**kw_niveles(con_doc.DEBIL_1000, con_doc.GOVERNANCE).to_dict(), "prueba": "Obligaciones débiles por 1.000 palabras (KW)"},
]), "t06_RQ1b_pruebas_niveles")

# Control por idioma: índice deóntico en originales en inglés vs traducidos, por nivel
tr = fuerza_doc[fuerza_doc.TRADUCIDO.notna()].copy()
if len(tr):
    tr["IDIOMA"] = np.where(tr.TRADUCIDO.astype(bool), "Traducido", "Original")
    guardar_tabla(resumen_por(tr, "GOVERNANCE", lambda g: comparar_2g(g.INDICE_DEONTICO, g.IDIOMA, "Traducido", "Original")),
                  "t06_indice_por_traduccion")

tab = fuerza_actor.groupby(["GOVERNANCE", "FUERZA_ETIQ"], observed=False).size().unstack(fill_value=0)
tab = tab.div(tab.sum(axis=1), axis=0)
fig, ax = plt.subplots()
izq = np.zeros(len(tab))
for i, c in enumerate(tab.columns):
    ax.barh(tab.index, tab[c], left=izq, color=plt.get_cmap("seguile_azul")(0.2 + 0.75 * i / max(1, len(tab.columns) - 1)), label=c)
    izq += tab[c].values
ax.set_xlabel("Share of actors"); ax.legend(title="Strongest instrument", bbox_to_anchor=(1, 1), fontsize=7)
guardar_fig(fig, "f06_RQ1b_fuerza_por_nivel", 9, 4)

guardar_obj(fuerza_doc, "fuerza_doc"); guardar_obj(fuerza_actor, "fuerza_actor")
print("Paso 6 (RQ1b) listo.")
```

## Paso 7. RQ1c: cambio tras la IA generativa (método 3.14)
Pre-GenAI (< 2023) vs Post-GenAI (≥ 2023), con el **documento** como unidad, dentro de cada nivel:
γ de cada tópico e índice deóntico (Wilcoxon, r biserial, BH entre tópicos). Orden temporal
descriptivo: primer año y año mediano en que cada tópico es relevante (γ ≥ 0,20) en cada nivel.

**[REVISIÓN]** La comparación pareada dentro de actor viene de los scripts R; el método la limita
(diseño transversal, limitación 2), así que aquí se reporta solo como sensibilidad.

```python
PASO = "Step7"
docs_meta = leer_obj("docs_meta"); doc_actor = leer_obj("doc_actor")
gamma_td = leer_obj("gamma_td"); etiquetas = leer_obj("etiquetas"); fuerza_doc = leer_obj("fuerza_doc")

docs_t = docs_meta[docs_meta.YEAR.notna()].copy()
docs_t["PERIODO"] = np.where(docs_t.YEAR >= ANIO_CORTE_GENAI, "Post", "Pre")
guardar_tabla(docs_t.groupby(["GOVERNANCE", "PERIODO"]).size().reset_index(name="n"), "t07_documentos_por_periodo")
registrar(PASO, "Documentos sin año (excluidos de RQ1c)", int(docs_meta.YEAR.isna().sum()))

# ---- 1. Tópicos: Post vs Pre dentro de cada nivel -------------------------------------------
gp = gamma_td.merge(docs_t[["CODE", "GOVERNANCE", "PERIODO"]], on="CODE")
temp_top = resumen_por(gp, ["GOVERNANCE", "topic"], lambda g: comparar_2g(g.gamma, g.PERIODO, "Post", "Pre"))
temp_top["p_BH"] = temp_top.groupby("GOVERNANCE").p_valor.transform(p_bh)
guardar_tabla(temp_top.merge(etiquetas, on="topic"), "t07_RQ1c_topicos_post_vs_pre")   # _1 = Post, _2 = Pre

# ---- 2. Fuerza normativa: Post vs Pre ----------------------------------------------------------
fp = fuerza_doc.drop(columns="GOVERNANCE").merge(docs_t[["CODE", "GOVERNANCE", "PERIODO"]], on="CODE")
fp = fp.melt(id_vars=["CODE", "GOVERNANCE", "PERIODO"], value_vars=["INDICE_DEONTICO", "FUERTE_1000", "DEBIL_1000", "NIVEL_FUERZA"],
             var_name="INDICADOR", value_name="valor")
temp_fza = resumen_por(fp, ["GOVERNANCE", "INDICADOR"], lambda g: comparar_2g(g.valor, g.PERIODO, "Post", "Pre"))
temp_fza["p_BH"] = temp_fza.groupby("GOVERNANCE").p_valor.transform(p_bh)
guardar_tabla(temp_fza, "t07_RQ1c_fuerza_post_vs_pre")

# ---- 3. Sensibilidad intra-actor (Wilcoxon pareado; solo sensibilidad) -----------------------
par = (gamma_td.merge(docs_t[["CODE", "PERIODO"]], on="CODE").merge(doc_actor[["CODE", "GOVERNANCE", "ACTOR"]], on="CODE")
       .groupby(["GOVERNANCE", "ACTOR", "topic", "PERIODO"]).gamma.mean().unstack().dropna(subset=["Pre", "Post"]).reset_index()
       if {"Pre", "Post"} <= set(docs_t.PERIODO) else pd.DataFrame())
if len(par):
    def pareado(g):
        p = stats.wilcoxon(g.Post, g.Pre, correction=True, method="approx").pvalue if len(g) >= 5 and (g.Post - g.Pre).abs().sum() > 0 else np.nan
        return pd.Series({"n_actores": len(g), "dif_mediana": np.median(g.Post - g.Pre), "p_valor": p})
    pt = resumen_por(par, ["GOVERNANCE", "topic"], pareado)
    pt["p_BH"] = pt.groupby("GOVERNANCE").p_valor.transform(p_bh)
    guardar_tabla(pt.merge(etiquetas, on="topic"), "t07_RQ1c_pareado_intra_actor_SENSIBILIDAD")

# ---- 4. Orden temporal de cada tópico por nivel (descriptivo) ----------------------------------
gy = gamma_td.merge(docs_t[["CODE", "GOVERNANCE", "YEAR"]], on="CODE")
primera = (gy[gy.gamma >= UMBRAL_MULTI].groupby(["GOVERNANCE", "topic"])
           .agg(anio_primero=("YEAR", "min"), anio_mediano=("YEAR", "median"), n_docs=("CODE", "size")).reset_index())
guardar_tabla(primera.merge(etiquetas, on="topic"), "t07_RQ1c_primera_aparicion")

serie = gy.assign(rel=gy.gamma >= UMBRAL_MULTI).groupby(["GOVERNANCE", "YEAR", "topic"]).rel.mean().reset_index()
K = gamma_td.topic.nunique(); ncol = 4; nfil = math.ceil(K / ncol)
fig, axs = plt.subplots(nfil, ncol, squeeze=False, sharey=True)
estilos = dict(zip(sorted(serie.GOVERNANCE.unique()), ["-", "--", ":"]))
for t, ax in zip(range(1, K + 1), axs.ravel()):
    for gov, d in serie[serie.topic == t].groupby("GOVERNANCE"):
        ax.plot(d.YEAR.astype(int), d.rel, linestyle="-", color=COLOR_GOB.get(gov, "#52514e"), label=gov)
    ax.axvline(ANIO_CORTE_GENAI - 0.5, color="grey"); ax.set_title(etiquetas.set_index("topic").TOPIC_LABEL[t][:40], fontsize=7)
    ax.tick_params(labelsize=6)
for ax in axs.ravel()[K:]:
    ax.axis("off")
axs[0, 0].set_ylabel(f"Share of documents with γ ≥ {UMBRAL_MULTI}"); axs[0, 0].legend(fontsize=6)
guardar_fig(fig, "f07_RQ1c_topicos_por_anio", 12, 3 * nfil)
print("Paso 7 (RQ1c) listo.")
```

## Paso 8. Taxonomía deductiva 5D, triangulación y alineamiento normativo (método 3.15)
Diccionario de unigramas, bigramas y trigramas por dimensión y subdimensión
(`diccionario_taxonomia_5D.csv`: DIMENSION, SUBDIMENSION, TERMINO, EXCLUIR; DIMENSION_NOMBRE opcional).
Textos y términos se normalizan igual (minúsculas, guiones y barras a espacios, sin *stop words* Snowball,
lematización); se cuenta de la expresión más larga a la más corta y un término en *k* subdimensiones
aporta 1/*k* a cada una. Medidas: densidad por 10.000 palabras y cuota de coincidencias.
Produce: `dim_doc`, `sub_doc`, `dim_actor`, `sub_actor`, `alineamiento_normativo`.

```python
PASO = "Step8"
f_dicc = DIR_PROYECTO / "diccionario_taxonomia_5D.csv"
if not f_dicc.exists():
    pd.DataFrame(columns=["DIMENSION", "SUBDIMENSION", "TERMINO", "EXCLUIR", "DIMENSION_NOMBRE"]).to_csv(
        DIR_PROYECTO / "diccionario_taxonomia_5D_PLANTILLA.csv", index=False)
    raise FileNotFoundError("Falta diccionario_taxonomia_5D.csv (se generó la plantilla). Súbelo al repositorio.")

corpus = leer_obj("corpus"); doc_actor = leer_obj("doc_actor")
gamma_td = leer_obj("gamma_td"); etiquetas = leer_obj("etiquetas")


def normalizar(txt):
    t = str(txt).lower()
    t = re.sub(r"['’‘]s\b", "", t); t = re.sub(r"['’‘]", "", t)
    t = re.sub(r"[\W_]+", " ", t).split()
    return " ".join(lematizar([w for w in t if w not in STOP_SNOWBALL]))


# ---- 1. Diccionario ---------------------------------------------------------------------------
dicc = pd.read_csv(f_dicc, dtype=str)
if "DIMENSION_NOMBRE" not in dicc:
    dicc["DIMENSION_NOMBRE"] = dicc.DIMENSION
dicc["EXCLUIR"] = dicc.EXCLUIR.fillna("FALSE").str.upper().isin(["TRUE", "T", "1", "SI", "SÍ", "YES"])
dicc["PATRON"] = dicc.TERMINO.map(normalizar)
dicc["N_PAL"] = dicc.PATRON.str.count(" ") + 1
dicc_activo = (dicc[~dicc.EXCLUIR & (dicc.PATRON != "")].drop_duplicates(["DIMENSION", "SUBDIMENSION", "PATRON"]).copy())
dicc_activo["PESO"] = 1 / dicc_activo.groupby("PATRON").PATRON.transform("size")   # término en k subdimensiones: 1/k
patrones = (dicc_activo.drop_duplicates("PATRON")[["PATRON", "N_PAL"]]
            .assign(L=lambda d: d.PATRON.str.len()).sort_values(["N_PAL", "L"], ascending=False))
registrar(PASO, "Términos activos del diccionario", len(patrones))
registrar(PASO, "Términos excluidos del diccionario", int(dicc.EXCLUIR.sum()))

# ---- 2. Conteo (frase más larga primero) --------------------------------------------------------
pats = [" " + p.replace(" ", "  ") + " " for p in patrones.PATRON]


def contar_doc(flujo):
    s = " " + flujo.replace(" ", "  ") + " "
    n = np.zeros(len(pats), dtype=int)
    for i, p in enumerate(pats):
        c = s.count(p)
        if c:
            n[i] = c; s = s.replace(p, " # ")
    return n


flujos = corpus[["CODE", "N_PALABRAS", "text"]].rename(columns={"N_PALABRAS": "N_PALABRAS_DOC"})
res = np.vstack([contar_doc(normalizar(t)) for t in flujos.text])
conteo_terminos = pd.DataFrame(res, index=flujos.CODE, columns=patrones.PATRON.values).stack().rename("n").reset_index()
conteo_terminos.columns = ["CODE", "PATRON", "n"]
conteo_terminos = conteo_terminos[conteo_terminos.n > 0]
hits = conteo_terminos.merge(dicc_activo[["DIMENSION", "SUBDIMENSION", "PATRON", "PESO"]], on="PATRON")
hits["n_pond"] = hits.n * hits.PESO


def puntuar(claves):
    base = dicc_activo[claves].drop_duplicates().merge(flujos[["CODE", "N_PALABRAS_DOC"]], how="cross")
    h = hits.groupby(["CODE"] + claves).n_pond.sum().rename("hits").reset_index()
    d = base.merge(h, on=["CODE"] + claves, how="left").fillna({"hits": 0})
    d["DENSIDAD"] = d.hits / d.N_PALABRAS_DOC * POR_DIEZ_MIL
    tot = d.groupby("CODE").hits.transform("sum")
    d["CUOTA"] = np.where(tot > 0, d.hits / tot.where(tot > 0), np.nan)
    return d


dim_doc = puntuar(["DIMENSION"]); sub_doc = puntuar(["DIMENSION", "SUBDIMENSION"])

# ---- 3. Cobertura [D7] --------------------------------------------------------------------------
cob = conteo_terminos.groupby("PATRON").agg(frecuencia=("n", "sum"), n_docs=("CODE", "nunique")).reset_index()
cob_t = dicc.merge(cob, on="PATRON", how="left").fillna({"frecuencia": 0, "n_docs": 0})
guardar_tabla(cob_t, "t08_cobertura_terminos")
guardar_tabla(cob_t[~cob_t.EXCLUIR & (cob_t.frecuencia == 0)], "t08_terminos_nunca_encontrados")
guardar_tabla(dim_doc.groupby("DIMENSION").agg(docs_con_hits=("hits", lambda s: int((s > 0).sum())),
                                               prop_docs=("hits", lambda s: (s > 0).mean()),
                                               densidad_media=("DENSIDAD", "mean")).reset_index(), "t08_cobertura_dimensiones")
registrar(PASO, "Documentos sin coincidencias del diccionario", int((dim_doc.groupby("CODE").hits.sum() == 0).sum()))

# ---- 4. Triangulación LDA x taxonomía (RQ1a) -----------------------------------------------------
tg = gamma_td.merge(dim_doc[["CODE", "DIMENSION", "DENSIDAD"]], on="CODE")
triang = resumen_por(tg, ["topic", "DIMENSION"],
                     lambda g: pd.Series({"rho": stats.spearmanr(g.gamma, g.DENSIDAD).statistic if g.DENSIDAD.nunique() > 1 else np.nan}))
triang = triang.merge(etiquetas, on="topic")
guardar_tabla(triang, "t08_triangulacion")
principal = triang.sort_values("rho", ascending=False).groupby("topic").head(1).copy()
principal["DIMENSION_PRINCIPAL"] = np.where(principal.rho >= RHO_MIN_TRIANG, principal.DIMENSION, "Sin correspondencia")
guardar_tabla(principal.sort_values("topic"), "t08_topico_a_dimension")
m = triang.pivot(index="TOPIC_LABEL", columns="DIMENSION", values="rho")
fig, ax = plt.subplots()
im = ax.imshow(m.values, cmap="seguile_div", aspect="auto", vmin=-1, vmax=1)
ax.set_xticks(range(m.shape[1])); ax.set_xticklabels(m.columns); ax.set_yticks(range(m.shape[0])); ax.set_yticklabels(m.index, fontsize=7)
for i in range(m.shape[0]):
    for j in range(m.shape[1]):
        ax.text(j, i, f"{m.values[i, j]:.2f}", ha="center", va="center", fontsize=7, color="white" if m.values[i, j] > .5 else "black")
fig.colorbar(im, label="Spearman ρ"); guardar_fig(fig, "f08_triangulacion", 9, 1.5 + 0.5 * len(m))

# ---- 5. Perfil por actor (misma agregación que los tópicos: media por documento) ----------------
CLAVES_ACTOR = ["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]


def agregar_actor(df, claves):
    d = (df.merge(doc_actor, on="CODE").groupby(CLAVES_ACTOR + claves, dropna=False)
         .agg(DENSIDAD=("DENSIDAD", "mean"), CUOTA=("CUOTA", "mean")).reset_index())
    d["CUOTA"] = d.CUOTA / d.groupby(["GOV_CODE", "ACTOR"]).CUOTA.transform("sum")
    return d


dim_actor = agregar_actor(dim_doc, ["DIMENSION"]); sub_actor = agregar_actor(sub_doc, ["DIMENSION", "SUBDIMENSION"])

kwd = resumen_por(dim_actor, "DIMENSION", lambda g: kw_niveles(g.DENSIDAD, g.GOVERNANCE)); kwd["p_BH"] = p_bh(kwd.p_valor)
kws = resumen_por(sub_actor, ["DIMENSION", "SUBDIMENSION"], lambda g: kw_niveles(g.DENSIDAD, g.GOVERNANCE)); kws["p_BH"] = p_bh(kws.p_valor)
guardar_tabla(kwd, "t08_taxonomia_por_nivel"); guardar_tabla(kws, "t08_subdimensiones_por_nivel")

# ---- 6. Alineamiento normativo con el referente global (RQ2) -------------------------------------
sin_perfil = dim_actor.groupby(["GOV_CODE", "ACTOR"]).CUOTA.apply(lambda s: s.isna().all()).reset_index(name="sin")
guardar_tabla(sin_perfil[sin_perfil.sin], "t08_actores_sin_perfil_normativo")
alineamiento_normativo = (distancia_referente(dim_actor[dim_actor.CUOTA.notna()], "DIMENSION", "CUOTA")
                          .rename(columns={"HELLINGER": "HELLINGER_NORMATIVO"})
                          .merge(dim_actor[CLAVES_ACTOR].drop_duplicates(), on=["GOV_CODE", "ACTOR"]))
alineamiento_normativo["ES_REFERENTE"] = (alineamiento_normativo.GOV_CODE == "G") & (alineamiento_normativo.BLOC == "Global")
filas = []
an = alineamiento_normativo[~alineamiento_normativo.ES_REFERENTE]
filas.append({**kw_niveles(an.HELLINGER_NORMATIVO, an.GOVERNANCE).to_dict(), "distancia": "normativa (dimensiones)"})
if existe_obj("actor_metricas"):
    am = leer_obj("actor_metricas"); am = am[~am.ES_REFERENTE]
    filas.append({**kw_niveles(am.HELLINGER_GLOBAL, am.GOVERNANCE).to_dict(), "distancia": "temática (tópicos)"})
al = pd.DataFrame(filas)
guardar_tabla(al, "t08_alineamiento_por_nivel")     # método 3.15: sin las organizaciones de referencia
guardar_tabla(alineamiento_normativo.groupby("GOVERNANCE").HELLINGER_NORMATIVO.describe().reset_index(),
              "t08_alineamiento_normativo_descriptivo")

for nombre, obj in {"dim_doc": dim_doc, "sub_doc": sub_doc, "dim_actor": dim_actor, "sub_actor": sub_actor,
                    "alineamiento_normativo": alineamiento_normativo}.items():
    guardar_obj(obj, nombre)
print("Paso 8 (taxonomía) listo.")
```

## Paso 9. RQ2: referencias explícitas a instrumentos globales (evidencia explícita)
**[REVISIÓN] El método v1.6 no tiene una sección para este paso**, aunque la 3.19 lo menciona

Requiere `diccionario_instrumentos.csv` (ID_INST, INSTRUMENTO, ACTOR_G, ANIO, PATRON [regex],
CODE_G [opcional, separados por ;]). Detecta menciones, excluye autocitas y menciones anacrónicas,
genera la muestra de validación de precisión **[C2]** y contrasta la cascada de referencias
universidad → país → organización con nulos por permutación.

```python
PASO = "Step9"
f_dicc = DIR_PROYECTO / "diccionario_instrumentos.csv"
if not f_dicc.exists():
    pd.DataFrame(columns=["ID_INST", "INSTRUMENTO", "ACTOR_G", "ANIO", "PATRON", "CODE_G"]).to_csv(
        DIR_PROYECTO / "diccionario_instrumentos_PLANTILLA.csv", index=False)
    print("[AVISO] Paso 9 omitido: falta diccionario_instrumentos.csv (se generó la plantilla).")
else:
    corpus = leer_obj("corpus"); doc_actor = leer_obj("doc_actor")
    inst = pd.read_csv(f_dicc, dtype={"ID_INST": str, "CODE_G": str})
    inst["ACTOR_G"] = canonizar_org(inst.ACTOR_G).values; inst["CODE_G"] = inst.CODE_G.fillna("")
    inst["ANIO"] = pd.to_numeric(inst.ANIO, errors="coerce")

    # ---- 1. Detección de menciones ------------------------------------------------------------
    filas = []
    for it in inst.itertuples():
        rx = re.compile(it.PATRON, flags=re.I)
        for code, txt, year in zip(corpus.CODE, corpus.text, corpus.YEAR):
            for m in rx.finditer(txt):
                filas.append({"CODE": code, "ID_INST": it.ID_INST, "inicio": m.start() + 1, "fin": m.end(),
                              "YEAR": year, "CONTEXTO": re.sub(r"\s+", " ", txt[max(0, m.start() - 150): m.end() + 150]).strip()})
    menciones = pd.DataFrame(filas, columns=["CODE", "ID_INST", "inicio", "fin", "YEAR", "CONTEXTO"]).merge(
        inst[["ID_INST", "ACTOR_G", "ANIO", "CODE_G"]], on="ID_INST", how="left")
    menciones["TEXTO_PROPIO"] = [c in [x.strip() for x in cg.split(";")] for c, cg in zip(menciones.CODE, menciones.CODE_G)]
    menciones["ANACRONICA"] = menciones.YEAR.notna() & menciones.ANIO.notna() & (menciones.YEAR < menciones.ANIO)
    auto = (menciones.merge(doc_actor[["CODE", "GOV_CODE", "ACTOR"]], on="CODE")
            .query("GOV_CODE == 'G' and ACTOR == ACTOR_G")[["CODE", "ID_INST"]].drop_duplicates())
    menciones = (menciones.merge(auto.assign(_auto=1), on=["CODE", "ID_INST"], how="left")
                 .query("_auto.isna() and not TEXTO_PROPIO", engine="python").drop(columns="_auto"))
    registrar(PASO, "Menciones anacrónicas", int(menciones.ANACRONICA.sum()))
    if EXCLUIR_ANACRONICAS:
        menciones = menciones[~menciones.ANACRONICA]

    # ---- 2. Validación de precisión [C2] --------------------------------------------------------
    (menciones.sample(n=min(N_MUESTRA_CITAS, len(menciones)), random_state=SEMILLA)
     [["CODE", "ID_INST", "inicio", "CONTEXTO"]].assign(CORRECTO="")
     .to_csv(DIR_PROYECTO / "muestra_validacion_citas_PLANTILLA.csv", index=False))
    f_val = DIR_PROYECTO / "validacion_citas.csv"      # misma estructura, CORRECTO = 1/0
    if f_val.exists():
        val = pd.read_csv(f_val, dtype={"ID_INST": str}); val = val[val.CORRECTO.notna()]
        prec = val.groupby("ID_INST").CORRECTO.agg(n_validadas="size", precision="mean").reset_index()
        guardar_tabla(pd.concat([prec, pd.DataFrame([{"ID_INST": "TOTAL", "n_validadas": len(val),
                                                      "precision": val.CORRECTO.mean()}])]), "t09_precision_citas")
        excluir = prec[(prec.n_validadas >= 5) & (prec.precision < UMBRAL_PRECISION)].ID_INST
        menciones = menciones[~menciones.ID_INST.isin(excluir)]
        registrar(PASO, "Instrumentos excluidos por precisión", len(excluir))
    else:
        print("[AVISO] [C2] Falta validacion_citas.csv: precisión no estimada.")

    # ---- 3. Matriz actor x organización citada ---------------------------------------------------
    ma = menciones.merge(doc_actor[["CODE", "GOV_CODE", "ACTOR"]], on="CODE")
    cita_actor_org = (ma.groupby(["GOV_CODE", "ACTOR", "ACTOR_G"])
                      .agg(n_menciones=("CODE", "size"), n_docs=("CODE", "nunique")).reset_index())
    guardar_tabla(cita_actor_org, "t09_citas_actor_org")
    guardar_tabla(ma.groupby(["ID_INST", "GOV_CODE"]).size().unstack(fill_value=0).reset_index()
                  .merge(inst[["ID_INST", "INSTRUMENTO", "ACTOR_G"]], on="ID_INST", how="left"), "t09_citas_por_instrumento")

    # ---- 4. Cascada de referencias y nulos -------------------------------------------------------
    paises = (doc_actor[(doc_actor.GOV_CODE == "P") & doc_actor.ISO2.notna()].drop_duplicates("ISO2")
              [["ISO2", "ACTOR"]].rename(columns={"ACTOR": "ACTOR_P"}))
    univ = (doc_actor[doc_actor.GOV_CODE == "U"].drop_duplicates("ACTOR")[["ACTOR", "ISO2", "REGION"]]
            .merge(paises, on="ISO2"))
    orgs = sorted(cita_actor_org.ACTOR_G.unique())
    if len(univ) and len(orgs):
        CU = pd.DataFrame(0, index=univ.ACTOR, columns=orgs)
        for r in cita_actor_org[(cita_actor_org.GOV_CODE == "U") & cita_actor_org.ACTOR.isin(univ.ACTOR)].itertuples():
            CU.loc[r.ACTOR, r.ACTOR_G] = 1
        CP = pd.DataFrame(0, index=paises.ACTOR_P, columns=orgs)
        for r in cita_actor_org[(cita_actor_org.GOV_CODE == "P") & cita_actor_org.ACTOR.isin(paises.ACTOR_P)].itertuples():
            CP.loc[r.ACTOR, r.ACTOR_G] = 1
        CUv = CU.values

        def estad(asig):
            P = CP.loc[asig].values
            return {"cascada": np.mean((CUv == 1) & (P == 1)), "bypass": np.mean((CUv == 1) & (P == 0)),
                    "relevo": np.mean((CUv == 0) & (P == 1)),
                    "lift": CUv[P == 1].mean() - CUv[P == 0].mean() if (P == 1).any() and (P == 0).any() else np.nan}

        def perm_region():
            p = univ.ACTOR_P.values.copy()
            for reg in univ.REGION.unique():
                i = np.where(univ.REGION.values == reg)[0]; p[i] = p[rng.permutation(i)]
            return p

        obs = estad(univ.ACTOR_P.values)
        nulos = {"global": pd.DataFrame([estad(rng.permutation(univ.ACTOR_P.values)) for _ in range(N_PERMUT)]),
                 "region": pd.DataFrame([estad(perm_region()) for _ in range(N_PERMUT)])}
        guardar_tabla(pd.DataFrame([{"NULO": nn, "ESTADISTICO": v, "observado": obs[v], "media_nula": nl[v].mean(),
                                     "p_superior": (1 + np.sum(nl[v] >= obs[v])) / (N_PERMUT + 1)}
                                    for nn, nl in nulos.items() for v in obs]), "t09_RQ2_cascada_referencias")
    guardar_obj(cita_actor_org, "cita_actor_org")
    print("Paso 9 (RQ2, referencias) listo.")
```

## Paso 10. RQ2 y RQ3b: red multinivel de alineamiento (método 3.17)
Siguiendo a Lazega y Snijders (2016) y Wang et al. (2016): redes dentro de cada nivel + redes meso
entre niveles, en dos relaciones (temática y normativa) y su multiplex.

1. **Lazos.** Matriz bipartita actor × artefacto (tópico si γ medio > 1/k; subdimensión si densidad >
   mediana entre actores), proyección sobre actores y *backbone* SDSM (Neal, 2014) con α = 0,05.
   El SDSM está programado aquí (modelo de configuración bipartito + distribución binomial de Poisson),
   porque el paquete `backbone` no existe en Python.
2. **Descripción** por bloque: densidad, centralización de grado, transitividad, componentes; por actor:
   grado dentro del nivel, grado entre niveles, grado hacia organizaciones, centralidad de vector propio.
3. **Cascada, relevo y bypass** universidad–país–organización con nulos.
4. **Espacio conjunto** (análisis de correspondencias; Zhu et al., 2016) y **posiciones**
   (equivalencia estructural, primero por nivel y luego conjunta; Žiberna y Lazega, 2016).
5. **Balance posicional Norte/Sur (RQ3b)** y homofilia de bloque contra permutación de etiquetas.
6. **ERGM** por bloques (opcional, en R; Paso 10b).

Los lazos se infieren de similitud textual: se habla de **alineamiento**, no de influencia ni difusión.

```python
from scipy.cluster.hierarchy import linkage, fcluster
import networkx as nx
PASO = "Step10"
actor_topico = leer_obj("actor_topico"); actores_estado = leer_obj("actores_estado")
sub_actor = leer_obj("sub_actor") if existe_obj("sub_actor") else None
K = actor_topico.topic.nunique()

# ---- 1. Nodos y afiliación observada universidad -> país -----------------------------------
nodos = (actor_topico[["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]].drop_duplicates(["GOV_CODE", "ACTOR"])
         .merge(actores_estado[["GOV_CODE", "ACTOR", "ISO2", "REGION", "ANIO_MEDIANO", "ORG_CATEGORIA"]],
                on=["GOV_CODE", "ACTOR"], how="left"))
nodos["ID"] = nodos.GOV_CODE + "::" + nodos.ACTOR.astype(str)
nodos["NIVEL"] = nodos.GOV_CODE
nodos["BLOC"] = nodos.BLOC.fillna("Unknown"); nodos["BLOC_SENS"] = nodos.BLOC_SENS.fillna("Unknown")
nodos = nodos.sort_values(["NIVEL", "ACTOR"], key=lambda s: s.map({"G": 0, "P": 1, "U": 2}) if s.name == "NIVEL" else s).reset_index(drop=True)
paises = nodos[(nodos.NIVEL == "P") & nodos.ISO2.notna()].drop_duplicates("ISO2").set_index("ISO2").ID
nodos["PAIS_ID"] = np.where(nodos.NIVEL == "U", nodos.ISO2.map(paises), None)
ids = nodos.ID.tolist(); pos = {i: k for k, i in enumerate(ids)}; nivel = nodos.NIVEL.to_numpy(dtype=object)
registrar(PASO, "Universidades con su país en la red", int(pd.notna(nodos.PAIS_ID).sum()))
registrar(PASO, "Universidades sin documento nacional de su país (excluidas de la cascada)",
          int(((nodos.NIVEL == "U") & pd.isna(nodos.PAIS_ID)).sum()))

# ---- 2. Matrices bipartitas actor x artefacto ------------------------------------------------
at = actor_topico.assign(ID=actor_topico.GOV_CODE + "::" + actor_topico.ACTOR.astype(str))
B_tema = (at.assign(v=(at.gamma > 1 / K).astype(int)).pivot_table(index="ID", columns="topic", values="v", fill_value=0)
          .reindex(ids, fill_value=0))
B_tema.columns = [f"T{c}" for c in B_tema.columns]
if sub_actor is not None:
    sa = sub_actor[sub_actor[VAR_LAZO_NORMATIVO].notna()].copy()
    sa["ID"] = sa.GOV_CODE + "::" + sa.ACTOR.astype(str); sa["SUB"] = sa.DIMENSION + "|" + sa.SUBDIMENSION
    med = sa.groupby("SUB")[VAR_LAZO_NORMATIVO].transform("median")
    sa["v"] = ((sa[VAR_LAZO_NORMATIVO] > med) & (sa[VAR_LAZO_NORMATIVO] > 0)).astype(int)
    B_norm = sa.pivot_table(index="ID", columns="SUB", values="v", fill_value=0).reindex(ids, fill_value=0)
    registrar(PASO, "Actores sin perfil normativo (sin lazos normativos)", int((B_norm.sum(axis=1) == 0).sum()))
else:
    B_norm = None
    print("[AVISO] Falta sub_actor (Paso 8): solo se construye la relación temática.")

# ---- 3. Backbone SDSM (funciones en el Paso 1, sección 7) ------------------------------------
A_tema = extraer_backbone(B_tema.values)
REL = {"tematica": {"B": B_tema.values, "A": A_tema}}
if B_norm is not None:
    A_norm = extraer_backbone(B_norm.values)
    REL["normativa"] = {"B": B_norm.values, "A": A_norm}
    A_multi = ((A_tema == 1) & (A_norm == 1)).astype(int)
    MATS = {"tematica": A_tema, "normativa": A_norm, "multiplex": A_multi}
else:
    A_norm = A_multi = None; MATS = {"tematica": A_tema}
for r, A in MATS.items():
    registrar(PASO, f"Lazos en el backbone ({r})", int(A.sum() // 2))
# [REVISIÓN] Diagnóstico de potencia del SDSM: con pocos artefactos (p. ej., k = 8 tópicos) ningún par de
# actores puede alcanzar p < α, y la red queda vacía aunque haya estructura (ver la nota al final del paso).
for r, d in REL.items():
    s = sdsm_pvalores(d["B"])
    pmin = float(s["p"].min()) if s is not None else np.nan
    registrar(PASO, f"p mínimo SDSM ({r}; {d['B'].shape[1]} artefactos)", round(pmin, 4))
    if not (pmin < ALFA_BACKBONE):
        print(f"[AVISO] Relación {r}: el p-valor SDSM más pequeño posible es {pmin:.3f} (≥ α = {ALFA_BACKBONE}). "
              f"Con {d['B'].shape[1]} artefactos la red {r} no puede tener lazos: es un límite del diseño, no un resultado.")

# ---- 4. Descripción por bloque y por actor (método 3.17) ---------------------------------------
import networkx as nx
NIVELES = ["G", "P", "U"]
idx_niv = {lv: np.where(nivel == lv)[0] for lv in NIVELES}


def centralizacion_grado(A):
    n = A.shape[0]
    if n < 3:
        return np.nan
    d = A.sum(axis=1)
    return float(np.sum(d.max() - d) / ((n - 1) * (n - 2)))


def describir(A, nombre):
    filas = []
    for a, b in itertools.combinations_with_replacement(NIVELES, 2):
        i, j = idx_niv[a], idx_niv[b]
        if a == b:
            sub = A[np.ix_(i, i)]; n = len(i)
            g = nx.from_numpy_array(sub)
            filas.append({"bloque": f"{a}-{a}", "n_nodos": n, "lazos": int(sub.sum() // 2),
                          "densidad": sub.sum() / (n * (n - 1)) if n > 1 else np.nan,
                          "centralizacion_grado": centralizacion_grado(sub),
                          "transitividad": nx.transitivity(g) if n > 2 else np.nan,
                          "componentes": nx.number_connected_components(g) if n else np.nan})
        else:
            sub = A[np.ix_(i, j)]
            filas.append({"bloque": f"{a}-{b}", "n_nodos": len(i) + len(j), "lazos": int(sub.sum()),
                          "densidad": sub.mean() if sub.size else np.nan})
    g = nx.from_numpy_array(A); n = A.shape[0]
    filas.append({"bloque": "Red completa", "n_nodos": n, "lazos": int(A.sum() // 2), "densidad": A.sum() / (n * (n - 1)),
                  "centralizacion_grado": centralizacion_grado(A), "transitividad": nx.transitivity(g),
                  "componentes": nx.number_connected_components(g)})
    return pd.DataFrame(filas).assign(RELACION=nombre)


guardar_tabla(pd.concat([describir(A, r) for r, A in MATS.items()]), "t10_RQ2_descripcion_bloques")

g_norte = np.where((nivel == "G") & (nodos.BLOC.to_numpy(dtype=object) == "North"))[0]
g_sur = np.where((nivel == "G") & (nodos.BLOC.to_numpy(dtype=object) == "South"))[0]
iG = idx_niv["G"]; nG = len(iG)


def metricas_nodo(A, r):
    g = nx.from_numpy_array(A)
    try:
        ev = nx.eigenvector_centrality_numpy(g) if A.sum() else {i: 0.0 for i in range(len(A))}
        ev = np.abs(np.array([ev[i] for i in range(len(A))])); ev = ev / ev.max() if ev.max() > 0 else ev
    except Exception:
        ev = np.full(len(A), np.nan)
    comp = {n: k for k, c in enumerate(sorted(nx.connected_components(g), key=len, reverse=True)) for n in c}
    mismo = nivel[:, None] == nivel[None, :]
    den_g = np.where(nivel == "G", nG - 1, nG)
    return pd.DataFrame({"ID": ids, "RELACION": r, "GRADO": A.sum(axis=1), "GRADO_DENTRO_NIVEL": (A * mismo).sum(axis=1),
                         "GRADO_ENTRE_NIVELES": (A * ~mismo).sum(axis=1), "GRADO_A_ORGS": A[:, iG].sum(axis=1),
                         "PROP_LAZOS_G": A[:, iG].sum(axis=1) / np.maximum(den_g, 1),
                         "PROP_G_NORTE": A[:, g_norte].sum(axis=1) / len(g_norte) if len(g_norte) else np.nan,
                         "PROP_G_SUR": A[:, g_sur].sum(axis=1) / len(g_sur) if len(g_sur) else np.nan,
                         "EIGEN": ev, "COMPONENTE": [comp[i] for i in range(len(A))]})


metricas_red = pd.concat([metricas_nodo(A, r) for r, A in MATS.items()]).merge(
    nodos[["ID", "GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]], on="ID")

# ---- 5. Cascada, relevo y bypass con nulos (método 3.17) -----------------------------------------
ctx = contexto_red(nodos)
U_idx, P_de_U = ctx["U_idx"], ctx["P_de_U"]

# Nulo del método: redes aleatorias que conservan el número de lazos de cada bloque
pares_bloque = {}
for a, b in itertools.combinations_with_replacement(NIVELES, 2):
    i, j = idx_niv[a], idx_niv[b]
    if a == b:
        t = np.triu_indices(len(i), 1); pares_bloque[(a, b)] = (i[t[0]], i[t[1]])
    else:
        ii, jj = np.meshgrid(i, j, indexing="ij"); pares_bloque[(a, b)] = (ii.ravel(), jj.ravel())


def aleatoria_por_bloques(A):
    R = np.zeros_like(A)
    for (pi, pj) in pares_bloque.values():
        m = int(A[pi, pj].sum())
        if m:
            k = rng.choice(len(pi), m, replace=False); R[pi[k], pj[k]] = R[pj[k], pi[k]] = 1
    return R


def contrastar(obs, nulo, etiqueta):
    nulo = pd.DataFrame(nulo)
    return pd.DataFrame([{"NULO": etiqueta, "ESTADISTICO": v, "observado": obs[v], "media_nula": nulo[v].mean(),
                          "de_nula": nulo[v].std(), "z": (obs[v] - nulo[v].mean()) / nulo[v].std() if nulo[v].std() > 0 else np.nan,
                          "p_superior": (1 + np.sum(nulo[v] >= obs[v] - 1e-12)) / (len(nulo) + 1),
                          "p_inferior": (1 + np.sum(nulo[v] <= obs[v] + 1e-12)) / (len(nulo) + 1)} for v in obs])


config, clases = [], []
for nombre, r in REL.items():
    A, B = r["A"], r["B"]
    obs = censo_configuraciones(A, ctx)
    if not obs:
        continue
    cl = clasificar_universidades(A, ctx)
    clases.append(cl.value_counts().rename("n_observado").reset_index().rename(columns={"index": "CLASE"}).assign(RELACION=nombre))
    print(f"Nulos para la relación {nombre}...")
    nulo_blq = [censo_configuraciones(aleatoria_por_bloques(A), ctx) for _ in range(N_NULO_BLOQUES)]
    u = umbral_sdsm(B)
    nulo_bip = [censo_configuraciones(backbone_con_umbral(curveball(B), u), ctx) for _ in range(N_NULO_BIPARTITO)] if u else []
    nulo_lab = []
    for _ in range(N_PERMUT):
        p = rng.permutation(len(A)); nulo_lab.append(censo_configuraciones(A[np.ix_(p, p)], ctx))
    partes = [contrastar(obs, nulo_blq, "lazos_por_bloque (método, principal)"),
              contrastar(obs, nulo_lab, "permutación_etiquetas (descriptivo)")]
    if nulo_bip:
        partes.insert(1, contrastar(obs, nulo_bip, "bipartito_curveball (scripts R)"))
    config.append(pd.concat(partes).assign(RELACION=nombre))
if config:
    guardar_tabla(pd.concat(config), "t10_RQ2_configuraciones_nulos")
    guardar_tabla(pd.concat(clases), "t10_RQ2_clasificacion_universidades")

# Transitividad observada frente al nulo por bloques (limitación 8: cierre inducido por la proyección)
guardar_tabla(pd.DataFrame([{"RELACION": r, "transitividad_obs": nx.transitivity(nx.from_numpy_array(A)),
                             "transitividad_nula_media": np.mean([nx.transitivity(nx.from_numpy_array(aleatoria_por_bloques(A)))
                                                                  for _ in range(100)])} for r, A in MATS.items()]),
              "t10_transitividad_vs_nulo")

# ---- 6. Convergencia citas-alineamiento (QAP; si existe el Paso 9) --------------------------------
if existe_obj("cita_actor_org"):
    cit = leer_obj("cita_actor_org")
    filas_pu = [i for i in range(len(ids)) if nivel[i] in ("P", "U")]
    C = np.zeros((len(filas_pu), nG))
    pf = {ids[i]: k for k, i in enumerate(filas_pu)}; pc = {ids[i]: k for k, i in enumerate(iG)}
    for r in cit.itertuples():
        f, c = f"{r.GOV_CODE}::{r.ACTOR}", f"G::{r.ACTOR_G}"
        if f in pf and c in pc:
            C[pf[f], pc[c]] = 1
    niv_f = nivel[filas_pu]; qap = []
    for nombre, A in MATS.items():
        Xm = A[np.ix_(filas_pu, iG)]
        obs = np.corrcoef(Xm.ravel(), C.ravel())[0, 1] if C.std() and Xm.std() else np.nan
        nulo = []
        for _ in range(N_PERMUT):
            p = np.arange(len(filas_pu))
            for lv in ("P", "U"):
                i = np.where(niv_f == lv)[0]; p[i] = p[rng.permutation(i)]
            nulo.append(np.corrcoef(Xm.ravel(), C[p].ravel())[0, 1] if C.std() and Xm.std() else np.nan)
        nulo = np.array(nulo)
        qap.append({"RELACION": nombre, "phi_obs": obs, "phi_nula_media": np.nanmean(nulo),
                    "p_superior": (1 + np.nansum(nulo >= obs)) / (N_PERMUT + 1)})
    guardar_tabla(pd.DataFrame(qap), "t10_RQ2_QAP_citas_alineamiento")

# ---- 7. Espacio conjunto: análisis de correspondencias (Zhu et al., 2016) --------------------------
Xca = pd.concat([B_tema, B_norm], axis=1) if B_norm is not None else B_tema.copy()
Xca = Xca.loc[Xca.sum(axis=1) > 0, Xca.sum(axis=0) > 0]
Pm = Xca.values / Xca.values.sum(); rm, cm = Pm.sum(axis=1), Pm.sum(axis=0)
S = (Pm - np.outer(rm, cm)) / np.sqrt(np.outer(rm, cm))
Uca, sv, Vt = np.linalg.svd(S, full_matrices=False)
F = (Uca * sv) / np.sqrt(rm)[:, None]; Gc = (Vt.T * sv) / np.sqrt(cm)[:, None]
inercia = sv ** 2
guardar_tabla(pd.DataFrame({"dimension": np.arange(1, len(inercia) + 1), "inercia": inercia,
                            "prop": inercia / inercia.sum()}).head(10), "t10_CA_inercias")
coord = pd.concat([pd.DataFrame({"ELEMENTO": Xca.index, "TIPO": "actor", "Dim1": F[:, 0], "Dim2": F[:, 1]}),
                   pd.DataFrame({"ELEMENTO": Xca.columns, "TIPO": "artefacto", "Dim1": Gc[:, 0], "Dim2": Gc[:, 1]})])
coord = coord.merge(nodos[["ID", "NIVEL", "BLOC"]].rename(columns={"ID": "ELEMENTO"}), on="ELEMENTO", how="left")
guardar_tabla(coord, "t10_CA_coordenadas")
fig, ax = plt.subplots()
for lv, mk in zip(NIVELES, ["s", "^", "o"]):
    d = coord[coord.NIVEL == lv]; ax.scatter(d.Dim1, d.Dim2, marker=mk, s=14, facecolors="none", edgecolors=COLOR_NIVEL[lv], label=lv)
d = coord[coord.TIPO == "artefacto"]; ax.scatter(d.Dim1, d.Dim2, marker="x", color="black", s=12)
for e, x, y in zip(d.ELEMENTO, d.Dim1, d.Dim2):
    ax.annotate(e.split("|")[-1][:18], (x, y), fontsize=6)
ax.axhline(0, color="lightgrey"); ax.axvline(0, color="lightgrey"); ax.legend(title="Level")
ax.set_xlabel(f"Dim 1 ({inercia[0] / inercia.sum():.0%})"); ax.set_ylabel(f"Dim 2 ({inercia[1] / inercia.sum():.0%})")
guardar_fig(fig, "f10_CA_espacio_conjunto", 9, 7)

# ---- 8. Posiciones: equivalencia estructural (por nivel y conjunta; Žiberna y Lazega, 2016) ---------
perfiles = np.hstack([A for r, A in MATS.items() if r != "multiplex"]).astype(float)
posiciones = []
for lv in NIVELES:
    i = idx_niv[lv]; k = min(K_POSICIONES[lv], len(i) - 1)
    if k < 2:
        posiciones.append(pd.DataFrame({"ID": [ids[x] for x in i], "POSICION": f"{lv}1"})); continue
    lab = fcluster(linkage(perfiles[i], method="ward"), k, "maxclust")
    posiciones.append(pd.DataFrame({"ID": [ids[x] for x in i], "POSICION": [f"{lv}{c}" for c in lab]}))
posiciones = pd.concat(posiciones)
k_conj = sum(K_POSICIONES.values())
posiciones["POSICION_CONJUNTA"] = posiciones.ID.map(dict(zip(ids, [f"J{c}" for c in fcluster(linkage(perfiles, method="ward"), k_conj, "maxclust")])))


def imagen(A, col):
    grupos = {p: [pos[i] for i in g.ID] for p, g in posiciones.groupby(col)}
    filas = []
    for a, b in itertools.product(grupos, repeat=2):
        M = A[np.ix_(grupos[a], grupos[b])]; n = len(grupos[a])
        filas.append({"desde": a, "hacia": b, "densidad": (M.sum() / (n * (n - 1)) if n > 1 else np.nan) if a == b else M.mean()})
    return pd.DataFrame(filas)


img = pd.concat([imagen(A, "POSICION").assign(RELACION=r) for r, A in MATS.items() if r != "multiplex"])
guardar_tabla(img, "t10_RQ2_matriz_imagen")
guardar_tabla(pd.concat([imagen(A, "POSICION_CONJUNTA").assign(RELACION=r) for r, A in MATS.items() if r != "multiplex"]),
              "t10_RQ2_matriz_imagen_conjunta")
pn = posiciones.merge(nodos[["ID", "NIVEL", "BLOC"]], on="ID")
guardar_tabla(pn.groupby(["POSICION", "BLOC"]).size().reset_index(name="n"), "t10_posiciones_por_bloc")
guardar_tabla(pn.groupby(["POSICION_CONJUNTA", "NIVEL"]).size().unstack(fill_value=0).reset_index(), "t10_posiciones_conjuntas_por_nivel")
rels = [r for r in MATS if r != "multiplex"]
fig, axs = plt.subplots(1, len(rels), squeeze=False)
for ax, r in zip(axs.ravel(), rels):
    m = img[img.RELACION == r].pivot(index="desde", columns="hacia", values="densidad")
    ax.imshow(m.values, cmap="seguile_azul", vmin=0, vmax=1)
    ax.set_xticks(range(m.shape[1])); ax.set_xticklabels(m.columns, fontsize=7); ax.set_yticks(range(m.shape[0]))
    ax.set_yticklabels(m.index, fontsize=7); ax.set_title(r)
    for a in range(m.shape[0]):
        for b in range(m.shape[1]):
            v = m.values[a, b]
            if pd.notna(v):
                ax.text(b, a, f"{v:.2f}", ha="center", va="center", fontsize=6, color="white" if v > .5 else "black")
guardar_fig(fig, "f10_RQ2_matriz_imagen", 12, 6)
metricas_red = metricas_red.merge(posiciones, on="ID", how="left")

# ---- 9. Balance posicional Norte/Sur y homofilia de bloque (RQ3b) ----------------------------------
bal, hom = [], []
for r, A in MATS.items():
    mr = metricas_red[metricas_red.RELACION == r]
    for lv in ("P", "U"):
        d = mr[mr.GOV_CODE == lv]
        for ind in ("EIGEN", "GRADO_ENTRE_NIVELES", "GRADO_A_ORGS"):
            for cb in ("BLOC", "BLOC_SENS"):
                bal.append({"RELACION": r, "NIVEL": lv, "INDICADOR": ind, "CLASIFICACION": cb,
                            **comparar_2g(d[ind], d[cb]).to_dict(), "p_perm": perm_test(d[ind], d[cb])})
        i = idx_niv[lv]; b = nodos.BLOC.to_numpy(dtype=object)[i]; ok = np.isin(b, ["North", "South"]); i, b = i[ok], b[ok]
        sub = A[np.ix_(i, i)]; t = np.triu_indices(len(i), 1); lazos = sub[t] == 1
        if lazos.sum() == 0:
            continue
        obs = np.mean(b[t[0]][lazos] == b[t[1]][lazos]); nulo = []
        for _ in range(N_PERMUT):
            bp = rng.permutation(b); nulo.append(np.mean(bp[t[0]][lazos] == bp[t[1]][lazos]))
        hom.append({"RELACION": r, "NIVEL": lv, "lazos": int(lazos.sum()), "prop_mismo_bloc_obs": obs,
                    "prop_mismo_bloc_nula": np.mean(nulo), "p_superior": (1 + np.sum(np.array(nulo) >= obs)) / (N_PERMUT + 1)})
bal = pd.DataFrame(bal); bal["p_BH"] = bal.groupby(["RELACION", "NIVEL", "CLASIFICACION"]).p_valor.transform(p_bh)
guardar_tabla(bal, "t10_RQ3b_balance_posicional")
guardar_tabla(pd.DataFrame(hom), "t10_RQ3b_homofilia_bloc")

guardar_tabla(metricas_red, "t10_metricas_nodo")
guardar_obj({"nodos": nodos, "B_tema": B_tema, "B_norm": B_norm, "A_tema": A_tema, "A_norm": A_norm,
             "A_multi": A_multi, "posiciones": posiciones}, "red_multinivel")
guardar_obj(metricas_red, "metricas_red")

# Exportación para Gephi (Cherven, 2015) y para los ERGM en R (Paso 10b)
DIR_RED = DIR_SALIDA / "red"; DIR_RED.mkdir(exist_ok=True)
nodos.to_csv(DIR_RED / "nodos.csv", index=False)
for r, A in MATS.items():
    pd.DataFrame(A, index=ids, columns=ids).to_csv(DIR_RED / f"A_{r}.csv")
    g = nx.from_numpy_array(A); nx.relabel_nodes(g, dict(enumerate(ids)), copy=False)
    for n, row in nodos.set_index("ID").iterrows():
        g.nodes[n].update({"nivel": row.NIVEL, "bloc": row.BLOC, "region": str(row.REGION)})
    nx.write_gexf(g, DIR_RED / f"red_{r}.gexf")
print("Paso 10 (RQ2, red) listo.")
```

## Paso 10b (opcional). ERGM por bloques, en R dentro de Colab (método 3.17, "Inference")
No hay una implementación madura de ERGM en Python, así que este paso escribe un script en R,
instala `network` y `ergm` (binarios de Posit Package Manager, unos minutos la primera vez) y lo ejecuta.
Activa `ESTIMAR_ERGM = True` en el Paso 1.

Modelos (Wang et al., 2016; los lazos dentro de nivel se tratan como fijos al modelar la red meso):
* **M1 U–G** (principal): aristas, estrellas alternantes por tipo de nodo (`gwb1degree`, `gwb2degree`),
  país alineado con la organización (`NAT`), su interacción con el alineamiento universidad–país (`CASC`,
  la cascada), bloc de la universidad, categoría de la organización y región compartida.
* **M2 P–G:** análogo, sin los términos de país.
* **M3 P–P y M4 U–U:** aristas, estrellas alternantes (`gwdegree`), `gwesp` y homofilia por bloc y región.

Limitación 8 del método: la proyección bipartita infla el cierre, así que los términos estructurales
son controles y la interpretación se centra en los efectos de atributo y entre niveles.

```python
R_ERGM = r'''
options(repos = c(CRAN = "https://packagemanager.posit.co/cran/__linux__/jammy/latest"),
        HTTPUserAgent = sprintf("R/%s R (%s)", getRversion(), paste(getRversion(), R.version$platform, R.version$arch, R.version$os)))
for (p in c("network", "ergm")) if (!requireNamespace(p, quietly = TRUE)) install.packages(p)
suppressPackageStartupMessages({library(network); library(ergm)})
args <- commandArgs(trailingOnly = TRUE); dir_red <- args[1]; semilla <- as.integer(args[2])
nodos <- read.csv(file.path(dir_red, "nodos.csv"), stringsAsFactors = FALSE)
f_mem <- args[3]
completar <- function(M) { n1 <- nrow(M); n2 <- ncol(M); Fm <- matrix(0, n1 + n2, n1 + n2)
  Fm[1:n1, n1 + seq_len(n2)] <- M; Fm[n1 + seq_len(n2), 1:n1] <- t(M); Fm }
tabla <- function(m, modelo) {
  if (is.null(m)) return(data.frame(MODELO = modelo, termino = NA))
  co <- summary(m)$coefficients
  data.frame(MODELO = modelo, termino = rownames(co), estimacion = co[, 1], ee = co[, 2], p_valor = co[, ncol(co)],
             AIC = AIC(m), BIC = BIC(m))
}
ajustar <- function(f) tryCatch(ergm(f, control = control.ergm(seed = semilla, MCMLE.maxit = 30)),
                                error = function(e) { message("ERGM no estimado: ", conditionMessage(e)); NULL })
res <- list(); MIEMBRO <- NULL
if (!is.na(f_mem) && file.exists(f_mem)) {                    # membresias_pais_org.csv: ISO2, ACTOR_G
  mem <- read.csv(f_mem, stringsAsFactors = FALSE)
  Pn <- nodos[nodos$NIVEL == "P", ]; Gn <- nodos[nodos$NIVEL == "G", ]
  MIEMBRO <- outer(Pn$ISO2, Gn$ACTOR, function(i, g) paste(i, g) %in% paste(mem$ISO2, mem$ACTOR_G)) * 1
  dimnames(MIEMBRO) <- list(Pn$ID, Gn$ID)
}
for (rel in c("tematica", "normativa")) {
  fa <- file.path(dir_red, paste0("A_", rel, ".csv")); if (!file.exists(fa)) next
  A <- as.matrix(read.csv(fa, row.names = 1, check.names = FALSE))
  G <- nodos[nodos$NIVEL == "G", ]; P <- nodos[nodos$NIVEL == "P", ]; U <- nodos[nodos$NIVEL == "U" & !is.na(nodos$PAIS_ID) & nodos$PAIS_ID != "", ]
  catG <- ifelse(is.na(G$ORG_CATEGORIA) | G$ORG_CATEGORIA == "", "Other", G$ORG_CATEGORIA)
  # M1: universidad-organización, condicionado a U-P y P-G observados
  if (nrow(U) > 2 && nrow(G) > 2) {
    Y1 <- A[U$ID, G$ID, drop = FALSE]; NAT <- A[U$PAIS_ID, G$ID, drop = FALSE]; ALIN <- A[cbind(U$ID, U$PAIS_ID)]
    CASC <- NAT * ALIN; REG <- outer(U$REGION, G$REGION, "==") * 1
    net1 <- network(Y1, matrix.type = "bipartite", directed = FALSE)
    set.vertex.attribute(net1, "BLOC", c(U$BLOC, G$BLOC)); set.vertex.attribute(net1, "CAT", c(rep("U", nrow(U)), catG))
    NATc <- completar(NAT); CASCc <- completar(CASC); REGc <- completar(REG)
    t1 <- c("edges", "gwb1degree(0.5, fixed = TRUE)", "gwb2degree(0.5, fixed = TRUE)", "edgecov(NATc)", "edgecov(CASCc)",
            "b1factor('BLOC')", "b2factor('CAT')")
    if (sum(REG) > 0) t1 <- c(t1, "edgecov(REGc)")
    if (!is.null(MIEMBRO)) { MIEc <- completar(MIEMBRO[U$PAIS_ID, G$ID, drop = FALSE]); t1 <- c(t1, "edgecov(MIEc)") }
    m1 <- ajustar(as.formula(paste("net1 ~", paste(t1, collapse = " + "))))
    res[[length(res) + 1]] <- cbind(tabla(m1, "M1 U-G"), RELACION = rel)
    if (!is.null(m1)) { pdf(file.path(dir_red, paste0("mcmc_M1_", rel, ".pdf"))); try(mcmc.diagnostics(m1)); dev.off() }
  }
  # M2: país-organización
  if (nrow(P) > 2 && nrow(G) > 2) {
    Y2 <- A[P$ID, G$ID, drop = FALSE]; REG2 <- completar(outer(P$REGION, G$REGION, "==") * 1)
    net2 <- network(Y2, matrix.type = "bipartite", directed = FALSE)
    set.vertex.attribute(net2, "BLOC", c(P$BLOC, G$BLOC)); set.vertex.attribute(net2, "CAT", c(rep("P", nrow(P)), catG))
    t2 <- c("edges", "gwb1degree(0.5, fixed = TRUE)", "gwb2degree(0.5, fixed = TRUE)", "b1factor('BLOC')", "b2factor('CAT')")
    if (sum(REG2) > 0) t2 <- c(t2, "edgecov(REG2)")
    if (!is.null(MIEMBRO)) { MIE2 <- completar(MIEMBRO[P$ID, G$ID, drop = FALSE]); t2 <- c(t2, "edgecov(MIE2)") }
    m2 <- ajustar(as.formula(paste("net2 ~", paste(t2, collapse = " + "))))
    res[[length(res) + 1]] <- cbind(tabla(m2, "M2 P-G"), RELACION = rel)
  }
  # M3 y M4: redes dentro de nivel
  for (lv in c("P", "U")) {
    N <- nodos[nodos$NIVEL == lv, ]; if (nrow(N) < 5) next
    net <- network(A[N$ID, N$ID], directed = FALSE)
    set.vertex.attribute(net, "BLOC", N$BLOC); set.vertex.attribute(net, "REGION", ifelse(is.na(N$REGION), "NA", N$REGION))
    m <- ajustar(net ~ edges + gwdegree(0.5, fixed = TRUE) + gwesp(0.5, fixed = TRUE) + nodematch("BLOC") + nodematch("REGION"))
    res[[length(res) + 1]] <- cbind(tabla(m, paste0(ifelse(lv == "P", "M3", "M4"), " ", lv, "-", lv)), RELACION = rel)
  }
}
out <- do.call(rbind, lapply(res, function(d) { for (c in c("termino","estimacion","ee","p_valor","AIC","BIC")) if (!c %in% names(d)) d[[c]] <- NA; d[, c("MODELO","termino","estimacion","ee","p_valor","AIC","BIC","RELACION")] }))
write.csv(out, file.path(dir_red, "ergm_resultados.csv"), row.names = FALSE)
cat("ERGM listos\n")
'''

if ESTIMAR_ERGM:
    import shutil
    if shutil.which("Rscript") is None:
        subprocess.run("apt-get -qq update && apt-get -qq install -y r-base > /dev/null", shell=True, check=True)
    DIR_RED = DIR_SALIDA / "red"
    (DIR_RED / "ergm.R").write_text(R_ERGM)
    f_mem = DIR_PROYECTO / "membresias_pais_org.csv"
    proc = subprocess.run(["Rscript", str(DIR_RED / "ergm.R"), str(DIR_RED), str(SEMILLA), str(f_mem)],
                          capture_output=True, text=True)
    print(proc.stdout[-3000:], proc.stderr[-3000:])
    if (DIR_RED / "ergm_resultados.csv").exists():
        guardar_tabla(pd.read_csv(DIR_RED / "ergm_resultados.csv"), "t10_RQ2_ergm")
        print(TAB["t10_RQ2_ergm"].to_string())
else:
    print("ERGM no estimados (ESTIMAR_ERGM = False).")
```

## Paso 11. RQ3: balance Norte/Sur dentro de países y universidades (método 3.12 y 3.16)
Con el **actor** como unidad, dentro de los niveles P y U: (a) énfasis temático (γ por tópico,
entropía, densidades de la taxonomía), (b) fuerza normativa (proporción de instrumentos obligatorios,
índice deóntico), (c) alineamiento (Hellinger temático y normativo, indicadores de red por permutación).
Wilcoxon + r biserial (positivo = mayor en el Norte), BH por nivel × familia, mínimo 3 actores por grupo.
**Robusto** = significativo y con el mismo signo con la clasificación principal y la alternativa.

**[REVISIÓN] Dependencia por país.** El bloc es un atributo del país, y las universidades del mismo país no
son independientes (manual de Humphrey y LeBreton, 2019: la no independencia infla el error de tipo I).
Para las universidades se reporta el ICC(1) por país y un p-valor por permutación **entre países**
(`p_perm_pais`). No está en el método v1.6: si lo usas, añádelo a la sección 3.18.

```python
PASO = "Step11"
actor_topico = leer_obj("actor_topico"); actor_metricas = leer_obj("actor_metricas")
fuerza_actor = leer_obj("fuerza_actor"); actores_estado = leer_obj("actores_estado")
claves = ["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]

# ---- 1. Composición (descriptivo) ----------------------------------------------------------
comp = actores_estado[actores_estado.GOV_CODE.isin(["P", "U"])].groupby(["GOVERNANCE", "BLOC", "AUSENCIA"]).size().reset_index(name="n")
guardar_tabla(comp, "t11_composicion")
if existe_obj("doc_actor"):
    da = leer_obj("doc_actor")
    guardar_tabla(da[da.GOV_CODE.isin(["P", "U"])].groupby(["GOVERNANCE", "BLOC", "ACTOR"]).CODE.nunique()
                  .groupby(["GOVERNANCE", "BLOC"]).describe().reset_index(), "t11_documentos_por_actor")

# ---- 2. Tabla larga de indicadores por actor ---------------------------------------------------
def bloque(df, familia, indicador, valor):
    d = df[claves].copy(); d["FAMILIA"] = familia; d["INDICADOR"] = indicador; d["valor"] = np.asarray(valor, float)
    return d


partes = [bloque(actor_topico, "a_tematica", "T" + actor_topico.topic.astype(str), actor_topico.gamma),
          bloque(actor_metricas, "a_tematica", "ENTROPIA", actor_metricas.ENTROPY_NORM),
          bloque(fuerza_actor, "b_fuerza", "FUERZA_MAX_con_ausencia", fuerza_actor.FUERZA_MAX)]
fd = fuerza_actor[fuerza_actor.n_docs > 0]
partes += [bloque(fd, "b_fuerza", "PROP_OBLIGATORIO", fd.PROP_VINCULANTE),
           bloque(fd, "b_fuerza", "INDICE_DEONTICO", fd.INDICE_DEONTICO),
           bloque(actor_metricas, "c_alineamiento", "HELLINGER_TEMATICO", actor_metricas.HELLINGER_GLOBAL)]
if existe_obj("dim_actor"):
    dim_actor = leer_obj("dim_actor"); sub_actor = leer_obj("sub_actor"); alin = leer_obj("alineamiento_normativo")
    partes += [bloque(dim_actor, "a_taxonomia", "DENS_" + dim_actor.DIMENSION.astype(str), dim_actor.DENSIDAD),
               bloque(sub_actor, "a_taxonomia_sub", "DENS_" + sub_actor.DIMENSION.astype(str) + "|" + sub_actor.SUBDIMENSION.astype(str), sub_actor.DENSIDAD),
               bloque(alin, "c_alineamiento", "HELLINGER_NORMATIVO", alin.HELLINGER_NORMATIVO)]
if existe_obj("metricas_red"):
    mr = leer_obj("metricas_red"); mr = mr[mr.RELACION.isin(["tematica", "normativa"])]
    for m in ("PROP_LAZOS_G", "PROP_G_NORTE", "PROP_G_SUR", "EIGEN", "GRADO_ENTRE_NIVELES"):
        partes.append(bloque(mr, "c_red", m + "_" + mr.RELACION, mr[m]))
largo = pd.concat(partes, ignore_index=True)
pais = actores_estado.set_index(["GOV_CODE", "ACTOR"]).ISO2
largo["ISO2"] = [pais.get((g, a)) for g, a in zip(largo.GOV_CODE, largo.ACTOR)]

# ---- 3. Pruebas Norte vs Sur con ambas clasificaciones -------------------------------------------
filas = []
for cb in ("BLOC", "BLOC_SENS"):
    d = largo[largo.GOV_CODE.isin(["P", "U"]) & largo[cb].isin(["North", "South"])]
    for (gov, fam, ind), g in d.groupby(["GOVERNANCE", "FAMILIA", "INDICADOR"]):
        r = comparar_2g(g.valor, g[cb]).to_dict()
        r["p_perm"] = perm_test(g.valor, g[cb]) if fam == "c_red" else np.nan
        es_u = g.GOV_CODE.iloc[0] == "U"
        r["p_perm_pais"] = perm_test(g.valor, g[cb], bloques=g.ISO2.fillna("NA")) if es_u else np.nan
        r["ICC1_pais"] = icc1(g.valor, g.ISO2) if es_u else np.nan
        filas.append({"GOVERNANCE": gov, "FAMILIA": fam, "INDICADOR": ind, "CLASIFICACION": cb, **r})
ns = pd.DataFrame(filas)
ns["p_usado"] = ns.p_perm.fillna(ns.p_valor)
ns["p_BH"] = ns.groupby(["CLASIFICACION", "GOVERNANCE", "FAMILIA"]).p_usado.transform(p_bh)
guardar_tabla(ns, "t11_RQ3_norte_sur")       # _1 = North, _2 = South; r_rb > 0: mayor en el Norte

rob = ns.pivot_table(index=["GOVERNANCE", "FAMILIA", "INDICADOR"], columns="CLASIFICACION", values=["r_rb", "p_BH"], aggfunc="first")
rob.columns = [f"{a}_{b}" for a, b in rob.columns]; rob = rob.reset_index()
rob["ROBUSTO"] = ((rob.p_BH_BLOC < 0.05) & (rob.p_BH_BLOC_SENS < 0.05) & (np.sign(rob.r_rb_BLOC) == np.sign(rob.r_rb_BLOC_SENS))).fillna(False)
guardar_tabla(rob, "t11_RQ3_robustez_clasificacion")

# ---- 4. Tópico dominante, categorías de instrumento y ausencia (método 3.12 y 3.16) ----------------
td, cat, ausf = [], [], []
fuerza_actor["CATEGORIA"] = fuerza_actor.FUERZA_ETIQ.astype(str)
for cb in ("BLOC", "BLOC_SENS"):
    for gov, g in actor_metricas[actor_metricas.GOV_CODE.isin(["P", "U"]) & actor_metricas[cb].isin(["North", "South"])].groupby("GOVERNANCE"):
        td.append({"GOVERNANCE": gov, "CLASIFICACION": cb, **chi_sim(g[cb], g.TOPIC_DOM).to_dict()})
    for gov, g in fuerza_actor[fuerza_actor.GOV_CODE.isin(["P", "U"]) & fuerza_actor[cb].isin(["North", "South"])].groupby("GOVERNANCE"):
        cat.append({"GOVERNANCE": gov, "CLASIFICACION": cb, **fisher_sim(g[cb], g.CATEGORIA).to_dict()})
    for gov, g in actores_estado[actores_estado.GOV_CODE.isin(["P", "U"]) & actores_estado[cb].isin(["North", "South"])].groupby("GOVERNANCE"):
        t = pd.crosstab(g[cb], g.AUSENCIA)
        p = stats.fisher_exact(t.values).pvalue if t.shape == (2, 2) else np.nan
        ausf.append({"GOVERNANCE": gov, "CLASIFICACION": cb, "prop_ausencia_north": g.AUSENCIA[g[cb] == "North"].mean(),
                     "prop_ausencia_south": g.AUSENCIA[g[cb] == "South"].mean(), "p_fisher": p})
guardar_tabla(pd.DataFrame(td), "t11_RQ3_topico_dominante")
guardar_tabla(pd.DataFrame(cat), "t11_RQ3_categorias_instrumento")      # método 3.16: Fisher simulado + V de Cramér
guardar_tabla(pd.DataFrame(ausf), "t11_RQ3_ausencia")

# ---- 5. Figura principal ---------------------------------------------------------------------------
fg = ns[(ns.CLASIFICACION == "BLOC") & ns.r_rb.notna() & (ns.FAMILIA != "a_taxonomia_sub")].merge(
    rob[["GOVERNANCE", "FAMILIA", "INDICADOR", "ROBUSTO"]], on=["GOVERNANCE", "FAMILIA", "INDICADOR"])
fg["estado"] = np.where(fg.ROBUSTO, "Robust (both classifications)", np.where(fg.p_BH < 0.05, "Main classification only", "Not significant"))
govs = sorted(fg.GOVERNANCE.unique())
fig, axs = plt.subplots(1, len(govs), squeeze=False, sharey=True)
inds = sorted(fg.INDICADOR.unique(), key=lambda s: (fg.FAMILIA[fg.INDICADOR == s].iloc[0], s))
marcas = {"Robust (both classifications)": ("o", "black"), "Main classification only": ("o", "none"), "Not significant": ("x", "grey")}
for ax, gov in zip(axs.ravel(), govs):
    d = fg[fg.GOVERNANCE == gov]
    for est, (mk, fc) in marcas.items():
        e = d[d.estado == est]
        ax.scatter(e.r_rb, [inds.index(i) for i in e.INDICADOR], marker=mk, facecolors=fc, edgecolors="black" if mk == "o" else None,
                   color="grey" if mk == "x" else None, label=est)
    ax.axvline(0, color="grey"); ax.set_title(gov, fontsize=8); ax.set_xlabel("Rank-biserial r (positive = higher in North)")
axs[0, 0].set_yticks(range(len(inds))); axs[0, 0].set_yticklabels(inds, fontsize=6); axs[0, 0].legend(fontsize=6)
guardar_fig(fig, "f11_RQ3_norte_sur", 11, max(6, 0.22 * len(inds)))
print("Paso 11 (RQ3) listo.")
```

## Paso 12. Robustez (método 3.18)
Las siete variaciones del método, más dos que traían los scripts R:

| Código | Variación | Dónde |
|---|---|---|
| R1 | Número de tópicos (*k* ± 2) y semilla del LDA | aquí |
| R2 | Clasificación Norte/Sur principal vs alternativa | Paso 11 (`t11_RQ3_robustez_clasificacion`) |
| R3 | Unidad de análisis: actor vs documento | aquí |
| R4 | Composición: sin instrumentos subnacionales y sin los documentos marcados en el cribado | aquí |
| R5 | Sin documentos con año inconsistente con su contenido | aquí (RQ1c) |
| R6 | Parámetros de construcción de lazos de la red | aquí |
| R7 | Sin comunicaciones institucionales | aquí |
| extra | Solo originales en inglés; sin casi duplicados ni documentos sin año | scripts R |

La tabla `t12_resumen_robustez` indica, para cada contraste Norte/Sur de tópicos, si el signo del
efecto se mantiene respecto del análisis principal.

```python
PASO = "Step12"
dtm = leer_obj("dtm"); beta_td = leer_obj("beta_td"); gamma_td = leer_obj("gamma_td")
doc_actor = leer_obj("doc_actor"); docs_meta = leer_obj("docs_meta")
actor_metricas = leer_obj("actor_metricas"); fuerza_doc = leer_obj("fuerza_doc")
from sklearn.decomposition import LatentDirichletAllocation
CLAVES_ACTOR = ["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC", "BLOC_SENS"]


def perfil_actor(gt):
    return gt.merge(doc_actor, on="CODE").groupby(CLAVES_ACTOR + ["topic"], dropna=False).gamma.mean().reset_index()


def ns_topicos(at, etiqueta, unidad="ACTOR"):
    d = at[at.GOV_CODE.isin(["P", "U"]) & at.BLOC.isin(["North", "South"])]
    r = resumen_por(d, ["GOVERNANCE", "topic"], lambda g: comparar_2g(g.gamma, g.BLOC))
    r["p_BH"] = r.groupby("GOVERNANCE").p_valor.transform(p_bh)
    return r.assign(ANALISIS=etiqueta)


def subconjunto(codes, etiqueta):
    return ns_topicos(perfil_actor(gamma_td[gamma_td.CODE.isin(codes)]), etiqueta)


principal = ns_topicos(perfil_actor(gamma_td), "Principal (actor)")

# ---- R1. Estabilidad de tópicos ------------------------------------------------------------------
K_FIN = gamma_td.topic.nunique()
Bm = beta_td.pivot(index="topic", columns="term", values="beta").fillna(0)
alternativos = {**{f"semilla_{SEMILLA + i}": (K_FIN, SEMILLA + i) for i in (1, 2, 3)},
                f"k_{K_FIN - 2}": (K_FIN - 2, SEMILLA), f"k_{K_FIN + 2}": (K_FIN + 2, SEMILLA)}
r1 = []
for nombre, (k, sem) in alternativos.items():
    mod = LatentDirichletAllocation(n_components=k, learning_method="batch", max_iter=LDA_MAX_ITER,
                                    random_state=sem, n_jobs=-1).fit(dtm["X"])
    Ba = pd.DataFrame(mod.components_ / mod.components_.sum(1, keepdims=True), columns=dtm["terms"])[Bm.columns]
    sim = (Bm.values @ Ba.values.T) / np.outer(np.linalg.norm(Bm.values, axis=1), np.linalg.norm(Ba.values, axis=1))
    g = mod.transform(dtm["X"])
    gt = pd.DataFrame([(dtm["docs"][i], t + 1, g[i, t]) for i in range(g.shape[0]) for t in range(k)], columns=["CODE", "topic", "gamma"])
    h = distancia_referente(perfil_actor(gt), "topic", "gamma").merge(actor_metricas[["GOV_CODE", "ACTOR", "HELLINGER_GLOBAL"]],
                                                                     on=["GOV_CODE", "ACTOR"])
    rho = stats.spearmanr(h.HELLINGER, h.HELLINGER_GLOBAL, nan_policy="omit").statistic
    r1 += [{"MODELO": nombre, "topic": int(t), "coseno_max": sim[i].max(), "rho_hellinger_actor": rho} for i, t in enumerate(Bm.index)]
    print(f"  R1 {nombre} listo")
guardar_tabla(pd.DataFrame(r1), "t12_R1_estabilidad_topicos")

# ---- R3. Unidad de análisis: documento ----------------------------------------------------------
bloc_doc = doc_actor.groupby("CODE").BLOC.agg(lambda s: s.iloc[0] if s.nunique() == 1 else np.nan).reset_index()
gd = gamma_td.merge(bloc_doc, on="CODE").merge(docs_meta[["CODE", "GOV_CODE", "GOVERNANCE"]], on="CODE")
r3 = ns_topicos(gd, "Documento")

# ---- R4. Composición: sin subnacionales ni documentos marcados en el cribado --------------------
r4 = subconjunto(docs_meta.CODE[~docs_meta.SUBNACIONAL & ~docs_meta.SENSIBILIDAD], "Sin subnacionales ni marcados en cribado")

# ---- R7. Sin comunicaciones institucionales ------------------------------------------------------
r7 = subconjunto(docs_meta.CODE[~docs_meta.COMUNICACION], "Sin comunicaciones institucionales")

# ---- Extras de los scripts R ----------------------------------------------------------------------
casi = leer_obj("casi_dup").CODE_2 if existe_obj("casi_dup") else pd.Series(dtype=str)
ex1 = subconjunto(docs_meta.CODE[docs_meta.YEAR.notna() & ~docs_meta.CODE.isin(casi)], "Sin casi duplicados ni sin año")
orig = docs_meta.CODE[docs_meta.TRADUCIDO.astype("boolean").eq(False).fillna(False)]
ex2 = subconjunto(orig, "Solo originales en inglés") if len(orig) else pd.DataFrame()
if len(orig):
    fo = (fuerza_doc[fuerza_doc.CODE.isin(orig)].drop(columns="GOVERNANCE")
          .merge(doc_actor[["CODE", "GOVERNANCE", "ACTOR", "BLOC"]], on="CODE")
          .groupby(["GOV_CODE", "GOVERNANCE", "ACTOR", "BLOC"]).INDICE_DEONTICO.mean().reset_index())
    fo = fo[fo.GOV_CODE.isin(["P", "U"]) & np.isfinite(fo.INDICE_DEONTICO)]
    guardar_tabla(resumen_por(fo, "GOVERNANCE", lambda g: comparar_2g(g.INDICE_DEONTICO, g.BLOC))
                  .assign(ANALISIS="Índice deóntico, originales"), "t12_extra_deontico_originales")

comparacion = pd.concat([principal, r3, r4, r7, ex1, ex2], ignore_index=True)[["ANALISIS", "GOVERNANCE", "topic", "r_rb", "p_BH"]]
ref = principal.set_index(["GOVERNANCE", "topic"]).r_rb
comparacion["mismo_signo_que_principal"] = [np.sign(r) == np.sign(ref.get((g, t), np.nan))
                                            for g, t, r in zip(comparacion.GOVERNANCE, comparacion.topic, comparacion.r_rb)]
guardar_tabla(comparacion, "t12_R3_R4_R7_norte_sur_topicos")

# ---- R5. Años inconsistentes con el contenido (afecta a RQ1c) -------------------------------------
dt = docs_meta[docs_meta.YEAR.notna() & ~docs_meta.ANIO_INCONSISTENTE].copy()
dt["PERIODO"] = np.where(dt.YEAR >= ANIO_CORTE_GENAI, "Post", "Pre")
r5 = resumen_por(gamma_td.merge(dt[["CODE", "GOVERNANCE", "PERIODO"]], on="CODE"), ["GOVERNANCE", "topic"],
                 lambda g: comparar_2g(g.gamma, g.PERIODO, "Post", "Pre"))
r5["p_BH"] = r5.groupby("GOVERNANCE").p_valor.transform(p_bh)
if "t07_RQ1c_topicos_post_vs_pre" in TAB or (DIR_TAB / "t07_RQ1c_topicos_post_vs_pre.csv").exists():
    p7 = pd.read_csv(DIR_TAB / "t07_RQ1c_topicos_post_vs_pre.csv").set_index(["GOVERNANCE", "topic"]).r_rb
    r5["mismo_signo_que_principal"] = [np.sign(r) == np.sign(p7.get((g, t), np.nan)) for g, t, r in zip(r5.GOVERNANCE, r5.topic, r5.r_rb)]
guardar_tabla(r5.assign(n_excluidos=int(docs_meta.ANIO_INCONSISTENTE.sum())), "t12_R5_RQ1c_sin_anios_inconsistentes")

# ---- R6. Parámetros de construcción de lazos de la red ---------------------------------------------
if existe_obj("red_multinivel"):
    red = leer_obj("red_multinivel"); at = leer_obj("actor_topico")
    at = at.assign(ID=at.GOV_CODE + "::" + at.ACTOR.astype(str))
    ids_red = red["nodos"].ID.tolist()
    B_tema_02 = (at.assign(v=(at.gamma >= UMBRAL_MULTI).astype(int)).pivot_table(index="ID", columns="topic", values="v", fill_value=0)
                 .reindex(ids_red, fill_value=0).values)
    variantes = {"principal_tematica": (red["B_tema"].values, ALFA_BACKBONE, MTC_BACKBONE),
                 "alfa_001_tematica": (red["B_tema"].values, 0.01, MTC_BACKBONE),
                 "holm_tematica": (red["B_tema"].values, ALFA_BACKBONE, "holm"),
                 "umbral_020_tematica": (B_tema_02, ALFA_BACKBONE, MTC_BACKBONE)}
    if red["B_norm"] is not None:
        variantes.update({"principal_normativa": (red["B_norm"].values, ALFA_BACKBONE, MTC_BACKBONE),
                          "alfa_001_normativa": (red["B_norm"].values, 0.01, MTC_BACKBONE),
                          "holm_normativa": (red["B_norm"].values, ALFA_BACKBONE, "holm")})
        sa = leer_obj("sub_actor"); alt = "CUOTA" if VAR_LAZO_NORMATIVO == "DENSIDAD" else "DENSIDAD"
        sa = sa[sa[alt].notna()].assign(ID=lambda d: d.GOV_CODE + "::" + d.ACTOR.astype(str), SUB=lambda d: d.DIMENSION + "|" + d.SUBDIMENSION)
        sa["v"] = ((sa[alt] > sa.groupby("SUB")[alt].transform("median")) & (sa[alt] > 0)).astype(int)
        variantes[f"lazo_por_{alt.lower()}_normativa"] = (sa.pivot_table(index="ID", columns="SUB", values="v", fill_value=0)
                                                         .reindex(ids_red, fill_value=0).values, ALFA_BACKBONE, MTC_BACKBONE)
    ctx = contexto_red(red["nodos"]); filas = []
    for n, (B, a, m) in variantes.items():
        A = extraer_backbone(B, a, m)
        filas.append({"VARIANTE": n, "lazos": int(A.sum() // 2), **censo_configuraciones(A, ctx)})
    guardar_tabla(pd.DataFrame(filas), "t12_R6_red_parametros_alternativos")

# ---- Resumen ----------------------------------------------------------------------------------------
res = (comparacion[comparacion.ANALISIS != "Principal (actor)"].groupby("ANALISIS").mismo_signo_que_principal
       .mean().rename("prop_contrastes_mismo_signo").reset_index())
guardar_tabla(res, "t12_resumen_robustez")
print(res.to_string(index=False))
print("Paso 12 (robustez) listo.")
```

```python
# Límite de palabras de la sección Results según la revista (escríbelo cuando lo confirmes; None = sin control)
LIMITE_PALABRAS_RESULTADOS = None

import base64, io, html as _html, datetime, traceback
import matplotlib.ticker
from matplotlib.colors import LinearSegmentedColormap, ListedColormap
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import networkx as nx

# ---- 1. Paleta y estilo (paleta validada para daltonismo) ----------------------------------
TINTA, TINTA2, TENUE = "#0b0b0b", "#52514e", "#898781"
SUPERFICIE, REJILLA, EJE = "#fcfcfb", "#e1e0d9", "#c3c2b7"
NIVEL_COLOR = {"G": "#2a78d6", "P": "#eb6834", "U": "#1baf7a"}
NIVEL_NOMBRE = {"G": "Global (organizations)", "P": "National (countries)", "U": "Institutional (universities)"}
NIVEL_CORTO = {"G": "Organizations", "P": "Countries", "U": "Universities"}
GOV_A_COD = {v: k for k, v in NIVEL_NOMBRE.items()}
NIVEL_ES = {"G": "organizaciones", "P": "países", "U": "universidades"}


def es(gov):
    """Nombre del nivel en español para las lecturas (las figuras quedan en inglés, para el artículo)."""
    return NIVEL_ES.get(GOV_A_COD.get(gov, gov), str(gov))
BLOC_COLOR = {"North": "#6250d6", "South": "#eda100", "Global": "#898781", "Mixed": "#c3c2b7", "Unknown": "#e1e0d9"}
AZUL = ["#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#2a78d6", "#1c5cab", "#104281", "#0d366b"]
CMAP_SEQ = LinearSegmentedColormap.from_list("seguile_azul", ["#f3f8fe"] + AZUL)
CMAP_DIV = LinearSegmentedColormap.from_list("seguile_div", ["#104281", "#3987e5", "#9ec5f4", "#f0efec",
                                                             "#f3b1b0", "#e34948", "#a32727"])
ORDINAL = ["#86b6ef", "#5598e7", "#2a78d6", "#1c5cab", "#104281"]     # ramp ordinal (no más clara que 250)
for _c in (CMAP_SEQ, CMAP_DIV):
    try:
        matplotlib.colormaps.register(_c, force=True)
    except Exception:
        pass

plt.rcParams.update({
    "figure.facecolor": SUPERFICIE, "axes.facecolor": SUPERFICIE, "savefig.facecolor": SUPERFICIE,
    "axes.edgecolor": EJE, "axes.labelcolor": TINTA2, "axes.titlecolor": TINTA, "text.color": TINTA,
    "xtick.color": TENUE, "ytick.color": TENUE, "xtick.labelcolor": TINTA2, "ytick.labelcolor": TINTA2,
    "axes.grid": True, "grid.color": REJILLA, "grid.linewidth": 0.6, "grid.linestyle": "-",
    "axes.spines.top": False, "axes.spines.right": False, "axes.axisbelow": True,
    "font.size": 9.5, "axes.titlesize": 11, "axes.titleweight": "bold", "axes.titlelocation": "left",
    "axes.labelsize": 9.5, "legend.frameon": False, "legend.fontsize": 8.5, "lines.linewidth": 2,
    "font.family": "sans-serif", "font.sans-serif": ["DejaVu Sans", "Arial", "Helvetica"],
})
DIR_INF = DIR_FIG / "informe"; DIR_INF.mkdir(parents=True, exist_ok=True)


def tabla(nombre):
    f = DIR_TAB / f"{nombre}.csv"
    if not f.exists():
        return None
    try:
        d = pd.read_csv(f)
    except pd.errors.EmptyDataError:
        return None
    return d if len(d) else None


def obj(nombre):
    return leer_obj(nombre) if existe_obj(nombre) else None


def corto(gov):
    return NIVEL_CORTO.get(GOV_A_COD.get(gov, gov), gov)


def color_nivel(gov):
    return NIVEL_COLOR.get(GOV_A_COD.get(gov, gov), TENUE)


def leyenda_niveles(ax, codigos=("G", "P", "U"), **kw):
    ax.legend(handles=[Patch(facecolor=NIVEL_COLOR[c], label=NIVEL_CORTO[c]) for c in codigos], **kw)


def etiqueta_topico(t, etiquetas, largo=34):
    e = etiquetas.set_index("topic").TOPIC_LABEL.get(t, f"T{t}") if etiquetas is not None else f"T{t}"
    return e if len(e) <= largo else e[:largo - 1] + "…"


def heatmap(ax, M, filas, cols, cmap, vmin=None, vmax=None, fmt="{:.2f}", cbar_label=""):
    im = ax.imshow(M, cmap=cmap, aspect="auto", vmin=vmin, vmax=vmax)
    ax.set_xticks(range(len(cols))); ax.set_xticklabels(cols, rotation=35, ha="right")
    ax.set_yticks(range(len(filas))); ax.set_yticklabels(filas)
    ax.grid(False)
    for s in ax.spines.values():
        s.set_visible(False)
    lim = np.nanmax(np.abs(M)) if vmax is None else max(abs(vmin or 0), abs(vmax))
    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            v = M[i, j]
            if np.isfinite(v):
                rel = (v - (vmin if vmin is not None else np.nanmin(M))) / ((vmax if vmax is not None else np.nanmax(M)) - (vmin if vmin is not None else np.nanmin(M)) + 1e-12)
                oscuro = rel > 0.62 if cmap is CMAP_SEQ else abs(v) > 0.55 * lim
                ax.text(j, i, fmt.format(v), ha="center", va="center", fontsize=7.5,
                        color="white" if oscuro else TINTA)
    cb = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cb.outline.set_visible(False); cb.set_label(cbar_label, color=TINTA2)
    return im


FIGURAS = {}


def fig_a_html(fig, nombre, ancho=None):
    ruta = DIR_INF / f"{nombre}.png"
    fig.savefig(ruta, dpi=200, bbox_inches="tight")
    buf = io.BytesIO(); fig.savefig(buf, format="png", dpi=110, bbox_inches="tight"); plt.close(fig)
    FIGURAS[nombre] = str(ruta)
    return f'<img src="data:image/png;base64,{base64.b64encode(buf.getvalue()).decode()}" alt="{nombre}">'


# ---- 2. Utilidades de interpretación (reglas fijas, borrador) -------------------------------
def mag_eps2(e):
    return "sin dato" if not np.isfinite(e) else "trivial" if e < .01 else "pequeño" if e < .08 else "moderado" if e < .26 else "grande"


def mag_r(r):
    a = abs(r) if np.isfinite(r) else np.nan
    return "sin dato" if not np.isfinite(a) else "trivial" if a < .1 else "pequeño" if a < .3 else "moderado" if a < .5 else "grande"


def mag_v(v):
    return "sin dato" if not np.isfinite(v) else "trivial" if v < .1 else "pequeña" if v < .3 else "moderada" if v < .5 else "fuerte"


def fp(p):
    return "n/d" if p is None or not np.isfinite(p) else ("p < 0,001" if p < .001 else f"p = {p:.3f}".replace(".", ","))


def fn(x, d=2):
    return "n/d" if x is None or not np.isfinite(x) else f"{x:.{d}f}".replace(".", ",")


def tabla_html(df, cols=None, n=20, renombrar=None, nota=None):
    if df is None or df.empty:
        return '<p class="aviso">Tabla no disponible.</p>'
    d = df[cols].copy() if cols else df.copy()
    if renombrar:
        d = d.rename(columns=renombrar)
    total = len(d); d = d.head(n)
    for c in d.columns:
        if pd.api.types.is_float_dtype(d[c]):
            d[c] = d[c].map(lambda v: "" if pd.isna(v) else (f"{v:.3f}" if abs(v) < 1000 else f"{v:,.0f}"))
    pie = f'<p class="nota">Se muestran {n} de {total} filas. ' if total > n else '<p class="nota">'
    pie += (nota or "") + "</p>"
    return d.to_html(index=False, border=0, classes="tabla", escape=True, na_rep="") + pie


SECCIONES = []


def seccion(id_, titulo, metodo, pregunta=None):
    def deco(fun):
        try:
            cuerpo = fun()
        except Exception as e:
            cuerpo = (f'<p class="aviso">No se pudo generar esta sección ({_html.escape(type(e).__name__)}: '
                      f'{_html.escape(str(e))[:300]}). Revisa que el paso correspondiente se haya corrido.</p>')
            traceback.print_exc()
        SECCIONES.append({"id": id_, "titulo": titulo, "metodo": metodo, "pregunta": pregunta, "cuerpo": cuerpo})
        print(f"  ✔ {titulo}")
        return fun
    return deco


def lectura(items):
    if not items:
        return ""
    lis = "".join(f"<li>{t}</li>" for t in items)
    return f'<div class="lectura"><h4>Lectura de los resultados <span>(borrador automático: revisar)</span></h4><ul>{lis}</ul></div>'


def figura(img, pie):
    return f'<figure>{img}<figcaption>{pie}</figcaption></figure>'


etiquetas = obj("etiquetas")
print("Generando el informe…")


# ---- 3. Secciones ---------------------------------------------------------------------------------
@seccion("corpus", "Corpus analizado", "Método 3.2–3.8")
def _():
    reg, corpus_, act = obj("registros"), obj("corpus"), obj("actores_estado")
    if reg is None:
        raise FileNotFoundError("Falta el Paso 2")
    filas = []
    for c in ("G", "P", "U"):
        r = reg[reg.GOV_CODE == c]
        filas.append({"Nivel": NIVEL_ES[c].capitalize(), "Registros": len(r), "Con documento": int((~r.SIN_DOCUMENTO).sum()),
                      "Textos únicos": int((corpus_.GOV_CODE == c).sum()) if corpus_ is not None else np.nan,
                      "Actores": int((act.GOV_CODE == c).sum()) if act is not None else np.nan,
                      "Ausencias": int(act[(act.GOV_CODE == c)].AUSENCIA.sum()) if act is not None else np.nan})
    comp = pd.DataFrame(filas)
    fig, ax = plt.subplots(figsize=(9, 3.6))
    metricas = ["Registros", "Con documento", "Textos únicos", "Actores"]
    y = np.arange(len(metricas)); h = 0.26
    for k, c in enumerate(("G", "P", "U")):
        vals = comp.loc[k, metricas].astype(float).values
        b = ax.barh(y + (k - 1) * h, vals, height=h - 0.03, color=NIVEL_COLOR[c], label=NIVEL_CORTO[c])
        for yy, v in zip(y + (k - 1) * h, vals):
            ax.text(v + max(comp[metricas].max()) * 0.01, yy, f"{v:.0f}", va="center", fontsize=8, color=TINTA2)
    ax.set_yticks(y); ax.set_yticklabels(metricas); ax.invert_yaxis(); ax.grid(axis="y", visible=False)
    ax.set_title("Composition of the corpus by level of governance"); ax.legend(loc="lower right")
    img = fig_a_html(fig, "f00_corpus")
    pend = tabla("t02_PENDIENTES_sin_txt")
    items = [f"El corpus tiene <b>{len(reg)}</b> registros; <b>{int((~reg.SIN_DOCUMENTO).sum())}</b> tienen documento "
             f"y <b>{len(corpus_) if corpus_ is not None else 'n/d'}</b> textos únicos entran en los modelos."]
    if act is not None:
        items.append(f"Hay <b>{int(act.AUSENCIA.sum())}</b> actores en ausencia (sin documento localizable), "
                     "que se reportan aparte y entran en la capa 1 de fuerza normativa como «Absence».")
    if pend is not None and len(pend):
        items.append(f"<b>Atención:</b> {len(pend)} registros con documento todavía no tienen .txt y quedaron fuera del "
                     "análisis de texto (tabla t02_PENDIENTES_sin_txt).")
    compartidos = tabla("t02_textos_compartidos")
    if compartidos is not None and len(compartidos):
        items.append(f"{len(compartidos)} textos están compartidos por varios registros; cada uno entra una sola vez en los modelos.")
    return figura(img, "Figura 0. Registros, documentos, textos y actores por nivel.") + tabla_html(comp) + lectura(items)


@seccion("modelo", "Modelo de tópicos (LDA)", "Método 3.9")
def _():
    cv, top = tabla("t04_seleccion_k"), tabla("t04_top_terminos")
    html_ = ""
    items = []
    if cv is not None and len(cv):
        fig, ax = plt.subplots(figsize=(8, 3.6))
        ax.fill_between(cv.k, cv.perplejidad_media - cv.perplejidad_de, cv.perplejidad_media + cv.perplejidad_de,
                        color=AZUL[1], alpha=0.5, linewidth=0)
        ax.plot(cv.k, cv.perplejidad_media, color=AZUL[4], marker="o", markersize=6)
        kf = etiquetas.topic.nunique() if etiquetas is not None else None
        if kf in set(cv.k):
            yk = cv.set_index("k").perplejidad_media[kf]
            ax.scatter([kf], [yk], s=110, facecolor=NIVEL_COLOR["P"], edgecolor=SUPERFICIE, linewidth=2, zorder=5)
            ax.annotate(f"k = {kf} (chosen)", (kf, yk), xytext=(8, 10), textcoords="offset points", color=TINTA2)
        ax.set_xticks(cv.k); ax.set_xlabel("Number of topics (k)"); ax.set_ylabel("Held-out perplexity")
        ax.set_title("Model selection: five-fold cross-validated perplexity")
        html_ += figura(fig_a_html(fig, "f01_seleccion_k"), "Figura 1. Perplejidad media (±1 DE) por número de tópicos.")
        mejor = int(cv.loc[cv.perplejidad_media.idxmin(), "k"])
        items.append(f"La perplejidad más baja se obtiene con k = {mejor}. El método fija k donde la mejora se estabiliza "
                     "y lo contrasta con la interpretabilidad de los tópicos [D1].")
    if top is not None and len(top):
        K = top.topic.nunique(); ncol = 4; nfil = int(np.ceil(K / ncol))
        fig, axs = plt.subplots(nfil, ncol, figsize=(13, 2.7 * nfil), squeeze=False)
        for t, ax in zip(sorted(top.topic.unique()), axs.ravel()):
            d = top[top.topic == t].nlargest(10, "beta").iloc[::-1]
            ax.barh(d.term, d.beta, color=AZUL[4], height=0.7)
            ax.set_title(etiqueta_topico(t, etiquetas, 30), fontsize=9); ax.tick_params(labelsize=7.5)
            ax.grid(axis="y", visible=False); ax.set_xticks([])
        for ax in axs.ravel()[K:]:
            ax.axis("off")
        fig.suptitle("Ten most probable terms per topic (β)", x=0.01, ha="left", fontweight="bold")
        fig.tight_layout()
        html_ += figura(fig_a_html(fig, "f02_terminos"), "Figura 2. Términos con mayor β en cada tópico. Úsalos, junto con "
                        "t04_docs_representativos, para nombrar los tópicos [D3].")
    return html_ + lectura(items)


@seccion("rq1a", "RQ1a · Narrativas por nivel y validación de la tipología", "Método 3.10–3.11", "¿Qué narrativas estructuran los documentos en cada nivel de gobernanza?")
def _():
    at, kw = obj("actor_topico"), tabla("t05_RQ1a_topicos_por_nivel")
    html_, items = "", []
    if at is not None:
        m = at.groupby(["GOV_CODE", "topic"]).gamma.mean().unstack()
        m = m.reindex([c for c in ("G", "P", "U") if c in m.index])
        fig, ax = plt.subplots(figsize=(11, 3.4))
        heatmap(ax, m.values, [NIVEL_CORTO[c] for c in m.index], [etiqueta_topico(t, etiquetas, 26) for t in m.columns],
                CMAP_SEQ, vmin=0, cbar_label="Mean γ (actors)")
        ax.set_title("Thematic profile of each level (mean topic proportion across actors)")
        html_ += figura(fig_a_html(fig, "f03_perfil_nivel"), "Figura 3. Proporción media de cada tópico por nivel (unidad: actor).")
    if kw is not None and len(kw):
        for r in kw.sort_values("epsilon2", ascending=False).itertuples():
            medias = {c: getattr(r, f"gamma_media_{c}", np.nan) for c in ("G", "P", "U")}
            alto = max(medias, key=lambda c: -1 if not np.isfinite(medias[c]) else medias[c])
            sig = "con diferencias significativas" if np.isfinite(r.p_BH) and r.p_BH < .05 else "sin diferencias significativas"
            items.append(f"<b>{_html.escape(etiqueta_topico(r.topic, etiquetas, 60))}</b>: más presente en "
                         f"{NIVEL_ES[alto]} (γ medio {fn(medias[alto])}); {sig} entre niveles "
                         f"(H = {fn(r.H)}, {fp(r.p_BH)} BH; ε² = {fn(r.epsilon2)}, efecto {mag_eps2(r.epsilon2)}).")
        html_ += tabla_html(kw, ["TOPIC_LABEL", "gamma_media_G", "gamma_media_P", "gamma_media_U", "H", "p_BH", "epsilon2"],
                            renombrar={"TOPIC_LABEL": "Tópico", "gamma_media_G": "γ G", "gamma_media_P": "γ P",
                                       "gamma_media_U": "γ U", "p_BH": "p (BH)", "epsilon2": "ε²"},
                            nota="Kruskal-Wallis por tópico entre niveles, con el actor como unidad.")
    tp = tabla("t05_tipo_perfil_por_nivel")
    if tp is not None and len(tp):
        orden = ["Diffuse", "Monothematic", "Bithematic", "Multithematic"]
        piv = tp.pivot(index="GOVERNANCE", columns="PROFILE_TYPE", values="prop").reindex(columns=orden).fillna(0)
        piv = piv.loc[[g for g in NIVEL_NOMBRE.values() if g in piv.index]]
        fig, ax = plt.subplots(figsize=(9, 2.8))
        izq = np.zeros(len(piv))
        for k, c in enumerate(orden):
            ax.barh([corto(g) for g in piv.index], piv[c], left=izq, color=ORDINAL[k + 1], edgecolor=SUPERFICIE, linewidth=2, label=c)
            for yy, (l, v) in enumerate(zip(izq, piv[c])):
                if v > .07:
                    ax.text(l + v / 2, yy, f"{v:.0%}", ha="center", va="center", color="white", fontsize=8)
            izq += piv[c].values
        ax.set_xlim(0, 1); ax.xaxis.set_major_formatter(matplotlib.ticker.PercentFormatter(1)); ax.grid(axis="y", visible=False)
        ax.legend(ncol=4, loc="upper center", bbox_to_anchor=(0.5, -0.18)); ax.invert_yaxis()
        ax.set_title(f"Documents by number of relevant topics (γ ≥ {UMBRAL_MULTI})")
        html_ += figura(fig_a_html(fig, "f04_tipo_perfil"), "Figura 4. Documentos monotemáticos, bitemáticos, multitemáticos y difusos por nivel.")
    val = tabla("t05_validacion_tipologias")
    mt = tabla("t05_tipologia_topico_modal")
    gt, dm = obj("gamma_td"), obj("docs_meta")
    if gt is not None and dm is not None:
        dom = dm[~dm.DOC_TYPE_H.fillna("").str.contains(r"\|") & dm.DOC_TYPE_H.fillna("").ne("")]
        m = gt.merge(dom[["CODE", "DOC_TYPE_H"]], on="CODE").groupby(["DOC_TYPE_H", "topic"]).gamma.mean().unstack()
        fig, ax = plt.subplots(figsize=(11, 0.45 * len(m) + 1.5))
        heatmap(ax, m.values, list(m.index), [etiqueta_topico(t, etiquetas, 26) for t in m.columns], CMAP_SEQ, vmin=0, cbar_label="Mean γ")
        ax.set_title("Title-based typology versus inductive topics")
        html_ += figura(fig_a_html(fig, "f05_tipologia_topico"), "Figura 5. γ medio de cada tópico según la tipología por título.")
    if val is not None and len(val):
        v = val.iloc[0]
        items.append(f"Validación de la tipología: V de Cramér = {fn(v.cramer_v)} (asociación {mag_v(v.cramer_v)}, uso descriptivo); "
                     f"el {fn(100 * v.coincidencia_obs, 0)} % de los documentos tiene el tópico de consenso de su tipo, frente a "
                     f"{fn(100 * v.coincidencia_nula_media, 0)} % esperado al azar ({fp(v.p_perm)}, permutación). "
                     f"{int(v.n_discrepantes)} documentos discrepantes deben leerse (t05_docs_discrepantes_LEER).")
    em = tabla("t05_topicos_emergentes")
    if em is not None and len(em):
        items.append("Tópicos que no son el consenso de ningún tipo (categorías emergentes candidatas): "
                     + ", ".join(_html.escape(e) for e in em.TOPIC_LABEL) + ".")
    am = obj("actor_metricas")
    if am is not None:
        d = am[~am.ES_REFERENTE].dropna(subset=["HELLINGER_GLOBAL"])
        fig, ax = plt.subplots(figsize=(8, 3.4))
        niveles = [c for c in ("G", "P", "U") if c in set(d.GOV_CODE)]
        for k, c in enumerate(niveles):
            x = d[d.GOV_CODE == c].HELLINGER_GLOBAL.values
            ax.boxplot(x, positions=[k], widths=0.45, patch_artist=True, showfliers=False,
                       boxprops=dict(facecolor=NIVEL_COLOR[c] + "33", edgecolor=NIVEL_COLOR[c], linewidth=1.5),
                       medianprops=dict(color=NIVEL_COLOR[c], linewidth=2.5), whiskerprops=dict(color=NIVEL_COLOR[c]),
                       capprops=dict(color=NIVEL_COLOR[c]))
            ax.scatter(k + rng.uniform(-0.15, 0.15, len(x)), x, s=16, color=NIVEL_COLOR[c], alpha=0.75, edgecolor=SUPERFICIE, linewidth=0.5)
        ax.set_xticks(range(len(niveles))); ax.set_xticklabels([NIVEL_CORTO[c] for c in niveles]); ax.grid(axis="x", visible=False)
        ax.set_ylabel("Hellinger distance"); ax.set_title("Thematic distance to the global reference (worldwide organizations)")
        html_ += figura(fig_a_html(fig, "f06_hellinger"), "Figura 6. Distancia de Hellinger de cada actor al perfil medio de las "
                        "organizaciones de alcance mundial (sin incluir al referente).")
        ind = tabla("t05_RQ1a_indicadores_por_nivel")
        if ind is not None and len(ind):
            r = ind.iloc[0]
            med = d.groupby("GOV_CODE").HELLINGER_GLOBAL.median()
            items.append(f"Distancia al referente global: medianas " + ", ".join(f"{NIVEL_ES[c]} {fn(med[c])}" for c in med.index)
                         + f"; diferencia entre niveles H = {fn(r.H)}, {fp(r.p_valor)}, ε² = {fn(r.epsilon2)} ({mag_eps2(r.epsilon2)}).")
    return html_ + lectura(items)


@seccion("rq1b", "RQ1b · Fuerza normativa", "Método 3.13", "¿Con qué fuerza normativa se expresan estas narrativas?")
def _():
    fa, fd = obj("fuerza_actor"), obj("fuerza_doc")
    html_, items = "", []
    if fa is not None:
        orden = ["Absence", "Descriptive", "Voluntary or soft", "Institutionally mandatory", "Legally binding"]
        colores = ["#e1e0d9"] + ORDINAL[1:5]
        t = fa.groupby(["GOVERNANCE", fa.FUERZA_ETIQ.astype(str)]).size().unstack(fill_value=0)
        t = t.reindex(columns=[c for c in orden if c in t.columns]).fillna(0)
        t = t.loc[[g for g in NIVEL_NOMBRE.values() if g in t.index]]
        p = t.div(t.sum(axis=1), axis=0)
        fig, ax = plt.subplots(figsize=(10, 3))
        izq = np.zeros(len(p))
        for c in p.columns:
            col = colores[orden.index(c)]
            ax.barh([corto(g) for g in p.index], p[c], left=izq, color=col, edgecolor=SUPERFICIE, linewidth=2, label=c)
            for yy, (l, v) in enumerate(zip(izq, p[c])):
                if v > .07:
                    ax.text(l + v / 2, yy, f"{v:.0%}", ha="center", va="center", fontsize=8,
                            color=TINTA if c == "Absence" else "white")
            izq += p[c].values
        ax.set_xlim(0, 1); ax.xaxis.set_major_formatter(matplotlib.ticker.PercentFormatter(1)); ax.grid(axis="y", visible=False)
        ax.invert_yaxis(); ax.legend(ncol=5, loc="upper center", bbox_to_anchor=(0.5, -0.18))
        ax.set_title("Strongest instrument issued by each actor (layer 1)")
        html_ += figura(fig_a_html(fig, "f07_fuerza_capa1"), "Figura 7. Actores según el instrumento más fuerte que emitieron (incluye ausencias).")
        for g in p.index:
            oblig = p.loc[g, [c for c in ("Institutionally mandatory", "Legally binding") if c in p.columns]].sum()
            items.append(f"{es(g).capitalize()}: {oblig:.0%} de los actores tiene al menos un instrumento obligatorio "
                         f"(legalmente vinculante o institucionalmente obligatorio).")
    if fd is not None:
        d = fd.dropna(subset=["INDICE_DEONTICO"]).copy(); d["GOV"] = d.GOVERNANCE.map(GOV_A_COD)
        fig, axs = plt.subplots(1, 2, figsize=(12, 3.6))
        niveles = [c for c in ("G", "P", "U") if c in set(d.GOV)]
        for k, c in enumerate(niveles):
            x = d[d.GOV == c].INDICE_DEONTICO.values
            vp = axs[0].violinplot(x, positions=[k], widths=0.7, showextrema=False)
            for b in vp["bodies"]:
                b.set_facecolor(NIVEL_COLOR[c]); b.set_alpha(0.3); b.set_edgecolor(NIVEL_COLOR[c])
            axs[0].scatter(k + rng.uniform(-0.12, 0.12, len(x)), x, s=10, color=NIVEL_COLOR[c], alpha=0.6, linewidth=0)
            axs[0].plot([k - 0.22, k + 0.22], [np.median(x)] * 2, color=TINTA, linewidth=2.5)
        axs[0].set_xticks(range(len(niveles))); axs[0].set_xticklabels([NIVEL_CORTO[c] for c in niveles]); axs[0].grid(axis="x", visible=False)
        axs[0].set_ylim(-0.03, 1.03); axs[0].set_ylabel("Deontic index (strong / total)")
        axs[0].set_title("Deontic intensity by level (layer 2)")
        cat = fd.dropna(subset=["INDICE_DEONTICO"]).groupby("FUERZA_ETIQ").INDICE_DEONTICO.median()
        cat = cat.reindex([c for c in ["Descriptive", "Voluntary or soft", "Institutionally mandatory", "Legally binding"] if c in cat.index])
        axs[1].hlines(range(len(cat)), 0, cat.values, color=AZUL[2], linewidth=2)
        axs[1].scatter(cat.values, range(len(cat)), s=80, color=AZUL[5], zorder=3)
        for yy, v in enumerate(cat.values):
            axs[1].text(v + 0.02, yy, fn(v), va="center", fontsize=8, color=TINTA2)
        axs[1].set_yticks(range(len(cat))); axs[1].set_yticklabels(cat.index); axs[1].set_xlim(0, 1.05)
        axs[1].grid(axis="y", visible=False); axs[1].set_xlabel("Median deontic index")
        axs[1].set_title("Convergent validity: index by instrument form")
        fig.tight_layout()
        html_ += figura(fig_a_html(fig, "f08_deontico"), "Figura 8. Índice deóntico por nivel (izquierda; la línea es la mediana) "
                        "y por categoría de instrumento (derecha).")
    vc = tabla("t06_validez_convergente")
    if vc is not None and len(vc):
        r = vc.iloc[0]
        items.append(f"Validez convergente: el índice deóntico difiere entre categorías de instrumento "
                     f"(H = {fn(r.H)}, {fp(r.p_valor)}, ε² = {fn(r.epsilon2)}, {mag_eps2(r.epsilon2)}). Si los instrumentos obligatorios "
                     "no muestran un índice mayor, las dos capas miden aspectos distintos de la fuerza.")
    fz = tabla("t06_fuerza_de_las_narrativas")
    if fz is not None and len(fz):
        f = fz.sort_values("fuerza_narrativa")
        fig, ax = plt.subplots(figsize=(9, 0.42 * len(f) + 1.2))
        ax.hlines(range(len(f)), f.fuerza_narrativa.min() * 0.9, f.fuerza_narrativa, color=AZUL[1], linewidth=2)
        ax.scatter(f.fuerza_narrativa, range(len(f)), s=90, color=AZUL[5], zorder=3)
        ax.set_yticks(range(len(f))); ax.set_yticklabels([etiqueta_topico(t, etiquetas, 40) for t in f.topic]); ax.grid(axis="y", visible=False)
        ax.set_xlabel("Deontic index weighted by γ"); ax.set_title("Normative force of each narrative")
        html_ += figura(fig_a_html(fig, "f09_fuerza_narrativas"), "Figura 9. Índice deóntico medio de los documentos ponderado por su γ en cada tópico.")
        items.append(f"La narrativa con mayor fuerza deóntica es «{_html.escape(etiqueta_topico(f.topic.iloc[-1], etiquetas, 60))}» "
                     f"({fn(f.fuerza_narrativa.iloc[-1])}) y la de menor «{_html.escape(etiqueta_topico(f.topic.iloc[0], etiquetas, 60))}» "
                     f"({fn(f.fuerza_narrativa.iloc[0])}).")
    pr = tabla("t06_RQ1b_pruebas_niveles")
    if pr is not None and len(pr):
        html_ += tabla_html(pr, [c for c in ["prueba", "n", "chi2", "p_valor_sim", "cramer_v", "H", "p_valor", "epsilon2"] if c in pr.columns],
                            renombrar={"prueba": "Prueba", "p_valor_sim": "p (simulado)", "cramer_v": "V de Cramér", "p_valor": "p", "epsilon2": "ε²"})
    return html_ + lectura(items)


@seccion("rq1c", "RQ1c · Cambio tras la IA generativa", "Método 3.14", "¿Cambiaron las narrativas y la fuerza normativa después de 2023?")
def _():
    gt, dm, tt = obj("gamma_td"), obj("docs_meta"), tabla("t07_RQ1c_topicos_post_vs_pre")
    html_, items = "", []
    if gt is not None and dm is not None:
        d = gt.merge(dm[dm.YEAR.notna()][["CODE", "GOV_CODE", "YEAR"]], on="CODE")
        d["rel"] = d.gamma >= UMBRAL_MULTI
        s = d.groupby(["GOV_CODE", "YEAR", "topic"]).rel.mean().reset_index()
        K = d.topic.nunique(); ncol = 4; nfil = int(np.ceil(K / ncol))
        fig, axs = plt.subplots(nfil, ncol, figsize=(13, 2.6 * nfil), squeeze=False, sharey=True)
        for t, ax in zip(sorted(d.topic.unique()), axs.ravel()):
            for c in ("G", "P", "U"):
                x = s[(s.topic == t) & (s.GOV_CODE == c)].sort_values("YEAR")
                if len(x):
                    ax.plot(x.YEAR.astype(int), x.rel, color=NIVEL_COLOR[c], linewidth=1.8, marker="o", markersize=3)
            ax.axvspan(ANIO_CORTE_GENAI - 0.5, d.YEAR.max() + 0.5, color=AZUL[0], alpha=0.35, linewidth=0)
            ax.set_title(etiqueta_topico(t, etiquetas, 30), fontsize=9); ax.tick_params(labelsize=7.5)
            ax.yaxis.set_major_formatter(matplotlib.ticker.PercentFormatter(1))
        for ax in axs.ravel()[K:]:
            ax.axis("off")
        leyenda_niveles(axs[0, 0], loc="upper left", fontsize=7)
        fig.suptitle(f"Share of documents where each topic is relevant (γ ≥ {UMBRAL_MULTI}); shaded = post-GenAI",
                     x=0.01, ha="left", fontweight="bold")
        fig.tight_layout()
        html_ += figura(fig_a_html(fig, "f10_topicos_anio"), "Figura 10. Evolución anual de cada tópico por nivel. La zona sombreada es el periodo Post-GenAI (≥ 2023).")
    if tt is not None and len(tt):
        govs = [g for g in NIVEL_NOMBRE.values() if g in set(tt.GOVERNANCE)]
        topicos = sorted(tt.topic.unique())
        fig, axs = plt.subplots(1, len(govs), figsize=(4.2 * len(govs), 0.4 * len(topicos) + 1.6), sharey=True, squeeze=False)
        for ax, g in zip(axs.ravel(), govs):
            x = tt[tt.GOVERNANCE == g].set_index("topic").reindex(topicos)
            col = color_nivel(g)
            for yy, (r, p) in enumerate(zip(x.r_rb, x.p_BH)):
                if np.isfinite(r):
                    ax.hlines(yy, 0, r, color=col, alpha=0.4, linewidth=2)
                    ax.scatter(r, yy, s=70, color=col if (np.isfinite(p) and p < .05) else SUPERFICIE, edgecolor=col, linewidth=1.8, zorder=3)
            ax.axvline(0, color=EJE, linewidth=1); ax.set_xlim(-1, 1); ax.set_title(corto(g)); ax.grid(axis="y", visible=False)
            ax.set_xlabel("r rank-biserial (Post − Pre)")
        axs[0, 0].set_yticks(range(len(topicos))); axs[0, 0].set_yticklabels([etiqueta_topico(t, etiquetas, 30) for t in topicos])
        axs[0, -1].legend(handles=[Line2D([], [], marker="o", color="none", markerfacecolor=TENUE, markeredgecolor=TENUE, markersize=8, label="p (BH) < 0.05"),
                                   Line2D([], [], marker="o", color="none", markerfacecolor=SUPERFICIE, markeredgecolor=TENUE, markersize=8, label="not significant")],
                          loc="lower right", fontsize=7.5)
        fig.suptitle("Change in topic proportions after the release of generative AI", x=0.01, ha="left", fontweight="bold")
        fig.tight_layout()
        html_ += figura(fig_a_html(fig, "f11_post_pre"), "Figura 11. Efecto del periodo en cada tópico (r biserial de rangos; positivo = más presente "
                        "Post-GenAI). Punto relleno: significativo tras BH.")
        sig = tt[tt.p_BH < .05].sort_values("r_rb", key=abs, ascending=False)
        if len(sig):
            for r in sig.head(8).itertuples():
                items.append(f"{es(r.GOVERNANCE).capitalize()}: «{_html.escape(etiqueta_topico(r.topic, etiquetas, 60))}» es "
                             f"{'más' if r.r_rb > 0 else 'menos'} frecuente después de 2023 (r = {fn(r.r_rb)}, {mag_r(r.r_rb)}; {fp(r.p_BH)} BH).")
        else:
            items.append("Ningún tópico cambia significativamente entre periodos tras la corrección BH.")
        items.append("El diseño es transversal: estas diferencias describen periodos, no cambios dentro de un mismo actor.")
    tf = tabla("t07_RQ1c_fuerza_post_vs_pre")
    if tf is not None and len(tf):
        x = tf[tf.INDICADOR == "INDICE_DEONTICO"]
        for r in x.itertuples():
            items.append(f"Índice deóntico en {es(r.GOVERNANCE)}: mediana {fn(r.mediana_2)} antes y {fn(r.mediana_1)} después "
                         f"(r = {fn(r.r_rb)}, {mag_r(r.r_rb)}; {fp(r.p_BH)} BH).")
    return html_ + lectura(items)


@seccion("tax", "Taxonomía deductiva 5D y triangulación", "Método 3.15", "¿Cómo se relacionan las narrativas inductivas con las dimensiones teóricas?")
def _():
    tr, da = tabla("t08_triangulacion"), obj("dim_actor")
    html_, items = "", []
    if tr is not None and len(tr):
        m = tr.pivot(index="topic", columns="DIMENSION", values="rho")
        fig, ax = plt.subplots(figsize=(8, 0.45 * len(m) + 1.5))
        heatmap(ax, m.values, [etiqueta_topico(t, etiquetas, 36) for t in m.index], list(m.columns), CMAP_DIV, vmin=-1, vmax=1,
                cbar_label="Spearman ρ")
        ax.set_title("Triangulation: topics (γ) × taxonomy dimensions (density)")
        html_ += figura(fig_a_html(fig, "f12_triangulacion"), "Figura 12. Correlación de Spearman entre el γ de cada tópico y la densidad de cada dimensión.")
        tp = tabla("t08_topico_a_dimension")
        if tp is not None:
            for r in tp.itertuples():
                items.append(f"«{_html.escape(etiqueta_topico(r.topic, etiquetas, 60))}» se asocia sobre todo con {r.DIMENSION} (ρ = {fn(r.rho)})"
                             + ("" if r.DIMENSION_PRINCIPAL != "Sin correspondencia" else f", por debajo del umbral ρ ≥ {RHO_MIN_TRIANG}: sin correspondencia clara") + ".")
    if da is not None:
        m = da.groupby(["DIMENSION", "GOV_CODE"]).DENSIDAD.mean().unstack().reindex(columns=["G", "P", "U"])
        fig, ax = plt.subplots(figsize=(9, 3.6))
        x = np.arange(len(m)); w = 0.26
        for k, c in enumerate([c for c in ("G", "P", "U") if c in m.columns]):
            ax.bar(x + (k - 1) * w, m[c], width=w - 0.03, color=NIVEL_COLOR[c], label=NIVEL_CORTO[c])
        ax.set_xticks(x); ax.set_xticklabels(m.index); ax.grid(axis="x", visible=False)
        ax.set_ylabel(f"Matches per {POR_DIEZ_MIL:,} words".replace(",", ".")); ax.legend()
        ax.set_title("Density of each taxonomy dimension by level (mean across actors)")
        html_ += figura(fig_a_html(fig, "f13_densidad_dimensiones"), "Figura 13. Densidad media de cada dimensión por nivel.")
    kw = tabla("t08_taxonomia_por_nivel")
    if kw is not None and len(kw):
        for r in kw.itertuples():
            items.append(f"{r.DIMENSION}: diferencia entre niveles H = {fn(r.H)}, {fp(r.p_BH)} BH, ε² = {fn(r.epsilon2)} ({mag_eps2(r.epsilon2)}).")
    cob = tabla("t08_cobertura_dimensiones")
    if cob is not None and len(cob):
        baja = cob[cob.prop_docs < 0.2]
        if len(baja):
            items.append("Cobertura baja (menos del 20 % de los documentos con coincidencias): " + ", ".join(baja.DIMENSION)
                         + ". El método indica no apoyar conclusiones en estas dimensiones [D7].")
        html_ += tabla_html(cob, renombrar={"docs_con_hits": "Docs con coincidencias", "prop_docs": "Proporción", "densidad_media": "Densidad media"},
                            nota="Cobertura del diccionario por dimensión (evidencia de validez).")
    al = tabla("t08_alineamiento_por_nivel")
    if al is not None and len(al):
        for r in al.itertuples():
            items.append(f"Distancia {r.distancia} al referente global: H = {fn(r.H)}, {fp(r.p_valor)}, ε² = {fn(r.epsilon2)} ({mag_eps2(r.epsilon2)}).")
    return html_ + lectura(items)


def dibujar_red(ax, A, nodos, titulo, disposicion="niveles"):
    """disposicion='niveles': tres bandas (organizaciones, países, universidades) y SOLO los lazos entre niveles
    (la red meso, foco de RQ2); 'fuerza': disposición por fuerzas con todos los lazos."""
    G = nx.from_numpy_array(A)
    ids = nodos.ID.tolist()
    niv = nodos.NIVEL.to_numpy(dtype=object)
    grado = A.sum(axis=1)
    if disposicion == "niveles":
        pos = {}
        for fila, c in enumerate(("G", "P", "U")):
            idx = [i for i in range(len(ids)) if niv[i] == c]
            orden = sorted(idx, key=lambda i: (str(nodos.REGION.iloc[i]), str(nodos.ACTOR.iloc[i])))
            n_ = max(len(orden), 1); filas_banda = 1 if n_ <= 45 else 2
            for k, i in enumerate(orden):
                desplaz = (k % filas_banda - (filas_banda - 1) / 2) * 0.16
                pos[i] = ((k + 0.5) / n_, 2 - fila + desplaz)
        aristas = [(a, b) for a, b in G.edges() if niv[a] != niv[b]]
        if aristas:
            nx.draw_networkx_edges(G, pos, edgelist=aristas, ax=ax, edge_color=AZUL[3], alpha=0.22, width=0.7)
        grado_m = np.array([sum(1 for a, b in aristas if i in (a, b)) for i in range(len(ids))])
    else:
        H = G.copy(); H.remove_nodes_from(list(nx.isolates(H)))
        pos = nx.spring_layout(H, seed=SEMILLA, k=1.8 / np.sqrt(max(H.number_of_nodes(), 1)), iterations=300)
        aislados = [i for i in G.nodes if i not in pos]
        for k, i in enumerate(aislados):                       # aislados en una fila al pie
            pos[i] = (-1.05 + 2.1 * (k + 0.5) / max(len(aislados), 1), -1.25)
        aristas = list(G.edges())
        mismo = [niv[a] == niv[b] for a, b in aristas]
        for flag, col, alfa in ((True, TENUE, 0.14), (False, AZUL[3], 0.3)):
            sel = [e for e, m in zip(aristas, mismo) if m == flag]
            nx.draw_networkx_edges(G, pos, edgelist=sel, ax=ax, edge_color=col, alpha=alfa, width=0.6)
        grado_m = grado
        if aislados:
            ax.text(0, -1.38, f"{len(aislados)} actors without ties", ha="center", fontsize=7.5, color=TENUE)
    tam = 16 + 140 * (grado_m / grado_m.max() if grado_m.max() > 0 else grado_m)
    for c in ("G", "P", "U"):
        idx = [i for i in range(len(ids)) if niv[i] == c]
        nx.draw_networkx_nodes(G, pos, nodelist=idx, ax=ax, node_size=tam[idx], node_color=NIVEL_COLOR[c],
                               edgecolors=SUPERFICIE, linewidths=0.8)
    if disposicion == "niveles":
        for fila, c in enumerate(("G", "P", "U")):
            ax.text(-0.015, 2 - fila, NIVEL_CORTO[c], ha="right", va="center", fontsize=9, color=TINTA2, fontweight="bold")
        for i in np.argsort(-grado_m)[:5]:
            if grado_m[i] > 0 and niv[i] == "G":
                ax.annotate(str(nodos.ACTOR.iloc[i])[:24], pos[i], xytext=(0, 10), textcoords="offset points", ha="center",
                            fontsize=7, color=TINTA2, bbox=dict(boxstyle="round,pad=0.15", fc=SUPERFICIE, ec="none", alpha=0.85))
        ax.set_xlim(-0.16, 1.02); ax.set_ylim(-0.45, 2.4)
    ax.set_title(titulo); ax.axis("off")


def dibujar_bipartita(ax, B, nodos, artefactos, titulo):
    """Red de afiliación actor–artefacto: útil sobre todo cuando el backbone queda vacío."""
    B = np.asarray(B); n, m = B.shape
    G = nx.Graph()
    for i, j in zip(*np.nonzero(B)):
        G.add_edge(int(i), n + int(j))
    if G.number_of_edges() == 0:
        ax.text(0.5, 0.5, "No affiliations", ha="center"); ax.axis("off"); return
    art = [n + j for j in range(m) if (n + j) in G]
    fijos = {a: (np.cos(2 * np.pi * k / len(art)), np.sin(2 * np.pi * k / len(art))) for k, a in enumerate(art)}
    pos = nx.spring_layout(G, pos=fijos, fixed=art, seed=SEMILLA, k=0.35, iterations=300)
    for v in list(pos):                                    # los actores quedan dentro del círculo de artefactos
        x, y = pos[v]; r = np.hypot(x, y)
        if v not in fijos and r > 0.92:
            pos[v] = (x / r * 0.92, y / r * 0.92)
    nx.draw_networkx_edges(G, pos, ax=ax, edge_color=TENUE, alpha=0.12, width=0.5)
    niv = nodos.NIVEL.to_numpy(dtype=object)
    for c in ("G", "P", "U"):
        idx = [i for i in range(n) if niv[i] == c and i in G]
        nx.draw_networkx_nodes(G, pos, nodelist=idx, ax=ax, node_size=24, node_color=NIVEL_COLOR[c], edgecolors=SUPERFICIE, linewidths=0.5)
    gr = B.sum(axis=0)
    nx.draw_networkx_nodes(G, pos, nodelist=art, ax=ax, node_shape="s", node_color=TINTA,
                           node_size=[70 + 260 * gr[a - n] / max(gr.max(), 1) for a in art], edgecolors=SUPERFICIE, linewidths=1)
    for a in art:
        x, y = pos[a]
        ax.annotate(str(artefactos[a - n])[:28], (x, y), xytext=(14 * np.sign(x or 1), 10 * np.sign(y or 1)), textcoords="offset points",
                    ha="left" if x >= 0 else "right", fontsize=7.5, color=TINTA, fontweight="bold",
                    bbox=dict(boxstyle="round,pad=0.2", fc=SUPERFICIE, ec="none", alpha=0.9))
    vacios = [artefactos[j] for j in range(m) if (n + j) not in G]
    if vacios:
        ax.text(0.5, -0.02, "Without actors: " + "; ".join(map(str, vacios))[:140], transform=ax.transAxes, ha="center",
                fontsize=7.5, color=TENUE)
    ax.set_xlim(-1.6, 1.6); ax.set_ylim(-1.25, 1.25); ax.set_aspect("equal")
    ax.set_title(titulo); ax.axis("off")


@seccion("rq2", "RQ2 · Alineamiento multinivel: referencias y red", "Método 3.17", "¿Cómo se alinean los niveles entre sí y con los instrumentos globales?")
def _():
    red = obj("red_multinivel")
    html_, items = "", []
    if red is None:
        raise FileNotFoundError("Falta el Paso 10")
    nodos = red["nodos"].reset_index(drop=True)
    leg = [Line2D([], [], marker="o", color="none", markerfacecolor=NIVEL_COLOR[c], markeredgecolor=SUPERFICIE, markersize=9,
                  label=NIVEL_CORTO[c]) for c in ("G", "P", "U")]
    for rel, A in (("tematica", red["A_tema"]), ("normativa", red.get("A_norm"))):
        if A is None:
            continue
        nombre = "thematic" if rel == "tematica" else "normative"
        lazos = int(A.sum() // 2)
        if lazos:
            fig, axs = plt.subplots(1, 2, figsize=(15, 6.2), gridspec_kw={"width_ratios": [1.25, 1]})
            entre = int(sum(1 for a, b in zip(*np.nonzero(np.triu(A))) if red["nodos"].NIVEL.iloc[a] != red["nodos"].NIVEL.iloc[b]))
            dibujar_red(axs[0], A, nodos, f"Cross-level ties only · {entre} of {lazos}", "niveles")
            dibujar_red(axs[1], A, nodos, f"All ties · force-directed layout", "fuerza")
            fig.legend(handles=leg + [Line2D([], [], color=AZUL[3], label="cross-level tie"), Line2D([], [], color=TENUE, label="within-level tie")],
                       loc="lower center", ncol=5, fontsize=8.5, bbox_to_anchor=(0.5, -0.02))
            fig.suptitle(f"Backbone of {nombre} alignment (SDSM, α = {ALFA_BACKBONE})", x=0.01, ha="left", fontweight="bold")
            fig.tight_layout()
            html_ += figura(fig_a_html(fig, f"f14_red_{rel}"), f"Figura 14{'a' if rel == 'tematica' else 'b'}. Red de alineamiento {rel} "
                            "(backbone SDSM). Izquierda: solo los lazos entre niveles (red meso), con organizaciones, países y universidades en bandas "
                            "ordenadas por región; tamaño = número de lazos entre niveles. Derecha: todos los lazos en disposición por fuerzas; "
                            "azul = lazo entre niveles, gris = dentro del nivel.")
            items.append(f"Red {rel}: {lazos} lazos de alineamiento significativos.")
        else:
            items.append(f"<b>Red {rel} sin lazos</b>: con el número de artefactos de esta relación, ningún par de actores puede "
                         f"superar lo esperado por el SDSM con α = {ALFA_BACKBONE} (ver reporte de calidad, «p mínimo SDSM»). Es un límite "
                         "del diseño, no un resultado; se muestra la red de afiliación como alternativa.")
        B = red["B_tema"] if rel == "tematica" else red.get("B_norm")
        if B is not None:
            cols = list(B.columns)
            if rel == "tematica" and etiquetas is not None:
                cols = [etiqueta_topico(int(c[1:]), etiquetas, 24) if str(c).startswith("T") else c for c in cols]
            elif rel == "normativa":
                cols = [str(c).split("|")[-1] for c in cols]
            if B.shape[1] <= 40:
                fig, ax = plt.subplots(figsize=(11, 8.5))
                dibujar_bipartita(ax, B.values, nodos, cols, f"Affiliation network: actors × {'topics' if rel == 'tematica' else 'subdimensions'}")
                ax.legend(handles=leg + [Line2D([], [], marker="s", color="none", markerfacecolor=TINTA, markersize=9,
                                                label="topic" if rel == "tematica" else "subdimension")], loc="upper left", bbox_to_anchor=(1.0, 1.0))
                html_ += figura(fig_a_html(fig, f"f15_afiliacion_{rel}"), f"Figura 15{'a' if rel == 'tematica' else 'b'}. Red bipartita de afiliación "
                                f"({'γ medio > 1/k' if rel == 'tematica' else 'densidad > mediana'}): cada actor se une a los artefactos que lo caracterizan. "
                                "Los cuadrados más grandes concentran más actores.")
    cf = tabla("t10_RQ2_configuraciones_nulos")
    if cf is not None and len(cf):
        claves = ["cascada (u-g, p-g, u-p)", "relevo_nacional (p-g, u-p, sin u-g)", "paralelo (u-g, p-g, sin u-p)",
                  "BYPASS_metodo (u-g sin u-p)", "desacoplado"]
        nombres = ["Cascade", "National relay", "Parallel", "Bypass", "Decoupled"]
        principal = cf[cf.NULO.str.startswith("lazos_por_bloque")]
        rels = list(principal.RELACION.unique())
        fig, axs = plt.subplots(1, len(rels), figsize=(6 * len(rels), 3.4), squeeze=False, sharey=True)
        for ax, rel in zip(axs.ravel(), rels):
            d = principal[principal.RELACION == rel].set_index("ESTADISTICO").reindex(claves)
            x = np.arange(len(claves))
            ax.bar(x - 0.18, d.observado, width=0.34, color=AZUL[4], label="Observed")
            ax.bar(x + 0.18, d.media_nula, width=0.34, color=AZUL[1], label="Null mean")
            ax.errorbar(x + 0.18, d.media_nula, yerr=d.de_nula, fmt="none", ecolor=TINTA2, capsize=3, linewidth=1)
            ax.set_xticks(x); ax.set_xticklabels(nombres, rotation=20, ha="right"); ax.grid(axis="x", visible=False)
            ax.yaxis.set_major_formatter(matplotlib.ticker.PercentFormatter(1)); ax.set_title(f"{rel.capitalize()} relation")
            for xx, (o, p) in enumerate(zip(d.observado, d.p_superior)):
                if np.isfinite(p) and p < .05:
                    ax.text(xx - 0.18, o, "*", ha="center", va="bottom", fontsize=13, color=TINTA)
        axs[0, 0].set_ylabel("Share of university–organization pairs"); axs[0, 0].legend()
        fig.suptitle("University–country–organization configurations versus the block-preserving null", x=0.01, ha="left", fontweight="bold")
        fig.tight_layout()
        html_ += figura(fig_a_html(fig, "f16_configuraciones"), "Figura 16. Configuraciones observadas frente al nulo que conserva los lazos de cada bloque "
                        "(* = más frecuente que el nulo, p < 0,05).")
        for r in principal[principal.ESTADISTICO.isin(claves)].itertuples():
            if np.isfinite(r.p_superior) and (r.p_superior < .05 or r.p_inferior < .05):
                items.append(f"Relación {r.RELACION}: «{nombres[claves.index(r.ESTADISTICO)]}» es "
                             f"{'más' if r.p_superior < .05 else 'menos'} frecuente que lo esperado ({fn(100 * r.observado, 1)} % vs "
                             f"{fn(100 * r.media_nula, 1)} %).")
    ca = tabla("t10_CA_coordenadas")
    if ca is not None and len(ca):
        inr = tabla("t10_CA_inercias")
        fig, ax = plt.subplots(figsize=(9, 6.5))
        for c in ("G", "P", "U"):
            d = ca[(ca.TIPO == "actor") & (ca.NIVEL == c)]
            ax.scatter(d.Dim1, d.Dim2, s=26, color=NIVEL_COLOR[c], alpha=0.75, edgecolor=SUPERFICIE, linewidth=0.5, label=NIVEL_CORTO[c])
        d = ca[ca.TIPO == "artefacto"]
        ax.scatter(d.Dim1, d.Dim2, marker="D", s=40, color=TINTA, label="topic / subdimension")
        for e, x, y in zip(d.ELEMENTO, d.Dim1, d.Dim2):
            et = etiqueta_topico(int(e[1:]), etiquetas, 22) if str(e).startswith("T") and str(e)[1:].isdigit() else str(e).split("|")[-1][:22]
            ax.annotate(et, (x, y), xytext=(4, 4), textcoords="offset points", fontsize=7, color=TINTA)
        ax.axhline(0, color=EJE, linewidth=0.8); ax.axvline(0, color=EJE, linewidth=0.8)
        if inr is not None and len(inr) > 1:
            ax.set_xlabel(f"Dimension 1 ({inr.prop.iloc[0]:.0%} of inertia)"); ax.set_ylabel(f"Dimension 2 ({inr.prop.iloc[1]:.0%})")
        ax.legend(loc="best"); ax.set_title("Joint space of actors and artifacts (correspondence analysis)")
        html_ += figura(fig_a_html(fig, "f17_espacio_conjunto"), "Figura 17. Análisis de correspondencias de la matriz actor × artefacto. "
                        "Actores cercanos comparten artefactos; un nivel concentrado en una zona indica un repertorio propio.")
    ds = tabla("t10_RQ2_descripcion_bloques")
    if ds is not None and len(ds):
        html_ += tabla_html(ds, renombrar={"bloque": "Bloque", "n_nodos": "Nodos", "lazos": "Lazos", "densidad": "Densidad",
                                           "centralizacion_grado": "Centralización", "transitividad": "Transitividad",
                                           "componentes": "Componentes", "RELACION": "Relación"}, n=30,
                            nota="Descripción por bloque. Las proyecciones bipartitas inflan la transitividad (limitación 8).")
    cr = tabla("t09_RQ2_cascada_referencias")
    if cr is not None and len(cr):
        for r in cr[cr.NULO == "region"].itertuples():
            items.append(f"Referencias explícitas ({r.ESTADISTICO}): observado {fn(r.observado, 3)} frente a {fn(r.media_nula, 3)} en el nulo "
                         f"regional (p superior = {fn(r.p_superior, 3)}).")
    items.append("Los lazos se infieren de similitud textual: hablan de alineamiento, no de influencia ni de difusión.")
    return html_ + lectura(items)


@seccion("rq3", "RQ3 · Balance Norte / Sur", "Método 3.12, 3.16 y 3.17 (RQ3b)", "¿Existen diferencias entre actores del Norte y del Sur global en énfasis, fuerza y alineamiento?")
def _():
    ns, rob = tabla("t11_RQ3_norte_sur"), tabla("t11_RQ3_robustez_clasificacion")
    html_, items = "", []
    if ns is None or ns.empty:
        raise FileNotFoundError("Falta el Paso 11")
    d = ns[(ns.CLASIFICACION == "BLOC") & ns.r_rb.notna() & (ns.FAMILIA != "a_taxonomia_sub")].merge(
        rob[["GOVERNANCE", "FAMILIA", "INDICADOR", "ROBUSTO"]], on=["GOVERNANCE", "FAMILIA", "INDICADOR"], how="left")
    fam_nombre = {"a_tematica": "Thematic emphasis", "a_taxonomia": "Taxonomy dimensions", "b_fuerza": "Normative force",
                  "c_alineamiento": "Alignment", "c_red": "Network position"}

    def nombre_ind(i):
        if i.startswith("T") and i[1:].isdigit():
            return etiqueta_topico(int(i[1:]), etiquetas, 30)
        return i.replace("_", " ").replace("con ausencia", "(incl. absence)").capitalize()[:32]

    govs = [g for g in ("National (countries)", "Institutional (universities)") if g in set(d.GOVERNANCE)]
    inds = d.sort_values(["FAMILIA", "INDICADOR"])[["FAMILIA", "INDICADOR"]].drop_duplicates()
    fig, axs = plt.subplots(1, len(govs), figsize=(6.2 * len(govs), 0.28 * len(inds) + 2), sharey=True, squeeze=False)
    yt = list(range(len(inds)))
    for ax, g in zip(axs.ravel(), govs):
        x = d[d.GOVERNANCE == g].set_index(["FAMILIA", "INDICADOR"]).reindex(list(inds.itertuples(index=False, name=None)))
        for yy, (r, p, rb) in enumerate(zip(x.r_rb, x.p_BH, x.ROBUSTO)):
            if not np.isfinite(r):
                continue
            col = BLOC_COLOR["North"] if r > 0 else BLOC_COLOR["South"] if r < 0 else TENUE
            ax.hlines(yy, 0, r, color=col, alpha=0.45, linewidth=2)
            if rb is True or rb == 1:
                ax.scatter(r, yy, s=70, color=col, edgecolor=TINTA, linewidth=1.2, zorder=4)
            elif np.isfinite(p) and p < .05:
                ax.scatter(r, yy, s=60, color=col, zorder=3)
            else:
                ax.scatter(r, yy, s=40, color=SUPERFICIE, edgecolor=col, linewidth=1.5, zorder=3)
        ax.axvline(0, color=EJE, linewidth=1); ax.set_xlim(-1, 1); ax.set_title(corto(g)); ax.grid(axis="y", visible=False)
        ax.set_xlabel("← higher in South      r rank-biserial      higher in North →")
        prev = None
        for yy, fam in enumerate(inds.FAMILIA):
            if fam != prev and yy:
                ax.axhline(yy - 0.5, color=REJILLA, linewidth=1)
            prev = fam
    axs[0, 0].set_yticks(yt); axs[0, 0].set_yticklabels([nombre_ind(i) for i in inds.INDICADOR], fontsize=7.5); axs[0, 0].invert_yaxis()
    axs[0, -1].legend(handles=[Line2D([], [], marker="o", color="none", markerfacecolor=TENUE, markeredgecolor=TINTA, markersize=8, label="robust (both classifications)"),
                               Line2D([], [], marker="o", color="none", markerfacecolor=TENUE, markeredgecolor=TENUE, markersize=7, label="main classification only"),
                               Line2D([], [], marker="o", color="none", markerfacecolor=SUPERFICIE, markeredgecolor=TENUE, markersize=7, label="not significant")],
                      loc="lower right", fontsize=7.5)
    fig.suptitle("North–South differences within countries and universities", x=0.01, ha="left", fontweight="bold")
    fig.tight_layout()
    html_ += figura(fig_a_html(fig, "f18_norte_sur"), "Figura 18. Diferencias Norte/Sur por indicador (r biserial de rangos; violeta = mayor en el "
                    "Norte, amarillo = mayor en el Sur). Borde negro: significativo y con el mismo signo con las dos clasificaciones.")
    robustos = d[d.ROBUSTO == True]
    if len(robustos):
        for r in robustos.sort_values("r_rb", key=abs, ascending=False).head(10).itertuples():
            items.append(f"{es(r.GOVERNANCE).capitalize()} · {_html.escape(nombre_ind(r.INDICADOR))}: mayor en el "
                         f"{'Norte' if r.r_rb > 0 else 'Sur'} (r = {fn(r.r_rb)}, efecto {mag_r(r.r_rb)}; {fp(r.p_BH)} BH); "
                         "se mantiene con la clasificación alternativa.")
    else:
        items.append("Ninguna diferencia Norte/Sur es robusta (significativa y con el mismo signo con ambas clasificaciones).")
    u = ns[(ns.GOVERNANCE == "Institutional (universities)") & (ns.CLASIFICACION == "BLOC") & ns.p_perm_pais.notna()]
    if len(u):
        cambian = u[(u.p_BH < .05) & (u.p_perm_pais >= .05)]
        items.append(f"Universidades: {len(cambian)} de las diferencias significativas dejan de serlo al permutar entre países "
                     "(p_perm_pais ≥ 0,05); deben leerse con cautela por la dependencia dentro de país.")
    au = tabla("t11_RQ3_ausencia")
    if au is not None:
        for r in au[au.CLASIFICACION == "BLOC"].itertuples():
            items.append(f"Ausencias en {es(r.GOVERNANCE)}: Norte {fn(100 * r.prop_ausencia_north, 1)} %, Sur "
                         f"{fn(100 * r.prop_ausencia_south, 1)} % (Fisher {fp(r.p_fisher)}).")
    hb = tabla("t10_RQ3b_homofilia_bloc")
    if hb is not None:
        for r in hb[hb.RELACION != "multiplex"].itertuples():
            items.append(f"Homofilia de bloc ({r.RELACION}, {NIVEL_ES.get(r.NIVEL, r.NIVEL)}): {fn(100 * r.prop_mismo_bloc_obs, 0)} % "
                         f"de los lazos unen actores del mismo bloc frente a {fn(100 * r.prop_mismo_bloc_nula, 0)} % esperado "
                         f"(p = {fn(r.p_superior, 3)}).")
    html_ += tabla_html(rob[rob.ROBUSTO == True] if rob is not None else None,
                        renombrar={"r_rb_BLOC": "r (principal)", "r_rb_BLOC_SENS": "r (alternativa)", "p_BH_BLOC": "p BH (principal)",
                                   "p_BH_BLOC_SENS": "p BH (alternativa)"}, nota="Diferencias robustas a la clasificación Norte/Sur.")
    return html_ + lectura(items)


@seccion("rob", "Robustez", "Método 3.18")
def _():
    res, r1 = tabla("t12_resumen_robustez"), tabla("t12_R1_estabilidad_topicos")
    html_, items = "", []
    if res is not None and len(res):
        r = res.sort_values("prop_contrastes_mismo_signo")
        fig, ax = plt.subplots(figsize=(9, 0.5 * len(r) + 1.3))
        ax.barh(r.ANALISIS, r.prop_contrastes_mismo_signo, color=AZUL[4], height=0.55)
        for yy, v in enumerate(r.prop_contrastes_mismo_signo):
            ax.text(v + 0.01, yy, f"{v:.0%}", va="center", fontsize=8, color=TINTA2)
        ax.set_xlim(0, 1.08); ax.xaxis.set_major_formatter(matplotlib.ticker.PercentFormatter(1)); ax.grid(axis="y", visible=False)
        ax.set_title("North–South topic contrasts that keep the sign of the main analysis")
        html_ += figura(fig_a_html(fig, "f19_robustez"), "Figura 19. Proporción de contrastes Norte/Sur con el mismo signo que el análisis principal.")
        for x in r.itertuples():
            items.append(f"{_html.escape(x.ANALISIS)}: {x.prop_contrastes_mismo_signo:.0%} de los contrastes conserva el signo.")
    if r1 is not None and len(r1):
        s = r1.groupby("MODELO").agg(coseno=("coseno_max", "mean"), rho=("rho_hellinger_actor", "first")).reset_index()
        html_ += tabla_html(s, renombrar={"MODELO": "Modelo alternativo", "coseno": "Coseno medio con el tópico más parecido",
                                          "rho": "ρ distancia al referente"}, nota="R1: estabilidad de los tópicos frente a semillas y k ± 2.")
        items.append(f"Estabilidad de los tópicos: coseno medio entre {fn(s.coseno.min())} y {fn(s.coseno.max())}; valores cercanos a 1 indican "
                     "que los tópicos se reproducen con otras semillas y otro k.")
    return html_ + lectura(items)


@seccion("articulo", "Resumen de resultados para el artículo (AI & SOCIETY)", "Borrador en inglés para la sección Results")
def _():
    """Redacta, solo con los números de las tablas, un borrador breve de la sección de resultados por pregunta de
    investigación, cuenta sus palabras frente al límite y propone qué figuras y tablas van en el texto principal."""
    e2 = lambda x: "n/a" if x is None or not np.isfinite(x) else f"{x:.2f}"
    ep = lambda p: "n/a" if p is None or not np.isfinite(p) else ("p < .001" if p < .001 else f"p = {p:.3f}".replace("0.", "."))
    mag_en = {"trivial": "negligible", "pequeño": "small", "moderado": "moderate", "grande": "large", "sin dato": "undetermined",
              "pequeña": "small", "moderada": "moderate", "fuerte": "strong"}
    adv_en = {"trivial": "negligibly", "pequeña": "weakly", "moderada": "moderately", "fuerte": "strongly", "sin dato": ""}
    NIV_EN = {"G": "organizations", "P": "countries", "U": "universities"}
    lvl = lambda g: NIV_EN.get(GOV_A_COD.get(g, g), str(g))
    top = lambda t: etiqueta_topico(t, etiquetas, 80)
    par = {}

    # Corpus
    reg, cor, act = obj("registros"), obj("corpus"), obj("actores_estado")
    if reg is not None and cor is not None and act is not None:
        n_doc = int((~reg.SIN_DOCUMENTO).sum())
        par["Corpus"] = (f"The corpus comprises {len(reg)} records ({', '.join(f'{int((reg.GOV_CODE == c).sum())} {NIV_EN[c]}' for c in ('G', 'P', 'U'))}); "
                         f"{n_doc} have an available document and {len(cor)} unique texts entered the models. "
                         f"{int(act.AUSENCIA.sum())} actors had no locatable document and are reported as absences.")
        pend = tabla("t02_PENDIENTES_sin_txt")
        if pend is not None:
            par["Corpus"] += f" [PENDING: {len(pend)} records still lack a text file.]"

    # RQ1a
    kw, val, ind = tabla("t05_RQ1a_topicos_por_nivel"), tabla("t05_validacion_tipologias"), tabla("t05_RQ1a_indicadores_por_nivel")
    if kw is not None:
        sig = kw[kw.p_BH < .05].sort_values("epsilon2", ascending=False)
        frases = []
        for r in sig.head(3).itertuples():
            med = {c: getattr(r, f"gamma_media_{c}", np.nan) for c in ("G", "P", "U")}
            alto = max(med, key=lambda c: -1 if not np.isfinite(med[c]) else med[c])
            frases.append(f"{top(r.topic)} (highest among {NIV_EN[alto]}, mean γ = {e2(med[alto])}; ε² = {e2(r.epsilon2)})")
        txt = (f"LDA identified {kw.topic.nunique()} narratives, {('all' if len(sig) == kw.topic.nunique() else len(sig))} of which differed significantly across levels after "
               f"Benjamini-Hochberg correction")
        txt += (": the largest differences concerned " + "; ".join(frases) + ".") if frases else "."
        if val is not None:
            v = val.iloc[0]
            txt += (f" The title-based typology was {adv_en.get(mag_v(v.cramer_v), '')} associated with dominant topics "
                    f"(Cramér's V = {e2(v.cramer_v)}); {v.coincidencia_obs:.0%} of documents matched the consensus topic of their type, "
                    f"against {v.coincidencia_nula_media:.0%} expected by chance ({ep(v.p_perm)}).")
        am = obj("actor_metricas")
        if am is not None and ind is not None:
            d = am[~am.ES_REFERENTE].dropna(subset=["HELLINGER_GLOBAL"]); med = d.groupby("GOV_CODE").HELLINGER_GLOBAL.median()
            r = ind.iloc[0]
            txt += (f" Median Hellinger distance to the global reference was " + ", ".join(f"{e2(med[c])} for {NIV_EN[c]}" for c in med.index)
                    + f" (H = {e2(r.H)}, {ep(r.p_valor)}, ε² = {e2(r.epsilon2)}).")
        par["RQ1a · Narratives across levels"] = txt

    # RQ1b
    fa, fd, vc = obj("fuerza_actor"), obj("fuerza_doc"), tabla("t06_validez_convergente")
    if fa is not None and fd is not None:
        f_ = fa.copy(); f_["OBLIG"] = f_.FUERZA_MAX >= 2
        ob = f_.groupby("GOV_CODE").OBLIG.mean()
        mi = fd.dropna(subset=["INDICE_DEONTICO"]).groupby("GOVERNANCE").INDICE_DEONTICO.median()
        txt = ("The share of actors with at least one mandatory instrument (legally binding or institutionally mandatory) was "
               + ", ".join(f"{ob[c]:.0%} among {NIV_EN[c]}" for c in ob.index if c in NIV_EN) + ". "
               + "Median deontic indices were " + ", ".join(f"{e2(v)} for {lvl(g)}" for g, v in mi.items()) + ".")
        pr = tabla("t06_RQ1b_pruebas_niveles")
        if pr is not None:
            x = pr[pr.prueba.str.startswith("Índice deóntico")]
            if len(x):
                r = x.iloc[0]
                txt += f" Differences in the deontic index across levels: H = {e2(r.H)}, {ep(r.p_valor)}, ε² = {e2(r.epsilon2)}."
        if vc is not None:
            r = vc.iloc[0]
            txt += (f" Convergent validity between layers was {mag_en.get(mag_eps2(r.epsilon2), '')} "
                    f"(Kruskal-Wallis across instrument categories, ε² = {e2(r.epsilon2)}, {ep(r.p_valor)}).")
        par["RQ1b · Normative force"] = txt

    # RQ1c
    tt = tabla("t07_RQ1c_topicos_post_vs_pre")
    if tt is not None:
        sig = tt[tt.p_BH < .05].sort_values("r_rb", key=abs, ascending=False)
        if len(sig):
            txt = "After 2023, " + "; ".join(f"{top(r.topic)} became {'more' if r.r_rb > 0 else 'less'} frequent among {lvl(r.GOVERNANCE)} "
                                            f"(r = {e2(r.r_rb)}, {ep(r.p_BH)})" for r in sig.head(4).itertuples()) + "."
        else:
            txt = "No topic proportion differed significantly between the pre- and post-generative-AI periods after correction."
        txt += " Because the design is cross-sectional, these differences describe periods rather than change within actors."
        par["RQ1c · Change after generative AI"] = txt

    # Taxonomía
    tp, kwd = tabla("t08_topico_a_dimension"), tabla("t08_taxonomia_por_nivel")
    if tp is not None:
        ok = tp[tp.DIMENSION_PRINCIPAL != "Sin correspondencia"]
        txt = (f"Triangulation linked {len(ok)} of {len(tp)} topics to a taxonomy dimension (Spearman ρ ≥ {RHO_MIN_TRIANG}): "
               + "; ".join(f"{top(r.topic)}–{r.DIMENSION} (ρ = {e2(r.rho)})" for r in ok.itertuples()) + ".")
        if kwd is not None:
            s = kwd[kwd.p_BH < .05]
            txt += f" {len(s)} of {len(kwd)} dimensions differed in density across levels" + (
                (": " + ", ".join(f"{r.DIMENSION} (ε² = {e2(r.epsilon2)})" for r in s.itertuples()) + ".") if len(s) else ".")
        par["Deductive taxonomy and triangulation"] = txt

    # RQ2
    red, cf = obj("red_multinivel"), tabla("t10_RQ2_configuraciones_nulos")
    if red is not None:
        lz = {r: int(A.sum() // 2) for r, A in (("thematic", red["A_tema"]), ("normative", red.get("A_norm"))) if A is not None}
        txt = "The SDSM backbone retained " + " and ".join(f"{v} {k} alignment ties" for k, v in lz.items()) + "."
        if any(v == 0 for v in lz.values()):
            txt += (" [CHECK: an empty backbone reflects too few artifacts for the SDSM test, not an absence of alignment; "
                    "revise the tie definition before reporting.]")
        if cf is not None:
            nom = {"cascada (u-g, p-g, u-p)": "cascades", "relevo_nacional (p-g, u-p, sin u-g)": "national relays",
                   "BYPASS_metodo (u-g sin u-p)": "bypasses", "desacoplado": "decoupled pairs"}
            pr_ = cf[cf.NULO.str.startswith("lazos_por_bloque") & cf.ESTADISTICO.isin(nom)]
            s = pr_[(pr_.p_superior < .05) | (pr_.p_inferior < .05)]
            if len(s):
                txt += " Relative to the block-preserving null, " + "; ".join(
                    f"{nom[r.ESTADISTICO]} were {'over' if r.p_superior < .05 else 'under'}-represented in the {r.RELACION.replace('tematica', 'thematic').replace('normativa', 'normative')} "
                    f"relation ({r.observado:.1%} vs {r.media_nula:.1%})" for r in s.itertuples()) + "."
            else:
                txt += " No university–country–organization configuration departed from the block-preserving null."
        par["RQ2 · Multilevel alignment"] = txt

    # RQ3
    rob = tabla("t11_RQ3_robustez_clasificacion")
    if rob is not None:
        rr = rob[rob.ROBUSTO == True].sort_values("r_rb_BLOC", key=abs, ascending=False)

        def ind_en(i):
            return top(int(i[1:])) if i.startswith("T") and i[1:].isdigit() else i.replace("_", " ").lower()
        if len(rr):
            txt = (f"{len(rr)} North–South differences were robust to both classifications: "
                   + "; ".join(f"{ind_en(r.INDICADOR)} among {lvl(r.GOVERNANCE)} (higher in the {'North' if r.r_rb_BLOC > 0 else 'South'}, "
                               f"r = {e2(r.r_rb_BLOC)})" for r in rr.head(5).itertuples()) + ".")
        else:
            txt = "No North–South difference was robust to both the main and the alternative classification."
        par["RQ3 · North–South balance"] = txt

    # Robustez
    res = tabla("t12_resumen_robustez")
    if res is not None:
        par["Robustness"] = ("Across robustness checks, the share of North–South topic contrasts keeping the sign of the main analysis ranged from "
                             f"{res.prop_contrastes_mismo_signo.min():.0%} to {res.prop_contrastes_mismo_signo.max():.0%}.")

    texto_md = "\n\n".join(f"**{k}.** {v}" for k, v in par.items())
    n_pal = len(re.findall(r"\b\w+\b", texto_md))
    avisos = []
    if ETIQUETAS_TOPICOS is None:
        avisos.append("Los tópicos todavía no tienen nombre [D3]: aparecen con sus términos más probables. Nómbralos en el Paso 1 "
                      "(ETIQUETAS_TOPICOS) y vuelve a correr desde el Paso 4 antes de usar este texto.")
    tt_ = tabla("t04_top_terminos")
    if tt_ is not None and "DUDOSO" in tt_ and tt_.DUDOSO.any():
        avisos.append("Hay posibles fragmentos de palabras (restos del PDF) entre los términos más probables de algún tópico: "
                      + ", ".join(sorted(set(tt_.term[tt_.DUDOSO]))) + ". Revisa t03_terminos_dudosos.")
    if EJECUTAR_CV_K:
        avisos.append("Este informe es de la pasada 1 (decisiones). Para el artículo usa el de la pasada 2 (EJECUTAR_CV_K = False).")
    lim = globals().get("LIMITE_PALABRAS_RESULTADOS")
    estado = (f"{n_pal} palabras" + (f" de un máximo de {lim} que fijaste para Results" if lim else
              " (fija LIMITE_PALABRAS_RESULTADOS con el límite de la revista para controlarlo)"))
    figs = [("Figure 1", "f03_perfil_nivel", "RQ1a: thematic profile by level"),
            ("Figure 2", "f07_fuerza_capa1", "RQ1b: strongest instrument by level"),
            ("Figure 3", "f11_post_pre", "RQ1c: change after generative AI"),
            ("Figure 4", "f14_red_normativa" if "f14_red_normativa" in FIGURAS else "f15_afiliacion_tematica", "RQ2: multilevel alignment network"),
            ("Figure 5", "f18_norte_sur", "RQ3: North–South differences")]
    tablas_ = [("Table 1", "Corpus composition by level (section «Corpus»)"),
               ("Table 2", "Topics: top terms, labels and level differences (t05_RQ1a_topicos_por_nivel)"),
               ("Table 3", "Robust North–South differences (t11_RQ3_robustez_clasificacion)")]
    (DIR_SALIDA / "resumen_resultados_articulo.md").write_text(
        "# Results (draft generated from the pipeline tables)\n\n" + texto_md
        + "\n\n## Suggested display items\n" + "\n".join(f"- {a}: {c} ({b}.png)" for a, b, c in figs)
        + "\n" + "\n".join(f"- {a}: {c}" for a, c in tablas_) + "\n\nAll other figures and tables → Supplementary Material.\n",
        encoding="utf-8")
    html_ = "".join(f'<p class="aviso">{_html.escape(a)}</p>' for a in avisos)
    html_ += (f'<p class="nota">Borrador en inglés redactado solo con los números de las tablas ({estado}). Se guarda también como '
              "<b>resumen_resultados_articulo.md</b>. La interpretación teórica y la discusión son tuyas.</p>")
    html_ += "".join(f"<p><b>{_html.escape(k)}.</b> {_html.escape(v)}</p>" for k, v in par.items())
    html_ += ("<h4>Figuras y tablas sugeridas para el texto principal</h4><ul>"
              + "".join(f"<li>{a}: {_html.escape(c)} <span class='nota'>({b}.png)</span></li>" for a, b, c in figs)
              + "".join(f"<li>{a}: {_html.escape(c)}</li>" for a, c in tablas_)
              + "<li>El resto, al material suplementario.</li></ul>")
    return html_


@seccion("calidad", "Control de calidad", "Método 3.19")
def _():
    c = tabla("t00_reporte_calidad")
    if c is not None:
        c = c.drop_duplicates(["paso", "indicador"], keep="last")
    return tabla_html(c, n=200, renombrar={"paso": "Paso", "indicador": "Indicador", "valor": "Valor"})


# ---- 4. Ensamblar el HTML ----------------------------------------------------------------------
reg_, cor_, act_ = obj("registros"), obj("corpus"), obj("actores_estado")
tiles = [("Registros", len(reg_) if reg_ is not None else "—"), ("Textos analizados", len(cor_) if cor_ is not None else "—"),
         ("Actores", len(act_) if act_ is not None else "—"), ("Tópicos (k)", etiquetas.topic.nunique() if etiquetas is not None else "—")]
toc = "".join(f'<a href="#{s["id"]}">{_html.escape(s["titulo"])}</a>' for s in SECCIONES)
cuerpo = ""
for s in SECCIONES:
    preg = f'<p class="pregunta">{_html.escape(s["pregunta"])}</p>' if s["pregunta"] else ""
    cuerpo += (f'<section id="{s["id"]}"><h2>{_html.escape(s["titulo"])}</h2><p class="metodo">{_html.escape(s["metodo"])}</p>'
               f'{preg}{s["cuerpo"]}</section>')
CSS = """
:root{--sup:#fcfcfb;--plano:#f9f9f7;--t1:#0b0b0b;--t2:#52514e;--t3:#898781;--rej:#e1e0d9;--azul:#2a78d6;--azul-c:#eef5fe}
*{box-sizing:border-box}body{margin:0;background:var(--plano);color:var(--t1);font:15px/1.55 system-ui,-apple-system,"Segoe UI",sans-serif}
header{background:var(--sup);border-bottom:1px solid var(--rej);padding:28px 16px}
.envoltura{max-width:1120px;margin:0 auto;padding:0 16px}
h1{margin:0 0 6px;font-size:26px}h2{font-size:21px;margin:0 0 4px}h4{margin:0 0 6px;font-size:14px}
.sub{color:var(--t2);margin:0}.tiles{display:flex;flex-wrap:wrap;gap:12px;margin-top:18px}
.tile{background:var(--plano);border:1px solid var(--rej);border-radius:10px;padding:10px 16px;min-width:140px}
.tile b{display:block;font-size:26px}.tile span{color:var(--t2);font-size:13px}
nav{position:sticky;top:0;z-index:2;background:var(--sup);border-bottom:1px solid var(--rej);overflow-x:auto;white-space:nowrap}
nav a{display:inline-block;padding:10px 12px;color:var(--t2);text-decoration:none;font-size:13.5px}nav a:hover{color:var(--azul)}
section{background:var(--sup);border:1px solid var(--rej);border-radius:12px;padding:22px;margin:22px 0}
.metodo{color:var(--t3);font-size:13px;margin:0 0 10px}.pregunta{font-style:italic;color:var(--t2);border-left:3px solid var(--azul);padding-left:10px}
figure{margin:18px 0}figure img{max-width:100%;height:auto;display:block;border-radius:6px}
figcaption{color:var(--t2);font-size:13px;margin-top:6px}
.tabla{border-collapse:collapse;width:100%;font-size:13px;margin:12px 0 4px;font-variant-numeric:tabular-nums}
.tabla th{text-align:left;color:var(--t2);font-weight:600;border-bottom:1px solid var(--t3);padding:6px 8px;background:var(--plano)}
.tabla td{border-bottom:1px solid var(--rej);padding:5px 8px;vertical-align:top}
.tabla-wrap{overflow-x:auto}.nota{color:var(--t3);font-size:12.5px;margin:2px 0 12px}
.lectura{background:var(--azul-c);border-radius:10px;padding:14px 18px;margin-top:14px}.lectura span{color:var(--t3);font-weight:400}
.lectura ul{margin:4px 0 0;padding-left:20px}.lectura li{margin:4px 0}
.aviso{background:#fff4e5;border-radius:8px;padding:10px 14px;color:#7a4a00}
footer{color:var(--t3);font-size:12.5px;padding:10px 16px 40px}
@media print{nav{display:none}section{break-inside:avoid-page;border:none}}
"""
cuerpo = cuerpo.replace('<table class="dataframe tabla">', '<div class="tabla-wrap"><table class="tabla">').replace("</table>", "</table></div>")
fecha = datetime.datetime.now().strftime("%d/%m/%Y %H:%M")
HTML = f"""<!doctype html><html lang="es"><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1">
<title>Informe de resultados SEGUILE</title><style>{CSS}</style></head><body>
<header><div class="envoltura"><h1>Gobernanza de la IA en tres niveles: informe de resultados</h1>
<p class="sub">Generado el {fecha} · k = {etiquetas.topic.nunique() if etiquetas is not None else '—'} · pasada {'1 (decisiones)' if EJECUTAR_CV_K else '2 (resultados)'}</p>
<div class="tiles">{''.join(f'<div class="tile"><b>{v}</b><span>{k}</span></div>' for k, v in tiles)}</div></div></header>
<nav><div class="envoltura">{toc}</div></nav><main class="envoltura">{cuerpo}</main>
<footer class="envoltura">Las lecturas automáticas aplican reglas fijas de magnitud del efecto (ε²: 0,01 / 0,08 / 0,26; r biserial y V de Cramér:
0,1 / 0,3 / 0,5) y significación con corrección de Benjamini-Hochberg. Son un borrador: verifica cada afirmación con las tablas antes de usarla.</footer>
</body></html>"""
destino = DIR_SALIDA / "informe_resultados.html"
destino.write_text(HTML, encoding="utf-8")
print(f"\nInforme guardado: {destino}  ({destino.stat().st_size / 1e6:.1f} MB, {len(FIGURAS)} figuras)")
if EN_COLAB:
    from IPython.display import HTML as _H, display
    display(_H(f'<p>Ábrelo desde tu Drive (<b>SEGUILE_resultados/informe_resultados.html</b>) o descárgalo:</p>'))
    try:
        from google.colab import files
        files.download(str(destino))
    except Exception:
        pass
```
